## **Before Object-Oriented Programming**

Before understanding **Object-Oriented Programming**, it helps to understand the problem it was trying to address.

Early computer programs were relatively small. A programmer could write a sequence of instructions describing exactly what the computer should do.

Conceptually:

```text
INPUT
  ↓
INSTRUCTION
  ↓
INSTRUCTION
  ↓
INSTRUCTION
  ↓
OUTPUT
```

This style of programming is closely associated with **procedural programming**.

A program is organized primarily around **procedures or functions** that operate on data.

For example:

```python
name = "Alice"
balance = 5000

def deposit(balance, amount):
    return balance + amount

balance = deposit(balance, 1000)
```

Here, the **data** and the **functions operating on that data** are separate.

We have:

```text
DATA
name
balance

       ↓ used by

FUNCTIONS
deposit()
withdraw()
transfer()
```

This approach works well for many problems.

However, as software systems became larger, developers needed better ways to organize increasingly complex combinations of:

* data,
* behavior,
* state,
* relationships,
* responsibilities,
* and interacting components.

This leads us toward one of the central ideas behind OOP:

> **What if a program could be organized around the entities participating in the problem rather than primarily around sequences of procedures?**

## **From Procedures to Objects**

Consider a university enrollment system.

We might have data describing:

```text
Student
Course
Instructor
Enrollment
Department
```

Each entity also has behavior.

A `Student` might:

```text
enroll()
drop_course()
view_grades()
```

A `Course` might:

```text
add_student()
remove_student()
calculate_class_average()
```

An `Instructor` might:

```text
assign_grade()
create_activity()
view_class_list()
```

Instead of keeping the data and operations completely separate, we can organize them together.

```{mermaid}
flowchart LR

A["Student Data"] --> S["Student Object"]
B["Student Behavior"] --> S

C["Course Data"] --> T["Course Object"]
D["Course Behavior"] --> T

E["Instructor Data"] --> I["Instructor Object"]
F["Instructor Behavior"] --> I
```

This is the basic direction of **object-oriented thinking**.

An object combines:

```text
STATE + BEHAVIOR
```

or, more concretely:

```text
DATA + OPERATIONS ON THAT DATA
```

## **Where Did OOP Come From?**

Object-oriented programming was not invented all at once.

Its development occurred over decades and was influenced by ideas from:

* simulation,
* interactive computing,
* computer graphics,
* programming languages,
* software engineering,
* biological systems,
* and networked computers.

A simplified historical progression is:

```{mermaid}
timeline
    title Development of Object-Oriented Programming

    1960s : Simula
          : Classes and objects
          : Simulation

    1970s : Smalltalk
          : Message passing
          : Everything as objects
          : Interactive computing

    1980s : C++
          : OOP enters mainstream systems programming

    1990s : Java
          : OOP becomes widely taught and adopted

    1991+ : Python
          : Multi-paradigm programming
          : Practical object-oriented programming

    Today : OOP across software engineering
          : Web systems
          : Data science
          : Machine learning libraries
```

The exact history is richer than this simplified timeline, but two languages are especially important to understanding the origins of OOP:

**Simula** and **Smalltalk**. Historical accounts identify Simula as foundational to classes, objects, inheritance, and dynamic lookup, while Smalltalk developed these ideas into a more thoroughly object-oriented environment. ([Cambridge University Press][1])

## **Simula: Objects Begin with Simulation**

### **The Simulation Problem**

In the 1960s, Norwegian computer scientists **Ole-Johan Dahl** and **Kristen Nygaard** worked on the programming language **Simula**.

As its name suggests, Simula was originally created for **simulation**. ([Cambridge University Press][1])

Imagine simulating a port.

You might need to represent:

```text
Ship
Cargo
Dock
Crane
Worker
Truck
```

Each entity has its own state.

For example:

```text
Ship
----------------
name
capacity
location
cargo
```

And its own behavior:

```text
Ship
----------------
load()
unload()
move()
```

Representing these entities naturally encouraged programmers to think about computational structures that represented **things with both state and behavior**.

## **Simula and the Class**

One of Simula's major contributions was the development of concepts we now associate with:

* classes,
* objects,
* inheritance,
* subtyping,
* and dynamic method lookup. ([Cambridge University Press][1])

The idea of a **class** is particularly important.

Think of a class as a description of a category of objects.

```{mermaid}
classDiagram

class Ship {
    name
    capacity
    location
    load()
    unload()
    move()
}

Ship --> ship1 : instance
Ship --> ship2 : instance
Ship --> ship3 : instance
```

The class describes what a `Ship` has and what a `Ship` can do.

Individual ships are **objects or instances** of that class.

Conceptually:

```text
CLASS
  ↓
Ship
  ↓
 ┌──────────┬──────────┬──────────┐
 ↓          ↓          ↓
Ship A     Ship B     Ship C
```

This distinction between **class** and **object** remains central to many modern OOP languages.

## **Smalltalk: A Different Vision of Computing**

Simula established many foundational mechanisms, but another language strongly shaped the philosophy of object-oriented programming:

**Smalltalk**.

Smalltalk was developed during the 1970s at **Xerox PARC**, associated particularly with the work of **Alan Kay, Dan Ingalls, Adele Goldberg**, and others. ([DOI System][2])

Smalltalk wasn't simply trying to improve simulation.

The researchers were exploring ideas about:

* personal computing,
* interactive programming,
* graphical interfaces,
* learning,
* dynamic systems,
* and communication among computational entities.

Alan Kay's historical account describes Smalltalk as part of a broader attempt to rethink computing rather than merely improve conventional programming. ([DOI System][2])

## **Alan Kay and the Meaning of "Object-Oriented"**

Alan Kay is closely associated with the term **object-oriented programming**.

Interestingly, Kay's original conception differed somewhat from the way OOP is commonly taught today.

Modern introductions frequently emphasize:

```text
Encapsulation
Abstraction
Inheritance
Polymorphism
```

Kay placed especially strong emphasis on **objects communicating through messages**, along with protected local state and late binding. ([Userpage][3])

The conceptual model is therefore not merely:

```text
OBJECT = DATA + METHODS
```

but also:

```text
OBJECT ↔ OBJECT ↔ OBJECT
       messages
```
