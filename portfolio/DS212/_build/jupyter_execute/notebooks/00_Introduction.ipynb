A **programming paradigm** is a general way of conceptualizing and structuring a computer program. Different paradigms encourage programmers to think about problems differently. ([Wikipedia][1])

For example, suppose we want to calculate the areas of several circles.

We could think about the problem as:

* What instructions should the computer execute?
* What reusable procedures or functions should perform the calculation?
* What objects exist in the problem, and what should each object know and do?
* What transformations should convert the input values into output values?
* What result do we want, rather than what exact sequence of instructions should produce it?

These approaches are called **programming paradigms**.

## **Programming Paradigms**

A programming paradigm is a general style or approach to programming. It describes how programmers think about a problem and how they organize instructions, data, and operations to create a solution.

Suppose we have the following scores:

```python
scores = [78, 92, 85, 67, 90]
```

Our task is simple:

> *Determine which scores are passing, where a passing score is at least 75.*

This makes the differences between paradigms easier to see.

### 1. **Imperative Programming**

> **Tell the computer HOW to perform the task, step by step.**

Imperative programming explicitly controls execution and changes program state through statements. ([Wikipedia][1])

```python
scores = [78, 92, 85, 67, 90]

passing_scores = []

for score in scores:
    if score >= 75:
        passing_scores.append(score)

print(passing_scores)
```

Output:

```text
[78, 92, 85, 90]
```

Think about what we told Python:

```text
Create an empty list.
↓
Take one score.
↓
Check whether score >= 75.
↓
If yes, append it.
↓
Repeat.
↓
Print the result.
```

The programmer explicitly describes the sequence of operations.

```{mermaid}
flowchart TD
    A[Create empty passing_scores] --> B[Get next score]
    B --> C{score >= 75?}
    C -->|Yes| D[Append score]
    C -->|No| E[Do nothing]
    D --> F{More scores?}
    E --> F
    F -->|Yes| B
    F -->|No| G[Print result]
```

### 2. **Procedural Programming**

Procedural programming belongs to the broader imperative family. Instead of organizing everything as one sequence of commands, we organize behavior into **procedures or functions** that can call one another. ([Wikipedia][2])

> **Organize the instructions into reusable procedures.**

```python
def is_passing(score):
    return score >= 75


def get_passing_scores(scores):
    passing = []

    for score in scores:
        if is_passing(score):
            passing.append(score)

    return passing


scores = [78, 92, 85, 67, 90]

result = get_passing_scores(scores)

print(result)
```

The computation itself is still imperative, but it has now been divided into meaningful units.

```{mermaid}

flowchart TD
    A[Main Program] --> B[get_passing_scores]
    B --> C[Loop through scores]
    C --> D[is_passing]
    D --> E{score >= 75?}
    E --> B
    B --> F[Return passing scores]
    F --> A

```

### 3. **Object-Oriented Programming**

Now we arrive at the paradigm that this course focuses on.

Object-oriented programming organizes programs around **objects containing data and associated behavior**. ([Wikipedia][3])

Instead of asking only:

> What functions do I need?

we begin asking:

> **What things exist in this problem? What information should they contain? What should they be able to do?**

For our example, one obvious thing is a **Student**.

A student can have:

* a `name`
* a `score`

and can perform behavior such as:

* determine whether they passed.

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passing(self):
        return self.score >= 75


student = Student("Ana", 85)

print(student.name)
print(student.is_passing())
```

Notice the fundamental change.

Procedural:

```python
is_passing(score)
```

Object-oriented:

```python
student.is_passing()
```

The data and the behavior related to that data are brought together.

```{mermaid}

classDiagram
    class Student {
        +name
        +score
        +is_passing()
    }
    
```

We can then create many objects from the same class:

```python
students = [
    Student("Ana", 85),
    Student("Ben", 67),
    Student("Carla", 92)
]

for student in students:
    if student.is_passing():
        print(student.name)
```

Conceptually:

```{mermaid}

classDiagram
    class Student {
        +name
        +score
        +is_passing()
    }

    Student <|-- Ana
    Student <|-- Ben
    Student <|-- Carla
    
```

### 4. **Functional Programming**

Functional programming approaches computation primarily through the **evaluation and composition of functions**, generally emphasizing transformations of values and avoiding unnecessary mutable state. ([Wikipedia][1])

> **Think about transforming data through functions.**

Our imperative version was:

```python
passing_scores = []

for score in scores:
    if score >= 75:
        passing_scores.append(score)
```

A more functional approach could be:

```python
scores = [78, 92, 85, 67, 90]

passing_scores = list(
    filter(lambda score: score >= 75, scores)
)

print(passing_scores)
```

Or, to demonstrate transformation:

```python
scores = [78, 92, 85, 67, 90]

statuses = list(
    map(
        lambda score: "Pass" if score >= 75 else "Fail",
        scores
    )
)

print(statuses)
```

Output:

```text
['Pass', 'Pass', 'Pass', 'Fail', 'Pass']
```

The conceptual model becomes:

```{mermaid}
flowchart LR
    A["[78, 92, 85, 67, 90]"] --> B["filter(score >= 75)"]
    B --> C["[78, 92, 85, 90]"]
```

Rather than focusing on repeatedly changing a `passing_scores` variable, we can think of the program as transforming:

```text
input
  ↓
function
  ↓
output
```

### 5. **Declarative Programming**

Declarative programming changes the question even further.

> **Describe WHAT result you want rather than explicitly describing every step required to obtain it.**

Declarative programming describes properties of the desired result without necessarily specifying the detailed sequence of state changes used to calculate it. ([Wikipedia][1])

SQL provides an intuitive example.

Imagine our scores were stored in a database table:

```text
students
+-------+-------+
| name  | score |
+-------+-------+
| Ana   | 85    |
| Ben   | 67    |
| Carla | 92    |
+-------+-------+
```

We could write:

```sql
SELECT name, score
FROM students
WHERE score >= 75;
```

Notice what we **didn't** specify.

We didn't write:

```text
Start at row 1.
Check score.
If score >= 75 copy it somewhere.
Move to row 2.
Check score.
...
```

Instead, we describe the desired result:

> Give me the students whose score is at least 75.

A subset of SQL queries is a familiar example of declarative programming. ([Wikipedia][4])

```{mermaid}
flowchart LR
    A[(Student Database)] --> B["WHERE score >= 75"]
    B --> C["Desired Result"]
```

## **Comparison of Common Paradigms**

| Paradigm            | Primary question                            | Typical organization         |
| ------------------- | ------------------------------------------- | ---------------------------- |
| **Imperative**      | What steps should happen?                   | Statements and state changes |
| **Procedural**      | What procedures should perform the steps?   | Functions/procedures         |
| **Object-Oriented** | What objects exist and what should they do? | Classes and objects          |
| **Functional**      | What transformations should occur?          | Functions and expressions    |
| **Declarative**     | What result do I want?                      | Rules, queries, expressions  |

The distinction is not absolute. Paradigms can overlap, and languages frequently support several of them. ([Wikipedia][1])

[1]: https://en.wikipedia.org/wiki/Programming_paradigm? "Programming paradigm"
[2]: https://en.wikipedia.org/wiki/Procedural_programming? "Procedural programming"
[3]: https://en.wikipedia.org/wiki/Object-oriented_programming? "Object-oriented programming"
[4]: https://en.wikipedia.org/wiki/Declarative_programming? "Declarative programming"
