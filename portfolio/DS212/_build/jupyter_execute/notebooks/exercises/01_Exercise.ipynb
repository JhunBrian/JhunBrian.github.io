# **Case Study**

## **Thinking in Objects**

**Learning Objective**

In this activity, you will identify a real-world system or situation that can be represented as a collection of **interacting entities**.

You will analyze the system from the perspective of the early ideas that led to Object-Oriented Programming.

> **You are not expected to write Python code or design classes yet.**

The goal is to explore **why thinking in terms of objects can be useful**.

## **Example: Traffic Simulation**

Consider a computer simulation of traffic in a small city.

The simulation might contain entities such as:

* cars,
* buses,
* pedestrians,
* traffic lights, and
* intersections.

Each entity has information associated with it and actions it can perform.

| Entity        | Information it knows       | Things it can do  |
| ------------- | -------------------------- | ----------------- |
| Car           | speed, location, direction | move, stop, turn  |
| Traffic Light | current signal, location   | change signal     |
| Pedestrian    | location, destination      | walk, wait, cross |

These entities must also interact.

```{mermaid}
flowchart LR
    Car -->|"approaches"| Intersection
    Intersection -->|"checks"| TrafficLight
    TrafficLight -->|"signals stop/go"| Car
```

A system like this illustrates why simulation played an important role in the early development of object-oriented ideas: a complex system can be viewed as **multiple entities that maintain information, perform actions, and interact with one another**.

**Your case study must use a different scenario.**

## **Your Task**

### **Part 1 — Choose a Scenario**

Choose a real-world system, process, or environment containing several entities that interact with one another.

Possible areas include transportation, education, business, healthcare, games, science, social systems, agriculture, or everyday life.

Do not simply choose an application such as *“Library System”* and immediately begin listing classes. Instead, briefly describe the **actual situation or process** you want a computer program to represent.

Write approximately **100–150 words** describing your chosen scenario.

### **Part 2 — Identify the Entities**

Identify at least **four important entities** involved in your scenario.

For each entity, determine:

* what information it needs to know; and
* what actions it should be able to perform.

Use the following format:

| Entity   | Information it knows | Things it can do |
| -------- | -------------------- | ---------------- |
| Entity 1 | ...                  | ...              |
| Entity 2 | ...                  | ...              |
| Entity 3 | ...                  | ...              |
| Entity 4 | ...                  | ...              |

At this point, think about the entities themselves rather than Python syntax.

### **Part 3 — Show How They Interact**

Your entities should not exist independently. Identify how they **communicate or interact** to accomplish something in your scenario.

Create a simple Mermaid diagram showing at least **three interactions**.

For example:

```{mermaid}
flowchart LR
    A["Entity A"] -->|"interaction"| B["Entity B"]
    B -->|"interaction"| C["Entity C"]
    C -->|"interaction"| A
```

Replace the example with entities and interactions from your own scenario.

Below your diagram, briefly explain what is happening.

### **Part 4 — The Programming Problem**

Now imagine implementing your scenario **without organizing the program around these entities**.

Suppose the information is stored across many separate variables and collections, while numerous procedures manipulate that information.

As the program becomes larger:

> **What problems might programmers encounter when trying to represent and manage your chosen system?**

Identify and explain at least **three potential difficulties**.
