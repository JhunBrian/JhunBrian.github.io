## **Conditional Statements and Boolean Logic**

Programs often need to **make decisions**.

### **Boolean values**

A Boolean has only two possible values:

```python
True
False
```

Most conditions come from comparisons:

```python
age = 20

print(age >= 18)   # True
print(age == 20)   # True
print(age < 18)    # False
```

Important comparison operators:

| Operator | Meaning               |
| -------- | --------------------- |
| `==`     | equal                 |
| `!=`     | not equal             |
| `<`      | less than             |
| `>`      | greater than          |
| `<=`     | less than or equal    |
| `>=`     | greater than or equal |

Be careful with:

```python
x = 10     # assignment
x == 10    # comparison
```

These mean completely different things.


### **Boolean operators**

Conditions can be combined using:

```python
and
or
not
```

For example:

```python
age = 20
has_id = True

print(age >= 18 and has_id)
```

* **`and`**

Both conditions must be `True`.

```python
score >= 70 and attendance >= 80
```

* **`or`**

At least one condition must be `True`.

```python
is_admin or is_instructor
```

* **`not`**

Reverses a Boolean.

```python
not is_logged_in
```


### **`if`, `elif`, and `else`**

Python uses conditional statements to execute different code depending on a condition.

```python
score = 82

if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Passed")
else:
    print("Failed")
```

Conceptually:

```{mermaid}
flowchart TD
    A{"score >= 90?"}
    A -- Yes --> B["Excellent"]
    A -- No --> C{"score >= 70?"}
    C -- Yes --> D["Passed"]
    C -- No --> E["Failed"]
```

Python checks the conditions **from top to bottom** and executes the first matching branch.

## **`for` and `while` Loops**

Loops repeat instructions.

There are two major kinds you'll need:

```text
for   → repeat over a collection/range
while → repeat while a condition remains true
```


### **`for` loops**

Use `for` when iterating through a sequence.

```python
scores = [85, 90, 78]

for score in scores:
    print(score)
```

Conceptually:

```{mermaid}
flowchart LR
    A["[85, 90, 78]"] --> B["85"]
    A --> C["90"]
    A --> D["78"]

    B --> E["score"]
    C --> E
    D --> E
```

The variable `score` temporarily refers to each element.

Another common pattern uses `range()`:

```python
for number in range(5):
    print(number)
```

Output:

```text
0
1
2
3
4
```

Remember:

```python
range(start, stop, step)
```

The `stop` value is **excluded**.

```python
for number in range(2, 10, 2):
    print(number)
```

produces:

```text
2
4
6
8
```


### **`while` loops**

A `while` loop repeats **while its condition remains `True`**.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```{mermaid}
flowchart TD
    A["count = 1"] --> B{"count <= 5?"}
    B -- True --> C["print(count)"]
    C --> D["count += 1"]
    D --> B
    B -- False --> E["Stop"]
```

The critical concept is that something normally needs to eventually make the condition false.

This is dangerous:

```python
count = 1

while count <= 5:
    print(count)
```

`count` never changes, producing an **infinite loop**.


### **`for` vs. `while`**

Use the loop that best expresses your intention.

```python
for student in students:
    ...
```

means:

> Do something for every student.

Whereas:

```python
while balance > 0:
    ...
```

means:

> Keep doing something while this condition remains true.

A useful rule:

| Situation                                   | Usually use |
| ------------------------------------------- | ----------- |
| Iterate through a list                      | `for`       |
| Iterate through a range                     | `for`       |
| Repeat until something happens              | `while`     |
| Number of iterations depends on a condition | `while`     |


### **`break` and `continue`**

`break` immediately exits a loop.

```python
numbers = [10, 20, -5, 30]

for number in numbers:
    if number < 0:
        break

    print(number)
```

Output:

```text
10
20
```

`continue` skips the **current iteration**.

```python
numbers = [10, -5, 20]

for number in numbers:
    if number < 0:
        continue

    print(number)
```

Output:

```text
10
20
```

So:

```text
break      → leave the loop
continue   → skip to the next iteration
```

## **Functions, Parameters, and Return Values**

A function groups instructions into a **reusable operation**.

Instead of:

```python
score1 = 80
score2 = 90
average = (score1 + score2) / 2
print(average)
```

we can define:

```python
def calculate_average(score1, score2):
    return (score1 + score2) / 2
```

and reuse it:

```python
student_a = calculate_average(80, 90)
student_b = calculate_average(75, 95)
```


### **Function anatomy**

```python
def calculate_average(score1, score2):
    average = (score1 + score2) / 2
    return average
```

The important pieces are:

```text
calculate_average → function name
score1, score2    → parameters
average           → local variable
return average    → result
```

```{mermaid}
flowchart LR
    A["Arguments<br/>80, 90"] --> B["Parameters<br/>score1, score2"]
    B --> C["Function Body<br/>(80 + 90) / 2"]
    C --> D["return"]
    D --> E["85.0"]
```


### **Parameters vs. arguments**

In:

```python
def greet(name):
    print(f"Hello, {name}")
```

`name` is a **parameter**.

When calling:

```python
greet("Maria")
```

`"Maria"` is the **argument**.

Think:

```text
parameter = variable in function definition
argument  = actual value supplied during function call
```


### **`return` is not `print`**

This distinction is extremely important.

* **Printing**

```python
def add(a, b):
    print(a + b)
```

displays something.

* **Returning**

```python
def add(a, b):
    return a + b
```

produces a value that the rest of the program can use:

```python
result = add(10, 20)

final_result = result * 2
```

Whenever possible, functions that **calculate something** should generally return their result rather than only printing it.


### **Functions should have clear responsibilities**

Avoid putting unrelated tasks into one enormous function.

Instead of thinking:

```text
do_everything()
```

think in smaller operations:

```python
calculate_average()
determine_status()
display_result()
```

For example:

```python
def calculate_average(scores):
    return sum(scores) / len(scores)


def determine_status(average):
    if average >= 70:
        return "Passed"

    return "Failed"
```

Then:

```python
scores = [80, 75, 90]

average = calculate_average(scores)
status = determine_status(average)

print(average)
print(status)
```

This idea becomes especially important in OOP because methods are essentially functions associated with objects.

## **Modules and Reusable Code**

Eventually, putting every function into one notebook or `.py` file becomes difficult to maintain.

Python lets us organize related code into **modules**.

A module is essentially a Python file containing reusable Python code.

For example:

```text
project/
│
├── main.py
└── grade_utils.py
```

`grade_utils.py`:

```python
def calculate_average(scores):
    return sum(scores) / len(scores)


def determine_status(average):
    if average >= 70:
        return "Passed"

    return "Failed"
```

Then `main.py` can import those functions:

```python
from grade_utils import calculate_average, determine_status

scores = [80, 90, 85]

average = calculate_average(scores)
status = determine_status(average)

print(average)
print(status)
```

```{mermaid}
flowchart LR
    A["main.py"] -->|"imports"| B["grade_utils.py"]
    B --> C["calculate_average()"]
    B --> D["determine_status()"]
```

The goal is **reuse**.

We write the logic once and import it wherever it is needed.


### **Common Import Styles**

Import the entire module:

```python
import math

print(math.sqrt(25))
```

Import a specific item:

```python
from math import sqrt

print(sqrt(25))
```

Use an alias:

```python
import math as m

print(m.sqrt(25))
```
