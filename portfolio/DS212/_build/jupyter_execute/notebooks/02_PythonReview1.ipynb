## **Python Syntax and Coding Conventions**

Before thinking about **objects**, **classes**, **inheritance**, or **polymorphism**, we need to be comfortable reading Python itself.

Consider:

```python
student_name = "Ana"
units = 21
rate_per_unit = 250.0

total_fee = units * rate_per_unit

print(f"{student_name}'s tuition is ₱{total_fee:.2f}")
```

Even this tiny program contains several fundamental Python ideas.

```{mermaid}
flowchart LR
```

There are three statements here:

```python
name = "Maria"
```

```python
age = 20
```

```python
print(name)
```

Python normally executes them **top to bottom**.

```{mermaid}
flowchart TD
    A["name = 'Maria'"] --> B["age = 20"]
    B --> C["print(name)"]
```

This becomes particularly important in Jupyter because cells can be executed in a different order.

### **Python is case-sensitive**

Python distinguishes uppercase and lowercase letters.

```python
student = "Ana"
Student = "Ben"
STUDENT = "Carlo"

print(student)
print(Student)
print(STUDENT)
```

These are **three different identifiers**.

That means this:

```python
score = 95
print(Score)
```

produces an error because `score` and `Score` are not the same name.

This matters later with classes because conventionally we'll write:

```python
class Student:
    pass
```

but an object might be:

```python
student = Student()
```

`Student` and `student` intentionally mean different things.

### **Indentation is syntax**

In many languages, braces indicate blocks.

Conceptually:

```text
if condition {
    do something
}
```

Python instead uses **indentation**.

```python
temperature = 38

if temperature > 37:
    print("High temperature")
    print("Consider resting")

print("Program finished")
```

Visualize the structure:

```{mermaid}
flowchart TD
    A["if temperature > 37"] --> B["Indented block"]
    B --> C["print('High temperature')"]
    B --> D["print('Consider resting')"]

    A --> E["Outside the block"]
    E --> F["print('Program finished')"]
```

The indentation isn't merely decoration.

It communicates **program structure** to Python.

Wrong:

```python
if temperature > 37:
print("High temperature")
```

Correct:

```python
if temperature > 37:
    print("High temperature")
```

The standard convention is **4 spaces per indentation level**.

### **Comments**

Comments allow programmers to explain code.

```python
# Calculate the student's final score
final_score = 90 * 0.4 + 85 * 0.6
```

Python ignores everything after `#` on that line.

But avoid comments that merely repeat the code.

Poor:

```python
# Set age to 20
age = 20
```

Better:

```python
# Minimum age required by the registration policy
minimum_age = 18
```

The second explains **why** the value exists.

## **Naming variables**

Compare these two programs.

* **Version A**

```python
x = 15
y = 120
z = x * y
```

* **Version B**

```python
quantity = 15
price_per_item = 120
total_cost = quantity * price_per_item
```

That's going to become increasingly important in OOP because we'll eventually have names such as:

```python
student
student_name
student_id
calculate_average
enrollment_status
```

instead of:

```python
x
a
data1
thing
func
```

### **`snake_case`**

Python convention generally uses **snake_case** for variables and functions.

```python
student_name = "Ana"
total_score = 95
number_of_students = 40
```

Rather than:

```python
studentName = "Ana"
TotalScore = 95
numberofstudents = 40
```

For functions:

```python
def calculate_average():
    pass
```

Later, you'll notice that **classes use a different convention**:

```python
class StudentRecord:
    pass
```

So remember:

| Element  | Convention   | Example               |
| -------- | ------------ | --------------------- |
| Variable | `snake_case` | `student_name`        |
| Function | `snake_case` | `calculate_average()` |
| Class    | `PascalCase` | `StudentRecord`       |
| Constant | `UPPER_CASE` | `MAX_STUDENTS`        |

This visual distinction helps us recognize what we're reading before understanding the implementation.

### **Whitespace**

These are technically very different in readability:

```python
total=price*quantity+tax
```

and

```python
total = price * quantity + tax
```

Prefer the second.

Similarly:

```python
x=10
name="Alice"
average=(x+20)/2
```

should normally be:

```python
x = 10
name = "Alice"
average = (x + 20) / 2
```

Readable code matters especially in OOP because programs will eventually contain multiple interacting classes and methods.

## **Multi-line expressions**

Long expressions can be wrapped using parentheses.

Instead of:

```python
final_grade = prelim_grade * 0.30 + midterm_grade * 0.30 + final_exam_grade * 0.40
```

we could write:

```python
final_grade = (
    prelim_grade * 0.30
    + midterm_grade * 0.30
    + final_exam_grade * 0.40
)
```

Same computation.

Much easier to inspect.

## **Python's readability philosophy**

A useful principle throughout this course is:

> Code is read far more often than it is written.

Suppose you encounter this:

```python
a = 80
b = 90
c = (a + b) / 2
```

You have to mentally discover what everything represents.

Compare:

```python
quiz_score = 80
exam_score = 90

average_score = (quiz_score + exam_score) / 2
```

The program almost explains itself.

And that idea connects directly to OOP.

Eventually, instead of scattered values such as:

```python
student_name = "Ana"
student_id = "2026-001"
student_grade = 92
```

we'll learn to model the concept itself:

```python
student = Student(...)
```
