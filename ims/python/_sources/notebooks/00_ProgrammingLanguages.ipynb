Every app on your phone, every spreadsheet formula, and every machine learning model began as a **program** — a precise set of instructions written by a person for a computer to follow.

Computers, however, understand only **binary**: long sequences of 0s and 1s. People understand natural language, which is full of ambiguity. A **programming language** sits between the two: precise enough for a computer to follow, readable enough for a person to write.

```{mermaid}

    P --> T["Translator<br/>compiler / interpreter"]

    M --> C["Computer executes"]
```

> **Analogy.** Think of a programming language as the language you use with a **very literal assistant**. If you tell a person "get me something to drink," they will figure it out. A computer needs: *walk to the refrigerator, open the door, take the bottle on the top shelf, close the door, walk back.* Programming is the skill of giving instructions that leave **no room for guessing**.

## **What Is a Program?**

A **program** is a sequence of instructions that tells a computer how to perform a task. Most programs follow the same basic pattern:

```text
INPUT  →  PROCESSING  →  OUTPUT
```

| Program | Input | Processing | Output |
| ------- | ----- | ---------- | ------ |
| Calculator | two numbers and an operator | arithmetic | the result |
| GWA calculator | grades and units | weighted average | the GWA |
| Weather app | location, sensor data | forecasting model | tomorrow's forecast |
| Spam filter | an email | classification model | "spam" or "not spam" |

**Programming** is the process of designing, writing, testing, and fixing (debugging) these instructions.

## **Levels of Programming Languages**

Programming languages are often described by how **close** they are to the hardware.

```{mermaid}
flowchart TB
    HL["High-level languages<br/>Python, Java, R, JavaScript<br/>closest to human language"]
    AL["Assembly language<br/>MOV, ADD, JMP<br/>human-readable names for machine instructions"]
    ML["Machine language<br/>10110000 01100001<br/>what the processor actually executes"]
    HW["Hardware (CPU)"]
    HL --> AL --> ML --> HW
```

| Level | Example | Easy for people? | Control over hardware | Portable? |
| ----- | ------- | ---------------- | --------------------- | --------- |
| **Machine language** | `10110000 01100001` | very hard | complete | no — specific to one processor family |
| **Assembly language** | `MOV AL, 61h` | hard | very high | no |
| **High-level language** | `total = price * quantity` | easier | limited | yes — runs on many systems |

Here is the same idea — adding two numbers — at different levels:

```text
Machine code (conceptual):   10110000 00000101   00000100 00000011
Assembly:                    MOV AL, 5
                             ADD AL, 3
Python:                      result = 5 + 3
```

> **Analogy.** Machine language is like describing a dish **molecule by molecule**; assembly is listing every **knife cut and burner setting**; a high-level language is a **recipe** a home cook can follow.

### **Generations of programming languages**

| Generation | Description | Examples |
| ---------- | ----------- | -------- |
| 1GL | machine language | binary instructions |
| 2GL | assembly language | x86 assembly, ARM assembly |
| 3GL | general-purpose high-level languages | C, Java, Python |
| 4GL | languages closer to the problem domain; often describe *what* rather than *how* | SQL, R (for statistics), spreadsheet formulas |

## **Translators: Compilers and Interpreters**

Computers cannot run high-level code directly; it must first be **translated** into machine code.

| | **Compiler** | **Interpreter** |
| - | ------------ | --------------- |
| How it works | translates the **whole program** before it runs | translates and runs the program **step by step** |
| Analogy | translating an entire **book** before publishing it | an **interpreter at a meeting** translating sentence by sentence |
| Errors are found | before running (at compile time) | while running, at the line where they occur |
| Speed | usually faster to run | usually slower to run |
| Convenience | edit → compile → run | edit → run immediately; great for experimenting |
| Examples | C, C++, Rust, Go | Python, R, JavaScript (in many settings) |


Many modern languages combine both ideas. As you will see below, Python first compiles code into **bytecode**, which is then run by the Python Virtual Machine.

## **Why Are There So Many Languages?**

Languages are designed for different **purposes**, **audiences**, and **ways of thinking** (called **paradigms** — the topic of a later lesson).

| Language | Created | Common uses | Main paradigm(s) |
| -------- | ------- | ----------- | ---------------- |
| C | 1972 | operating systems, embedded devices | imperative, procedural |
| SQL | 1970s | querying databases | declarative |
| Java | 1995 | enterprise systems, Android apps | object-oriented |
| JavaScript | 1995 | interactive web pages | multi-paradigm |
| R | 1993 | statistics, data analysis | functional, procedural |
| Python | 1991 | data science, automation, web, AI | multi-paradigm |

> **Analogy.** Languages are like **vehicles**. A motorcycle, a jeepney, a truck, and a boat all move things from place to place, but each is best for a particular job. A good programmer chooses the right vehicle — and in data science, Python is the most common choice.

### **Why Python for data science?**

* **Readable** — code often looks like pseudocode, so you can focus on the problem.
* **Interactive** — run code in small pieces with Jupyter notebooks and see results immediately.
* **Huge ecosystem** — NumPy, pandas, Matplotlib, scikit-learn, and thousands more libraries.
* **Free and open source**, with a large community and learning resources.
* **General purpose** — the same language handles data collection, analysis, visualization, and deployment.

## **A Short History of Python**

### **The Beginning: Guido van Rossum and the ABC Language**

Python was created by Dutch programmer **Guido van Rossum** in the late 1980s while working at the **Centrum Wiskunde & Informatica (CWI)** in the Netherlands. Before developing Python, van Rossum had worked on **ABC**, an experimental programming language designed to make programming easier for non-experts.

ABC introduced several ideas that later influenced Python, particularly its emphasis on readability and simplicity. However, ABC had limitations that made it difficult to extend and integrate with other systems. Van Rossum wanted a language that retained ABC's simplicity while being more practical, extensible, and suitable for real-world software development.

During the Christmas holidays of **1989**, van Rossum began working on a new language as a personal project. This project eventually became Python.

### Why Is It Called "Python"?

Despite the name, Python was **not named after the snake**. Van Rossum was a fan of the British comedy television series *Monty Python's Flying Circus*. He wanted a name that was short, distinctive, and somewhat playful, so he chose **Python**.

This playful influence remains visible in Python culture, where examples sometimes use terms such as `spam` and `eggs`, references associated with Monty Python.

### Python's Public Introduction

Van Rossum began implementing Python around 1989, and in **February 1991**, he publicly released an early version, commonly identified as **Python 0.9.0**, through the Usenet newsgroup `alt.sources`.

Even at this early stage, Python already contained several recognizable features, including functions, classes, exception handling, and fundamental data types such as lists and dictionaries. This is important because Python was not initially designed merely as a scripting calculator; many of the concepts associated with modern Python were present from its early development.

Python's design emphasized code that programmers could easily read. Instead of relying heavily on braces or other symbols to delimit blocks, Python used **indentation as part of its syntax**. This contributed to one of Python's most recognizable characteristics: programs that often resemble structured pseudocode.

### Python 1.x: Establishing the Language

**Python 1.0** was officially released in **1994**. During the Python 1.x era, the language continued developing features associated with functional programming, including tools such as `lambda`, `map()`, `filter()`, and `reduce()`.

Python was therefore developing as a **multi-paradigm language**. A programmer was not restricted to a single programming style. Python could support procedural programming, object-oriented programming, and elements of functional programming.

This flexibility eventually became one of Python's major strengths. Rather than forcing programmers to structure every problem in the same way, Python allowed them to choose programming techniques appropriate to the problem.

### Python 2.0: A More Mature Python

A major milestone came with the release of **Python 2.0 in October 2000**. It introduced several important improvements, including **list comprehensions**, an improved garbage-collection system capable of handling reference cycles, and better support for Unicode.

For example, list comprehensions provided a compact way of constructing lists:

```python
squares = [x**2 for x in range(10)]
```

This illustrates an important aspect of Python's evolution: features were frequently designed not simply to shorten programs, but to allow programmers to express computational ideas more clearly.

The Python 2.x series became extremely popular and was widely adopted for scripting, web development, system administration, education, and scientific computing.

However, Python accumulated certain design decisions that were difficult to correct without breaking existing programs. As the language grew, its developers faced a difficult question: **Should Python maintain complete backward compatibility, or should some old design decisions be changed to improve the language?**

The eventual answer led to Python 3.

### Python 3: Breaking Compatibility to Improve the Language

**Python 3.0 was released in December 2008.** Unlike an ordinary update, Python 3 intentionally introduced changes that were **not completely backward-compatible with Python 2**.

One of the simplest examples involved `print`.

Python 2 commonly used:

```python
print "Hello, World!"
```

Python 3 uses:

```python
print("Hello, World!")
```

Here, `print` became a regular function rather than a special statement.

Python 3 also introduced significant changes to areas such as Unicode handling, division, iteration, and the organization of the standard library. For instance:

```python
5 / 2
```

In Python 3, this produces:

```text
2.5
```

whereas traditional Python 2 integer division would produce `2` when both operands were integers.

The transition was initially challenging because a large amount of existing software had been written for Python 2. Consequently, Python 2 and Python 3 existed side-by-side for many years.

Eventually, the Python community increasingly moved toward Python 3. **Python 2.7 officially reached end-of-life on January 1, 2020**, marking the effective end of the Python 2 era.

### Python's Rise in Scientific Computing and Data Science

Although Python was not originally created specifically for **data science**, its simplicity and extensibility made it particularly suitable for scientific computing.

A crucial part of Python's success was its ability to serve as a relatively simple interface while computationally demanding operations could be implemented in highly optimized lower-level languages such as C, C++, and Fortran.

Over time, an extensive scientific ecosystem developed around Python. Libraries and frameworks such as **NumPy**, **SciPy**, **Matplotlib**, **pandas**, **scikit-learn**, **TensorFlow**, and **PyTorch** expanded Python's use in numerical computing, visualization, machine learning, and deep learning.

This produced an important characteristic of modern Python: **Python itself is only part of the reason for its popularity. Its ecosystem is equally important.**

A data scientist, for example, can write relatively simple code such as:

```python
import pandas as pd

data = pd.read_csv("dataset.csv")
print(data.describe())
```

while substantial complexity involved in parsing files, managing data structures, and performing numerical computations is handled by libraries underneath.

### Python and Interactive Computing

Python also became strongly associated with **interactive and exploratory programming**. Tools such as IPython and later **Jupyter Notebook** allowed programmers, scientists, researchers, and students to execute programs incrementally rather than repeatedly running an entire application.

This made Python particularly useful for experimentation:

```python
data = load_data()

cleaned = clean_data(data)

model = train_model(cleaned)

evaluate(model)
```

Each stage can be inspected independently. This style fits scientific and data-oriented workflows in which programmers frequently explore data, test hypotheses, visualize intermediate results, modify parameters, and rerun experiments.

Consequently, Python became prominent not only as a traditional programming language but also as a language for **computational exploration**.

### Modern Python

Modern Python continues to evolve through the **Python Enhancement Proposal (PEP)** process. PEPs provide a structured mechanism through which major language features, standards, and design decisions can be proposed and discussed.

One particularly influential document is **PEP 20, "The Zen of Python,"** which expresses several principles associated with Python's philosophy. These include ideas such as readability, simplicity, explicitness, and avoiding unnecessary complexity.

The philosophy can be observed directly from Python:

```python
import this
```

Modern Python has introduced many additional capabilities, including type hints, asynchronous programming with `async` and `await`, structural pattern matching, improvements to performance, and continued refinements to the standard library and language syntax.

Python therefore continues to balance two goals that have existed since its beginning: **keeping the language approachable while allowing it to support increasingly sophisticated applications.**

## **Major Characteristics of the Python Programming Language**

Python is a **high-level, general-purpose programming language** designed with an emphasis on readability, simplicity, and programmer productivity. Although its syntax is relatively easy to learn, Python supports a wide range of applications, from simple automation scripts to web applications, scientific computing, data science, and artificial intelligence. Several characteristics distinguish Python from many other programming languages.

### 1. High-Level, Readable, and Expressive

Python is considered a **high-level language** because programmers work with abstractions such as variables, functions, objects, lists, and dictionaries without directly managing many low-level details such as memory addresses and manual memory allocation.

One of Python's most recognizable characteristics is its emphasis on **readability**. Python uses indentation to define blocks of code rather than relying primarily on braces `{}`.

For example:

```python
numbers = [10, 20, 30, 40]

for number in numbers:
    if number >= 20:
        print(number)
```

The structure of the program can be understood visually from its indentation. This is intentional: Python was designed so that source code could be relatively close to how programmers naturally describe an algorithm.

Python is also **expressive**, meaning that relatively complex operations can often be represented using a small amount of readable code:

```python
squares = [x**2 for x in range(10)]
```

This combination of readability and expressiveness contributes to Python's popularity in education, rapid prototyping, and exploratory programming.

### 2. Dynamically Typed and Automatically Managed

Python is a **dynamically typed language**. A programmer does not normally declare the data type of a variable before assigning a value to it.

```python
x = 10
x = 3.14
x = "Python"
```

The same variable name `x` can refer to objects of different types during execution. In contrast, many statically typed languages require the variable's type to be specified and checked before execution.

However, this does **not** mean Python has no types. Objects still have specific types:

```python
x = 10

print(type(x))
# <class 'int'>
```

Python is therefore commonly described as **dynamically but strongly typed**. It determines types at runtime, but it generally does not automatically treat incompatible types as interchangeable:

```python
"10" + 5
```

This results in a `TypeError` rather than automatically converting one value.

Python also provides **automatic memory management**. Objects are allocated and reclaimed by the Python runtime, so programmers normally do not manually allocate and free memory as they might in C.

### 3. Interpreted and Interactive

Python is commonly described as an **interpreted language** because Python programs are executed through a Python interpreter rather than typically being distributed as directly compiled native machine-code programs.

For example:

```bash
python program.py
```

In the standard CPython implementation, Python source code is first compiled into an intermediate representation called **bytecode**, which is then executed by the Python Virtual Machine.

Conceptually:

```text
Python Source Code
       ↓
    Bytecode
       ↓
Python Virtual Machine
       ↓
    Execution
```

This execution model also makes Python highly suitable for **interactive programming**. A programmer can execute individual statements and immediately inspect their results:

```python
>>> x = 10
>>> x * 5
50
```

Interactive environments such as **IPython and Jupyter Notebook** extend this idea further. Instead of repeatedly compiling and executing an entire program, programmers can experiment with small portions of code, inspect variables, visualize results, and modify their approach incrementally.

This characteristic is particularly useful in **data science and scientific computing**, where experimentation and exploration are often part of the programming process.

### 4. Multi-Paradigm, General-Purpose, and Extensible

Python does not require programmers to follow only one programming paradigm. It is a **multi-paradigm language**, supporting several approaches to organizing programs.

Procedural programming is possible through sequences of instructions and functions:

```python
def calculate_area(width, height):
    return width * height
```

Object-oriented programming is supported through classes and objects:

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
```

Python also supports many ideas associated with functional programming:

```python
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x**2, numbers))
```

This flexibility allows programmers to choose an approach based on the nature of the problem rather than being restricted to one paradigm.

Python is also **general-purpose**. The core language was not designed exclusively for a single field. Instead, its extensive ecosystem of modules, packages, and libraries has allowed it to be used across many domains.

```text
Python
 ├── Web Development
 ├── Automation and Scripting
 ├── Scientific Computing
 ├── Data Science
 ├── Machine Learning
 ├── Artificial Intelligence
 └── General Software Development
```

### 5. Ease of Use Comes with Performance and Low-Level Trade-offs

One of Python's major advantages—its high level of abstraction—is also responsible for some of its limitations. Compared with lower-level languages such as **C and C++**, Python generally provides programmers with less direct control over memory and hardware and can have slower execution for computation performed directly in Python.

For example, consider a simple loop:

```python
total = 0

for i in range(1_000_000):
    total += i
```

Python can express this operation clearly, but executing a large number of Python-level operations introduces runtime overhead. In a compiled language such as C, similar operations can usually be translated into highly optimized machine instructions and executed much faster.

The difference partly comes from Python's characteristics discussed earlier. Python is **dynamically typed**, so information about objects and operations often needs to be handled at runtime. Python also provides **automatic memory management** and numerous abstractions that make programming convenient but introduce additional overhead.

Python therefore generally gives up some **execution performance and low-level control in exchange for programmer productivity, readability, and convenience**.

This does not mean Python is unsuitable for computationally intensive applications. Instead, Python commonly relies on optimized libraries implemented partly in lower-level languages. For example:

```python
import numpy as np

x = np.arange(1_000_000)
total = np.sum(x)
```

Although the programmer writes Python, much of the numerical computation can occur in highly optimized compiled code underneath libraries such as NumPy.

This leads to an important distinction:

| Python emphasizes           | Lower-level languages such as C emphasize |
| --------------------------- | ----------------------------------------- |
| Programmer productivity     | Execution efficiency                      |
| Readability                 | Fine-grained control                      |
| High-level abstractions     | Hardware-level operations                 |
| Automatic memory management | Greater memory control                    |
| Rapid development           | Predictable low-level performance         |

Consequently, Python is not necessarily intended to replace lower-level languages. In many systems, the two approaches **complement each other**: Python can provide a convenient high-level interface, while languages such as C or C++ handle operations where performance and direct control over system resources are critical.

Python's speed trade-off can be seen directly. The cell below compares a plain Python loop with the same computation done by NumPy (which you will study in Weeks 10–12):

````python
import time

import numpy as np

start = time.perf_counter()
total = 0
for i in range(1_000_000):
    total += i
loop_seconds = time.perf_counter() - start

start = time.perf_counter()
np_total = int(np.arange(1_000_000).sum())
numpy_seconds = time.perf_counter() - start

print(total == np_total)
print(f"Python loop is roughly {loop_seconds / numpy_seconds:.0f}× slower than NumPy on this machine")
````
