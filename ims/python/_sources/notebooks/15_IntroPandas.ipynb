NumPy arrays are fast, but they have no **labels**: column 3 is just "column 3", and every element must share one type. Real datasets have **named columns**, **row labels**, **mixed types** (names, dates, numbers), and **missing values**.

**pandas** adds exactly that. It provides two main data structures:

| Structure | Think of it as | Example |
| --------- | -------------- | ------- |
| **Series** | one **labeled column** | the final exam scores, indexed by student name |
| **DataFrame** | a **table** of Series sharing the same row labels | the whole gradebook |

> **Analogy.** A NumPy array is a **grid of numbers on graph paper**. A pandas DataFrame is a **spreadsheet**: the same grid, but with **column headers**, **row labels**, and columns that can hold different kinds of data.

```{mermaid}
flowchart LR

    S --> D["pandas DataFrame<br/>columns of Series + shared index"]
```

````python
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)

print(pd.__version__)
````

## **Series**

A **Series** is a 1D array of values with an **index** (labels).

````python
finals = pd.Series([85, 72, 94, 66], index=["Ana", "Ben", "Carlo", "Diana"], name="final")
print(finals)
print(finals["Carlo"], finals.iloc[0])     # by label, by position
print(finals.index.tolist(), finals.values, finals.dtype)
````

Series support vectorized operations and masks, just like NumPy:

````python
print(finals + 5)
print(finals[finals >= 75])
print(finals.mean(), finals.idxmax())
````

### **Label alignment**

When two Series are combined, pandas matches values **by label**, not by position. Labels missing on one side give `NaN` (*not a number* — pandas' marker for missing data).

````python
quiz = pd.Series({"Ana": 88, "Ben": 75, "Carlo": 91})
lab = pd.Series({"Carlo": 85, "Ana": 92, "Diana": 70})
print((quiz + lab) / 2)
````

## **DataFrames**

A **DataFrame** is a table. It can be created from a dictionary of columns or a list of row dictionaries:

````python
df = pd.DataFrame({
    "name": ["Ana", "Ben", "Carlo", "Diana", "Ella"],
    "section": ["DS1A", "DS1B", "DS1A", "DS1C", "DS1B"],
    "quiz": [88, 75, 91, 62, 80],
    "final": [85, 72, 94, 66, 90],
})
df
````

````python
rows = [
    {"code": "RICE", "item": "Rice", "price": 15},
    {"code": "ADOBO", "item": "Chicken adobo", "price": 80},
]
pd.DataFrame(rows)
````

### **Anatomy of a DataFrame**

```text
            columns →
          name   section  quiz  final
index  0  Ana    DS1A      88    85     ← a row
  ↓    1  Ben    DS1B      75    72
       2  Carlo  DS1A      91    94
                            ↑
                   each column is a Series with its own dtype
```

````python
print(df.shape)
print(df.columns.tolist())
print(df.index)
print(df.dtypes)
````

## **Essential Functionality**

| Method / attribute | Purpose |
| ------------------ | ------- |
| `head(n)`, `tail(n)` | first / last rows |
| `info()` | columns, dtypes, non-missing counts, memory |
| `describe()` | summary statistics of numeric columns |
| `df["col"]` | one column (a Series) |
| `df[["a", "b"]]` | several columns (a DataFrame) |
| `df["new"] = …` | add or replace a column |
| `drop(columns=…)`, `drop(index=…)` | remove columns / rows (returns a new DataFrame) |
| `rename(columns={…})` | rename |
| `sort_values(by)`, `sort_index()` | sort by values / by index |
| `set_index("col")`, `reset_index()` | choose / restore the row labels |

````python
df.info()
````

````python
df.describe()
````

````python
df["average"] = (df["quiz"] + df["final"]) / 2           # new column, vectorized
df["passed"] = df["average"] >= 75
df.sort_values("average", ascending=False)
````

````python
by_name = df.set_index("name")
print(by_name.loc["Carlo"])
print(by_name.drop(columns=["passed"]).head(2))
````

Most pandas methods **return a new DataFrame** and leave the original unchanged. Assign the result to keep it: `df = df.sort_values("final")`.

## **Data Types in pandas**

| pandas dtype | Holds |
| ------------ | ----- |
| `int64`, `float64` | numbers (`float64` also allows `NaN`) |
| `bool` | `True`/`False` |
| `object` / `str` | text |
| `datetime64` | dates and times |
| `category` | a small set of repeated labels (e.g., section) |

````python
df["section"] = df["section"].astype("category")
print(df.dtypes)
````

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| Series | labeled 1D values; aligns by label |
| DataFrame | table of columns sharing an index |
| index / columns | row labels / column labels |
| `head`, `info`, `describe` | first look at any dataset |
| adding columns | vectorized: `df["avg"] = (df["a"] + df["b"]) / 2` |
| sorting / dropping | `sort_values`, `drop` — return new DataFrames |
