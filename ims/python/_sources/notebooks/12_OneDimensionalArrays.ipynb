A **one-dimensional (1D) array** is a single row of values — like a list of all quiz scores, daily temperatures, or prices. This lesson covers how to **select**, **compute with**, and **apply** 1D arrays.

```python
import numpy as np

scores = np.array([78, 92, 64, 85, 70, 99, 58, 88])
print(scores, scores.shape)
```

## **Indexing and Slicing**

1D arrays are indexed exactly like lists: positions start at 0, negative indexes count from the end, and `start:stop:step` slices.

```python
print(scores[0], scores[-1])
print(scores[2:5])
print(scores[::2])
print(scores[::-1])
```

### **Views, not copies**

A **slice of an array is a view** of the same data — changing it changes the original. (List slices, by contrast, are copies.) Use `.copy()` when you need an independent array.

```python
first_three = scores[:3]
first_three[0] = 100
print(scores[:3])                  # the original changed!

scores[0] = 78                     # restore
safe = scores[:3].copy()
safe[0] = 0
print(scores[:3], safe)
```

> **Analogy.** A view is like a **window** into a room — moving furniture seen through the window moves it in the room. A copy is a **photograph** of the room.

### **Fancy indexing**

Pass a list of positions to pick several elements at once:

```python
print(scores[[0, 3, 5]])
```

## **Vectorized Arithmetic**

Operations apply to **every element** at once — no loop needed.

```python
print(scores + 5)
print(scores * 1.1)
print(scores / 100)
print(scores ** 2)

quiz = np.array([80, 90, 70])
exam = np.array([85, 75, 95])
print(0.4 * quiz + 0.6 * exam)          # element by element
```

Two arrays used together must have **compatible shapes** (for 1D arrays: the same length). A single number is automatically applied to every element (this is **broadcasting**, covered in detail next week).

## **Boolean Masks**

A comparison produces an array of `True`/`False` values called a **mask**. Using a mask as an index **selects** the elements where it is `True`.

```python
passing_mask = scores >= 75
print(passing_mask)
print(scores[passing_mask])               # only passing scores
print(scores[scores < 75])                # failing scores

print(passing_mask.sum(), "passed")       # True counts as 1
print(scores[(scores >= 70) & (scores < 90)])   # use & (and), | (or), ~ (not) with parentheses
```

> **Analogy.** A mask is like a **stencil** placed over the row of numbers — only the numbers under the holes show through.

### **`np.where()`**

`np.where(condition, value_if_true, value_if_false)` builds a new array element by element:

```python
print(np.where(scores >= 75, "Passed", "Failed"))
print(np.where(scores + 5 > 100, 100, scores + 5))     # bonus, capped at 100
```

## **Quick Aggregations**

```python
print(scores.sum(), scores.mean(), scores.min(), scores.max())
print(round(scores.std(), 2), np.median(scores))
print(scores.argmax(), "is the position of the highest score")
```

## **Applications**

### **1. Curving and capping grades**

```python
curved = np.minimum(scores + 8, 100)
print(curved)
print(f"Passing rate: {np.mean(scores >= 75):.0%} → {np.mean(curved >= 75):.0%}")
```

### **2. Normalization**

Data is often rescaled before analysis. **Min–max scaling** maps values to 0–1; **z-scores** measure how many standard deviations a value is from the mean:

$$x' = \frac{x - \min(x)}{\max(x) - \min(x)} \qquad z = \frac{x - \bar{x}}{\sigma}$$

```python
minmax = (scores - scores.min()) / (scores.max() - scores.min())
z = (scores - scores.mean()) / scores.std()
print(minmax.round(2))
print(z.round(2))
```

### **3. Temperature conversion and threshold counting**

```python
celsius = np.array([28.4, 31.2, 33.5, 29.9, 35.1, 30.0, 27.8])
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit.round(1))
print("Hot days (≥ 33 °C):", np.count_nonzero(celsius >= 33))
print("Day of the hottest reading:", celsius.argmax() + 1)
```

### **4. Loop vs. vectorized — the same answer**

```python
# loop version
total = 0
for s in scores:
    if s >= 75:
        total += min(s + 5, 100)
count = sum(1 for s in scores if s >= 75)
print(total / count)

# vectorized version
passing = scores[scores >= 75]
print(np.minimum(passing + 5, 100).mean())
```

:::{admonition} Paradigm lens · functional pipeline
:class: tip

`scores[scores >= 75]` is a **filter**, `np.minimum(... + 5, 100)` is a **map**, and `.mean()` is a **reduce** — the same pipeline from the *Programming Paradigms* lesson, now running on fast compiled code.
:::

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| indexing / slicing | same as lists, but slices are **views** |
| fancy indexing | `a[[0, 3, 5]]` |
| vectorized arithmetic | operations apply to every element |
| boolean mask | `a[a >= 75]`; combine with `&`, `|`, `~` |
| `np.where` | element-wise choice between two values |
| aggregations | `sum`, `mean`, `min`, `max`, `std`, `argmax` |
