# **Seatwork: Objects and Classes**

**Learning Objective**

Define classes with constructors, attributes, and methods, and create and use objects.

## **Laboratory Exercise: Library Book Tracker**

### **Task 1 — Objects you already use**

For each value below, print its **type** and call **one method** on it: `"data science"`, `[3, 1, 2]`, `{"a": 1}`.

Expected output:

```text
str: Data Science
list: [1, 2, 3]
dict: ['a']
```

### **Task 2 — Define a class**

Create a class `Book` with attributes `title`, `author`, `copies` (default 1), and `borrowed` (starting at 0). Add a `__str__` method returning `"<title> by <author> (<available>/<copies> available)"`. Create and print two books.

Expected output:

```text
Think Python by Allen Downey (2/2 available)
Python for Data Analysis by Wes McKinney (1/1 available)
```

### **Task 3 — Methods that change state**

Add methods `available()`, `borrow()` (returns `True` if successful, `False` if no copies are left), and `give_back()`. Borrow *Think Python* three times, printing the result each time, then return one copy and print the book.

Expected output:

```text
True
True
False
Think Python by Allen Downey (1/2 available)
```

### **Task 4 — Many objects**

Create a list of four `Book` objects (your choice of titles, at least one with 3 copies). Use a loop to borrow one copy of **each**, then print only the books that still have copies available.

Expected output:

```text
Think Python by Allen Downey (2/3 available)
Python Data Science Handbook by Jake VanderPlas (1/2 available)
```

### **Task 5 — Class diagram**

Draw the class diagram of your `Book` class (attributes and methods) using Mermaid or on paper.

### **Task 6 — Reflection**

Answer in a Markdown cell:

1. What does `self` refer to inside `borrow()`?
2. How would you write the same tracker **without** a class? Which version is easier to extend, and why?
