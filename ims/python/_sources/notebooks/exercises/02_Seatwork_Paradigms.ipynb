# **Seatwork: Same Problem, Different Paradigms**

**Learning Objective**

Determine the differences among programming paradigms by solving the same problem in structured, procedural, and functional styles.

## **The Problem**

A sari-sari store records the sales of one day as a list of `(item, quantity, unit price)` records:

```text
("Coke 1.5L", 3, 75)   ("Pandesal", 20, 3)   ("Rice 1kg", 2, 55)
("Sardines", 4, 28)    ("Coke 1.5L", 1, 75)  ("Eggs (dozen)", 1, 120)
```

The owner wants the **total revenue from items that earned at least ₱100 in a single record**, with a **5% discount** applied to each of those records.

### **Task 1 — Structured**

Write **pseudocode** using only sequence, selection, and iteration (one main block, with a running total), and draw its **flowchart**.

### **Task 2 — Procedural**

Decompose the problem into **at least three procedures** (for example `record_amount`, `qualifies`, `apply_discount`). Draw a **decomposition tree**, write pseudocode for each procedure, and write the main program that calls them.

### **Task 3 — Functional**

Describe the solution as a **filter → map → reduce** pipeline. Draw it as a left-to-right diagram, label what each stage receives and produces, and write the one-line pipeline in pseudocode, for example:

```text
total ← reduce(add, map(discount, filter(qualifies, map(amount, records))))
```

### **Task 4 — Desk check**

Trace any **one** of your three solutions with the data above, showing each record's amount and whether it qualifies. The correct total is **₱538.65**.

### **Task 5 — Analogy**

Invent **your own analogy** (not one from the lesson) for each of the three paradigms.

## **Reflection**

1. Which version would be easiest to change if the discount became 10%? Which if the rule became "≥ ₱150"?
2. Which version would be easiest to **test** piece by piece? Explain using the idea of pure functions.
3. Which version feels most natural to you right now, and why?

Submit one notebook (or PDF) with your pseudocode, diagrams, trace table, analogies, and reflection.
