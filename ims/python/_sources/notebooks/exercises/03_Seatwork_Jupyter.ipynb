# **Seatwork: Jupyter Notebook Exercises**

**Learning Objective**

Explore Jupyter IDE functionalities and translate algorithmic logic into a Jupyter notebook.

## **Instructions**

Create a **new notebook** named `Surname_Jupyter.ipynb` and complete the tasks below. Use the keyboard shortcuts from the lesson as much as possible.

### **Task 1 — Markdown formatting**

Create a Markdown cell that contains:

* a level-1 heading with your full name and a level-2 heading with your section;
* a short paragraph with one **bold** and one *italic* phrase;
* a bulleted list of three things you hope to learn in DS112;
* a table with two columns (`Subject`, `Units`) and three rows;
* a link to the USTP website;
* the formula for the average in LaTeX: $\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$

### **Task 2 — Your first code cells**

Create code cells that:

1. store your name, age, and program in three variables;
2. print a sentence using an f-string, for example: `Hi, I'm Ana, 18, BS Data Science.`;
3. compute and display how many **days** old you are approximately (`age * 365`).

### **Task 3 — Execution order**

1. Create a cell with `x = 10`, a second with `x = x * 2`, and a third with `print(x)`.
2. Run the second cell **three times**, then the third. Record the output.
3. Choose **Kernel → Restart & Run All** and record the output again.

Explain in a Markdown cell why the two outputs differ.

Expected output:

```text
80
20
```

### **Task 4 — From flowchart to notebook**

Using the **electricity bill** problem from the *Algorithm Making* seatwork (₱11.50/kWh for the first 200 kWh, ₱13.20/kWh above 200, plus ₱50 service charge):

1. write the algorithm in a Markdown cell;
2. translate it into a code cell using a variable `kwh` (set its value directly — do not use `input()` yet);
3. run it for 150, 200, and 280 kWh and confirm the results match your desk check.

Expected output:

```text
150 kWh → ₱1,775.00
200 kWh → ₱2,350.00
280 kWh → ₱3,406.00
```

### **Task 5 — Helpers**

Use `help(len)`, `%timeit sum(range(10_000))`, and `%whos` in separate cells. Describe in one sentence what each one shows.

### **Task 6 — Reflection**

Answer in a Markdown cell:

1. What is the difference between the **notebook** and the **kernel**?
2. Why should you always run *Restart & Run All* before submitting?
3. Which three keyboard shortcuts will you use most?

## **Submission Requirements**

Submit **one Jupyter Notebook** containing all tasks in order. Each task should have:

* the code that solves it;
* the output produced by running the code; and
* the Markdown answers to the questions asked.

Make sure the notebook runs from top to bottom without errors (**Kernel → Restart & Run All**) before submitting.
