## **Laboratory Exercise: GWA Toolkit**

Each subject is a tuple `(code, grade, units)`, where grades follow the university scale (1.00 is highest, 3.00 is the lowest passing grade, 5.00 is failing).

```python
subjects = [("DS112", 1.25, 3), ("MATH101", 1.75, 3), ("GE1", 1.50, 3), ("NSTP1", 1.00, 3), ("PE1", 2.00, 2)]
```

### **Task 6 — Pure vs. impure**

The function below has a **side effect**. Rewrite it as a **pure** function `curved(grades, bonus)` that returns a new list and leaves the original unchanged. Show that `original` is unchanged after calling your function.

```python
def curve(grades, bonus):
    for i in range(len(grades)):
        grades[i] = max(1.00, grades[i] - bonus)
    return grades
```

Expected output:

```text
[1.0, 1.25, 1.5]
[1.25, 1.75, 2.0]
```
