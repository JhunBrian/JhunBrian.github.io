## **Laboratory Exercise: Weather Stations** (`data/weather.csv`, synthetic)

### **Task 1 — Derived columns**

Read the file with `date` parsed as dates. Add `temp_f` (Fahrenheit, 1 decimal) and `rainy` (`True` if rain ≥ 1 mm). Print the number of rainy days per station.

Expected output:

```text
station
Cagayan de Oro    22
Iligan            30
Malaybalay        32
Name: rainy, dtype: int64
```

### **Task 2 — Binning**

Use `pd.cut` to label rain as `"none"` (0), `"light"` (> 0 to 10), `"moderate"` (> 10 to 30), and `"heavy"` (> 30), and print the counts (in that order).

Expected output:

```text
rain_level
none        83
light       80
moderate    18
heavy        2
Name: count, dtype: int64
```

### **Task 3 — Rolling and cumulative**

For **Malaybalay**, compute the 7-day rolling mean temperature and the cumulative rainfall. Print the last rolling value (2 decimals) and the total rainfall.

Expected output:

```text
22.56 200.2
```

### **Task 4 — Correlation**

Print the correlation matrix of `temp_c`, `rain_mm`, and `humidity` (2 decimals).

Expected output:

```text
          temp_c  rain_mm  humidity
temp_c      1.00     0.03      0.04
rain_mm     0.03     1.00     -0.01
humidity    0.04    -0.01      1.00
```

### **Task 5 — Charts**

Create four charts, each with a title and axis labels:

1. a **line** chart of daily temperature for all three stations (one line per station — hint: `pivot`);
2. a **bar** chart of total rainfall per station;
3. a **histogram** of humidity;
4. a **scatter** plot of humidity vs. rainfall.
