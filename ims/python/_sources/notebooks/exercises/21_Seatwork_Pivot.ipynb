# **Seatwork: Reshaping and Pivot Tables**

**Learning Objective**

Reshape data and implement pivot functionality with pandas.

## **Tasks**

### **Task 1 — Pivot**

Read `data/weather.csv` (parse dates) and pivot it so each **station** is a column of `rain_mm`. Print the shape and the total rain per station from the wide table.

Expected output:

```text
(61, 3)
station
Cagayan de Oro    196.6
Iligan            240.0
Malaybalay        200.2
dtype: float64
```

### **Task 2 — Melt**

Melt `data/grades.csv` into long format with columns `name`, `section`, `assessment`, `score`, and print the number of rows and the mean score per assessment (2 decimals).

Expected output:

```text
144
assessment
final      77.60
lab        84.19
midterm    75.69
quiz       79.74
Name: score, dtype: float64
```

### **Task 3 — Pivot table**

Create a pivot table of the **mean score** by `section` (rows) and `assessment` (columns) from the long table, rounded to 1 decimal, with row and column averages (`margins=True`).

Expected output:

```text
assessment  final   lab  midterm  quiz   All
section                                     
DS1A         76.3  82.8     74.8  77.2  77.8
DS1B         79.2  84.6     80.1  80.5  81.0
DS1C         74.1  88.2     74.4  86.5  80.8
DS1D         79.3  83.4     75.6  78.5  79.3
All          77.6  84.2     75.7  79.7  79.3
```

### **Task 4 — Crosstab**

Add `passed` (final ≥ 75) to the gradebook and print a crosstab of section vs. passed with proportions per section (2 decimals).

Expected output:

```text
passed   False  True 
section              
DS1A       0.5    0.5
DS1B       0.4    0.6
DS1C       0.5    0.5
DS1D       0.4    0.6
```

### **Task 5 — Sales pivot**

Merge canteen sales with the menu, add `revenue`, and build a pivot table of total revenue by `category` (rows) and `branch` (columns) with a `Total` margin.
