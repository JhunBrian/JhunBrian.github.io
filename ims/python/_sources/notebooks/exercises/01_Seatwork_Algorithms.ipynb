# **Seatwork: Algorithm Making**

**Learning Objective**

Create algorithms, pseudocode, and flowcharts for programming problems, and verify them by desk checking.

## **Instructions**

For each problem: (a) write the algorithm in numbered steps, (b) write **pseudocode** using the course conventions (`INPUT`, `SET`, `IF … THEN … ELSE … ENDIF`, `WHILE … ENDWHILE`, `FOR … ENDFOR`, `OUTPUT`), (c) draw a **flowchart** with the standard symbols, and (d) **desk-check** it with the given test cases in a trace table.

You may draw flowcharts on paper (attach a photo), in draw.io, or with a Mermaid block in a Markdown cell:

````text
```{mermaid}
flowchart TD
    S([Start]) --> I[/Input x/]
    ...
```
````

### **Problem 1 — Electricity bill (sequence and selection)**

A household is charged **₱11.50 per kWh** for the first 200 kWh and **₱13.20 per kWh** for every kWh above 200. A **₱50.00** fixed service charge is added to every bill.

| Test | kWh | Expected bill |
| ---- | --- | ------------- |
| 1 | 150 | ₱1,775.00 |
| 2 | 200 | ₱2,350.00 |
| 3 | 280 | ₱3,406.00 |

### **Problem 2 — Leap year (nested selection)**

A year is a leap year if it is divisible by 4, **except** years divisible by 100, **unless** they are also divisible by 400.

| Test | Year | Expected |
| ---- | ---- | -------- |
| 1 | 2024 | Leap year |
| 2 | 1900 | Not a leap year |
| 3 | 2000 | Leap year |
| 4 | 2026 | Not a leap year |

### **Problem 3 — Savings goal (iteration with `WHILE`)**

A student saves ₱50 in week 1 and increases the amount by ₱10 every week (₱60 in week 2, ₱70 in week 3, …). How many weeks are needed to save **at least** a target amount?

| Test | Target | Expected weeks | Expected total |
| ---- | ------ | -------------- | -------------- |
| 1 | ₱500 | 7 | ₱560 |
| 2 | ₱1,000 | 11 | ₱1,100 |

### **Problem 4 — Counting grades (iteration with `FOR` and selection)**

Given a list of scores, count how many are **A** (≥ 90), **B** (80–89), **C** (75–79), and **F** (< 75).

| Test | Scores | Expected |
| ---- | ------ | -------- |
| 1 | 95, 82, 74, 78, 91, 60 | A = 2, B = 1, C = 1, F = 2 |

### **Problem 5 — Your own problem**

Write a problem from your daily life (for example: computing a jeepney fare with a student discount, splitting a restaurant bill, checking a password's length rules). Provide the algorithm, pseudocode, flowchart, and at least **three** test cases that cover every path of your flowchart.

## **Reflection**

2. In Problem 2, does the **order** of the decisions matter? Show what happens if you check divisibility by 4 first and stop there.
3. Why do good test cases include **boundary values** such as exactly 200 kWh?
