# **Seatwork: Files and Exceptions**

**Learning Objective**

Read and write files and handle errors using exceptions.

## **Laboratory Exercise: Processing the Gradebook File**

Use the file `data/grades.csv` (download it from this book's `data` folder and place it in a `data` folder next to your notebook). Some cells are **empty**.

### **Task 1 — Read and inspect**

Using `csv.DictReader`, read the file into a list of dictionaries. Print the number of rows and the column names.

Expected output:

```text
36
['student_id', 'name', 'section', 'quiz', 'lab', 'midterm', 'final']
```

### **Task 2 — Safe conversion**

Write a function `to_float(text)` that returns a `float`, or `None` if the text cannot be converted. Use it to count how many **midterm** values are missing.

Expected output:

```text
2 missing midterms
```

### **Task 3 — Compute with error handling**

For each student, compute `term = 0.2*quiz + 0.3*lab + 0.2*midterm + 0.3*final`. If any value is missing, mark the student `INC` instead. Print how many students are complete and how many are `INC`, then list the `INC` names.

Expected output:

```text
33 complete, 3 INC
['Felix', 'Leo', 'Rafael']
```

### **Task 4 — Write a report file**

Write `report.csv` (in your notebook's folder) with columns `name`, `term`, and `remark` (`Passed` if term ≥ 75, otherwise `Failed`) for the complete students, sorted by term from highest to lowest. Then read the file back and print its **first four lines**.

Expected output:

```text
name,term,remark
Wena,95.85,Passed
Ivy,94.47,Passed
Carlo,92.96,Passed
```

### **Task 5 — Handling a missing file**

Write a function `load_rows(path)` that returns the rows of a CSV file, or an **empty list** and a printed message if the file does not exist. Test it with `"data/grades_2025.csv"`.

Expected output:

```text
File not found: data/grades_2025.csv
0
```

### **Task 6 — Raising an exception**

Write `validate_score(value)` that raises `ValueError("score must be between 0 and 100")` for values outside 0–100 and returns the value otherwise. Call it on `[95, 120, -3, 80]` inside a loop, printing either the valid value or the error message.

Expected output:

```text
95
120: score must be between 0 and 100
-3: score must be between 0 and 100
80
```

### **Task 7 — Reflection**

Answer in a Markdown cell:

1. Why is `with open(...)` safer than `open()` followed by `close()`?
2. What would happen in Task 3 without exception handling?
3. Why should you catch `ValueError` specifically instead of using a bare `except:`?
