**Learning Objective**

Apply split–apply–combine aggregation and transformation to datasets.

## **Laboratory Exercise: Canteen Sales** (`data/canteen_sales.csv`, `data/menu.csv`, synthetic)

### **Task 1 — Units per branch**

Read both files (parse `date`). Print the total units (`qty`) per branch and the number of order lines per branch.

Expected output:

```text
         sum  count
branch             
Annex   1816    896
Main    3088   1554
```

### **Task 2 — Named aggregation**

For each `item_code`, compute `units` (sum of qty), `orders` (number of rows), and `avg_qty` (mean qty, 2 decimals), sorted by units descending.

Expected output:

```text
           units  orders  avg_qty
item_code                        
RICE        1170     592     1.98
ADOBO        687     351     1.96
COFFEE       639     307     2.08
LUMPIA       587     302     1.94
SINIG        519     257     2.02
PANCIT       488     243     2.01
BUKO         442     216     2.05
HALO         372     182     2.04
```

### **Task 3 — Two keys and unstack**

Compute total units by `branch` and `item_code`, then `unstack` the branch so each branch is a column.

Expected output:

```text
branch     Annex  Main
item_code             
ADOBO        249   438
BUKO         158   284
COFFEE       255   384
HALO         150   222
LUMPIA       206   381
PANCIT       197   291
RICE         430   740
SINIG        171   348
```

### **Task 4 — Grouping by date parts**

Add a `weekday` column and print the total units per weekday in calendar order (Monday → Saturday).

Expected output:

```text
weekday
Monday       764
Tuesday      946
Wednesday    937
Thursday     778
Friday       734
Saturday     745
Name: qty, dtype: int64
```

### **Task 5 — Transform**

Compute daily units per branch (`daily`), then add a column `share` = each day's units divided by that **branch's** total for the month (use `transform`, 3 decimals). Print the first 4 rows.

Expected output:

```text
  branch       date  qty  share
0  Annex 2026-09-01   77  0.042
1  Annex 2026-09-02   92  0.051
2  Annex 2026-09-03   60  0.033
3  Annex 2026-09-04   62  0.034
```
