## **Laboratory Exercise: Four Weeks of Temperatures**

```python
import numpy as np
temps = np.array([28.4, 29.1, 30.2, 31.0, 29.8, 28.7, 27.9,
                  28.8, 30.5, 31.6, 32.2, 31.4, 29.9, 28.5,
                  29.2, 30.0, 30.8, 31.9, 32.5, 30.7, 29.4,
                  28.1, 28.9, 29.5, 30.1, 30.9, 29.6, 28.8])      # °C, 28 days
```

### **Task 1 — Reshape**

Reshape `temps` into a 4 × 7 array `weeks` (one row per week). Print its shape and week 3.

```text
(4, 7)
[29.2 30.  30.8 31.9 32.5 30.7 29.4]
```
