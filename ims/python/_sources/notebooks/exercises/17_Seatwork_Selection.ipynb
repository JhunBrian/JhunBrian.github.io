**Learning Objective**

## **Laboratory Exercise: The DS112 Gradebook** (`data/grades.csv`)

### **Task 1 — loc and iloc**

Read the file with `student_id` as the index. Print (a) the name and final score of `2026-0010` using `loc`, (b) the first three rows and first three columns using `iloc`.

Expected output:

```text
Jose 52.7
             name section  quiz
student_id                     
2026-0001     Ana    DS1A  56.8
2026-0002     Ben    DS1A  91.0
2026-0003   Carlo    DS1D  95.0
```

### **Task 2 — Filtering**

Print the **number** of students in DS1C, the names of students with **both** quiz and final ≥ 85, and the names in DS1A **or** DS1B with a final below 70.

Expected output:

```text
6
['Carlo', 'Ella', 'Gina', 'Ivy', 'Wena']
['Ana', 'Hector', 'Jose', 'Uma', 'Grace']
```

### **Task 3 — Missing values**

Print the number of missing values per column, the names of students with any missing value, and the number of rows after `dropna()`.

Expected output:

```text
{'name': 0, 'section': 0, 'quiz': 0, 'lab': 1, 'midterm': 2, 'final': 0}
['Felix', 'Leo', 'Rafael']
33
```

### **Task 4 — Filling and updating**

Create `clean = g.copy()`. Fill missing midterms with the **median** midterm and missing labs with the **mean** lab (1 decimal). Then cap every final above 95 at 95 using `loc`. Print the number of missing values left and the maximum final.

Expected output:

```text
0 95.0
```

### **Task 5 — Summaries**
