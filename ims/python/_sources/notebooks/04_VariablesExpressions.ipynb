## **Values and Types**

A **value** is a piece of data a program works with. Every value has a **type**, which decides what you can do with it.

| Type | Meaning | Examples |
| ---- | ------- | -------- |
| `int` | whole numbers | `21`, `-5`, `0` |
| `float` | numbers with a decimal point | `3.14`, `75.0`, `-0.5` |
| `str` | text (a *string* of characters) | `"Ana"`, `'DS112'`, `"75"` |
| `bool` | truth values | `True`, `False` |
| `NoneType` | "no value" | `None` |

Use `type()` to ask Python for the type of a value:

````python
print(type(21))
print(type(3.14))
print(type("DS112"))
print(type(True))
print(type("75"))      # quotes make it text, not a number
````

## **Variables and Assignment**

A **variable** is a **name** that refers to a value. The assignment statement `=` **binds** a name to a value.

> **Analogy.** A variable is like a **label on a storage box**. `score = 85` sticks the label *score* on a box containing 85. Assigning again, `score = 90`, moves the label to a box containing 90.

```text
score  ──►  85
score = 90
score  ──►  90
```

Read `=` as **"gets"** or **"is assigned"**, not "equals". `count = count + 1` means *"count gets its old value plus one"* — perfectly sensible in programming, even though it is impossible in algebra.

````python
score = 85
print(score)

score = score + 5          # update using the old value
print(score)

score += 5                 # shorthand for score = score + 5
print(score)

# assign several variables at once
quiz, lab, exam = 88, 92, 79
print(quiz, lab, exam)
````

### **Naming rules and conventions**

**Rules** (breaking them causes an error):

* names contain letters, digits, and underscores only;
* they cannot start with a digit (`2nd_score` ✗);
* they cannot be a **keyword** (`if`, `for`, `class`, `True`, …);
* they are **case-sensitive**: `score`, `Score`, and `SCORE` are three different names.

**Conventions** from **PEP 8**, Python's official style guide (breaking them still works, but looks unprofessional):

| Kind | Convention | Example |
| ---- | ---------- | ------- |
| variables and functions | `snake_case` | `final_grade`, `compute_average()` |
| constants | `UPPER_CASE` | `PASSING_GRADE = 75` |
| classes | `PascalCase` | `StudentRecord` |

Choose **meaningful** names: `price_per_unit` explains itself; `x2` does not.

## **Operators and Expressions**

An **expression** is a combination of values, variables, and operators that Python **evaluates** to a single value.

### **Arithmetic operators**

| Operator | Meaning | Example | Result |
| -------- | ------- | ------- | ------ |
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division (always `float`) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | exponent | `7 ** 2` | `49` |

````python
minutes = 135
print(minutes // 60, "hours and", minutes % 60, "minutes")

print(10 / 4, 10 // 4, 10 % 4, 2 ** 10)

# % is handy for checking even/odd and divisibility
print(17 % 2 == 0, 18 % 2 == 0)
````

### **Order of operations (precedence)**

Python follows the familiar **PEMDAS** order: parentheses → exponents → multiplication/division (`* / // %`) → addition/subtraction, evaluated left to right.

```python
2 + 3 * 4        # 14
(2 + 3) * 4      # 20
-3 ** 2          # -9  (exponent before the minus sign)
(-3) ** 2        # 9
```

When in doubt, add parentheses — they make your intent clear to readers too.

### **Comparison and logical operators**

Comparisons produce `True` or `False`: `==`, `!=`, `<`, `>`, `<=`, `>=`. Logical operators combine them: `and`, `or`, `not`.

````python
final = 82.5
attendance = 0.92

print(final >= 75)
print(final >= 75 and attendance >= 0.80)
print(not final >= 75)
print(70 <= final < 90)         # chained comparison
````

### **Strings in expressions**

`+` joins (concatenates) strings and `*` repeats them:

````python
first, last = "Ana", "Cruz"
print(first + " " + last)
print("=" * 20)

try:
    print("Score: " + 85)
except TypeError as error:
    print("TypeError:", error)
````

Python will not add a string and a number — it is **strongly typed**. Convert first with `str()`, or better, use an **f-string**.

## **Type Conversion**

| Function | Converts to | Example |
| -------- | ----------- | ------- |
| `int()` | integer (drops decimals) | `int("42")` → `42`, `int(3.9)` → `3` |
| `float()` | decimal number | `float("2.5")` → `2.5` |
| `str()` | text | `str(85)` → `"85"` |
| `bool()` | truth value | `bool(0)` → `False`, `bool("hi")` → `True` |
| `round(x, n)` | rounded number | `round(3.14159, 2)` → `3.14` |

````python
text_score = "87"
print(int(text_score) + 3)
print(float("2.75") * 2)
print(round(2 / 3, 3))

try:
    int("eighty-seven")
except ValueError as error:
    print("ValueError:", error)
````

## **Input and Output**

### **Output with `print()` and f-strings**

An **f-string** (formatted string) starts with `f` and places expressions inside `{ }`. A **format specifier** after `:` controls how values look.

````python
name = "Ana"
final = 87.456
tuition = 18250

print(f"{name}'s final grade is {final:.2f}")
print(f"Tuition: ₱{tuition:,.2f}")
print(f"[{name:<8}] [{name:>8}] [{name:^8}]")
print("A", "B", "C", sep=" | ", end=" ←\n")
````

### **Input with `input()`**

`input()` displays a prompt, waits for the user to type, and **always returns a string**:

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))            # convert before doing math
print(f"Hi {name}! Next year you'll be {age + 1}.")
```

```text
Enter your name: Ana
Enter your age: 18
Hi Ana! Next year you'll be 19.
```

> In this book, `input()` examples are shown as text because the pages are built automatically with no one to type. In your own notebooks, run them normally.

## **Statements and Comments**

A **statement** is a complete instruction that Python executes: an assignment, a `print()` call, an `if`, a `for`… An **expression** produces a value; a **statement** does something.

```python
final = 0.3 * quiz + 0.3 * lab + 0.4 * exam     # statement containing an expression
```

**Comments** start with `#` and are ignored by Python. Use them to explain **why**, not to repeat **what**:

```python
# Weights follow the DS112 grading system in the syllabus
final = 0.3 * quiz + 0.3 * lab + 0.4 * exam
```

## **Putting It Together — Translating an Algorithm**

The final-grade algorithm from Week 2, now in Python:

````python
# Input (fixed values here; use input() in your own notebook)
quiz = 88
lab = 92
exam = 71

# Processing
QUIZ_WEIGHT, LAB_WEIGHT, EXAM_WEIGHT = 0.30, 0.30, 0.40
final = QUIZ_WEIGHT * quiz + LAB_WEIGHT * lab + EXAM_WEIGHT * exam
passed = final >= 75

# Output
print(f"Final grade: {final:.2f}")
print(f"Passed: {passed}")
````

## **Writing Readable Code: PEP 8 Essentials**

| Guideline | Do | Don't |
| --------- | -- | ----- |
| spaces around operators | `total = price * qty` | `total=price*qty` |
| one statement per line | two lines | `a = 1; b = 2` |
| indentation | 4 spaces | tabs or 2 spaces |
| line length | ≤ 79 characters | very long lines |
| names | `student_count` | `sc`, `StudentCount` (for variables) |
| constants | `MAX_UNITS = 26` | magic numbers scattered in code |

The **Zen of Python** (PEP 20) summarizes the philosophy — *"Readability counts."*

## **Common Errors**

| Error | Cause | Example |
| ----- | ----- | ------- |
| `SyntaxError` | the code is not valid Python | `final grade = 90` (space in a name) |
| `NameError` | using a name that does not exist | `print(scroe)` |
| `TypeError` | an operation on the wrong type | `"Score: " + 85` |
| `ValueError` | the right type with an unusable value | `int("abc")` |
| `ZeroDivisionError` | dividing by zero | `total / 0` |

Read the **last line** of an error message first — it names the problem.
