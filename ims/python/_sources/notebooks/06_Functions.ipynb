As programs grow, the same steps appear again and again, and one long script becomes hard to read. A **function** is a **named block of code that performs one task**. You define it once and **call** it whenever you need it.

> **Analogy.** A function is like a **rice cooker**: you put in rice and water (**inputs**), press one button (**call**), and get cooked rice (**output**). You don't need to know or repeat the heating steps every time.

You have already used many built-in functions: `print()`, `len()`, `round()`, `type()`. Now you will write your own.

## **Defining and Calling Functions**

```python
def function_name(parameter1, parameter2):
    '''Docstring: what the function does.'''
    # body (indented)
    return result
```

````python
def compute_final(quiz, lab, exam):
    '''Return the weighted final grade.'''
    return 0.30 * quiz + 0.30 * lab + 0.40 * exam


ana = compute_final(88, 92, 71)          # calling the function
ben = compute_final(75, 80, 90)
print(round(ana, 2), round(ben, 2))
````

```{mermaid}
flowchart LR
    A["arguments<br/>88, 92, 71"] --> P["parameters<br/>quiz, lab, exam"]
    P --> B["body<br/>0.30·quiz + 0.30·lab + 0.40·exam"]
    B --> R["return 82.4"]
    R --> V["ana = 82.4"]
```

* **Parameters** are the names in the definition (`quiz`, `lab`, `exam`).
* **Arguments** are the actual values passed in the call (`88`, `92`, `71`).

## **`return` vs. `print`**

`print()` only **displays** a value. `return` **gives the value back** so the rest of the program can use it.

````python
def show_total(a, b):
    print(a + b)          # displays, returns None


def get_total(a, b):
    return a + b          # returns the value


x = show_total(2, 3)
y = get_total(2, 3)
print("x is", x, "| y is", y, "| y * 10 is", y * 10)
````

**Rule of thumb:** functions that **compute** something should `return` it. Leave printing to the part of the program that talks to the user.

A function can return **several values** (as a tuple):

````python
def min_max(values):
    return min(values), max(values)


lowest, highest = min_max([78, 92, 64, 85])
print(lowest, highest)
````

## **Default and Keyword Arguments**

Parameters can have **default values**, and arguments can be passed **by name**:

````python
def jeepney_fare(distance, discounted=False, base=13.00, per_km=1.80):
    fare = base if distance <= 4 else base + (distance - 4) * per_km
    if discounted:
        fare *= 0.80
    return round(fare, 2)


print(jeepney_fare(10))                         # uses defaults
print(jeepney_fare(10, discounted=True))        # keyword argument
print(jeepney_fare(per_km=2.00, distance=10))   # any order when named
````

## **Scope: Local and Global Variables**

Variables created **inside** a function are **local**: they exist only while the function runs. This keeps functions from interfering with each other.

> **Analogy.** What happens inside a **fitting room** stays there — try on as many clothes as you like; the store's shelves (the rest of the program) are unaffected.

````python
PASSING = 75                      # global constant: readable everywhere


def remark(grade):
    status = "Passed" if grade >= PASSING else "Failed"     # local variable
    return status


print(remark(80))

try:
    print(status)
except NameError as error:
    print("NameError:", error)
````

Avoid changing global variables from inside functions. Pass values **in** as parameters and **return** results instead.

## **Docstrings**

A **docstring** is a string right below the `def` line that documents the function. `help()` displays it.

````python
def letter_grade(score):
    '''Convert a numeric score (0–100) to a letter grade.

    Parameters:
        score: the numeric score.

    Returns:
        "A", "B", "C", or "F".
    '''
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 75:
        return "C"
    return "F"


help(letter_grade)
````

## **Designing with Functions: Decomposition**

Large problems become manageable when split into small functions, each with **one responsibility**.

> **Problem.** Compute a student's GWA (general weighted average) from subjects with grades and units, then report the honors standing (using example cut-offs — check the actual policy in the USTP Student Handbook).

```{mermaid}
flowchart TB
    M["report(subjects)"] --> G["gwa(subjects)"]
    M --> H["honors(gwa)"]
    G --> T["total_units(subjects)"]
```

````python
def total_units(subjects):
    return sum(units for _, _, units in subjects)


def gwa(subjects):
    weighted = sum(grade * units for _, grade, units in subjects)
    return weighted / total_units(subjects)


def honors(value):
    if value <= 1.45:
        return "President's lister"
    if value <= 1.75:
        return "Dean's lister"
    return "No honors"


def report(subjects):
    value = gwa(subjects)
    return f"GWA {value:.3f} — {honors(value)}"


subjects = [("DS112", 1.25, 3), ("MATH101", 1.50, 3), ("GE1", 1.75, 3), ("PE1", 1.00, 2)]
print(report(subjects))
````

:::{admonition} Paradigm lens · procedural
:class: tip

Decomposing a problem into named procedures that call one another is **procedural programming**. The main logic (`report`) reads like a summary, each function can be tested alone, and a rule (such as the honors cut-offs) changes in **one** place.
:::

## **Using Modules**

A **module** is a file of ready-made functions. Bring one in with `import`:

````python
import math
import random
import statistics

print(math.sqrt(144), math.ceil(4.2), math.pi)
print(statistics.mean([78, 92, 64]), statistics.median([78, 92, 64]))

rng = random.Random(112)            # a fixed seed makes results reproducible
print(rng.randint(1, 6), rng.choice(["Ana", "Ben", "Carlo"]))
````

## **Functions as Values**

In Python, functions are **values**: they can be stored in variables and **passed to other functions**.

````python
def shout(text):
    return text.upper() + "!"


announce = shout              # no parentheses: the function itself, not a call
print(announce("class dismissed"))

students = [("Ben", 85), ("Ana", 92), ("Carlo", 78)]
print(sorted(students, key=lambda pair: pair[1], reverse=True))     # sort by score
````

### **`lambda` — small anonymous functions**

`lambda parameters: expression` creates a short, unnamed function, useful when passing simple behavior to another function:

```python
lambda score: score >= 75        # same as: def f(score): return score >= 75
```

### **`map()` and `filter()`**

* `map(function, items)` applies a function to **every** item.
* `filter(function, items)` keeps the items for which the function returns `True`.

````python
scores = [78, 92, 64, 85, 70, 99]

passing = list(filter(lambda s: s >= 75, scores))
with_bonus = list(map(lambda s: min(s + 5, 100), passing))
print(passing)
print(with_bonus)
print(sum(with_bonus) / len(with_bonus))
````

### **Pure functions and side effects**

A **pure function** returns a result that depends **only** on its inputs and changes **nothing** outside itself. An **impure** function has **side effects**, such as modifying a list it was given.

````python
def add_bonus_impure(scores):
    for i in range(len(scores)):
        scores[i] += 5            # changes the caller's list!
    return scores


def add_bonus_pure(scores):
    return [s + 5 for s in scores]   # builds a new list


original = [70, 80]
print(add_bonus_pure(original), original)
print(add_bonus_impure(original), original)
````

:::{admonition} Paradigm lens · functional
:class: tip

Passing functions as values, using `map`/`filter`, and preferring **pure functions** that return new data instead of modifying old data are the core ideas of **functional programming**. They make code easier to test and combine — and they are exactly how NumPy and pandas work.
:::

## **Recursion (a brief look)**

A function may call **itself** on a smaller version of the problem. Every recursive function needs a **base case** that stops it.

````python
def factorial(n):
    if n == 0:                 # base case
        return 1
    return n * factorial(n - 1)


print(factorial(5))
````
