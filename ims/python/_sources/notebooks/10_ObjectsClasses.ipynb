**ILO :** Define objects and classes in Python.

You have been using **objects** since your first line of Python. Every value — a number, a string, a list — is an object: a bundle of **data** together with **actions** (called **methods**) that work on that data.

````python
name = "ana cruz"
scores = [78, 92, 64]

print(name.title())          # a str object's method
scores.append(85)            # a list object's method
print(scores)
print(type(name), type(scores), type(3.5), type(print))
````

`name.title()` reads as *"ask the string `name` to give a title-cased version of itself."* The dot connects an **object** with something it **has** or **can do**.

> **Analogy.** An object is like a **smartphone**: it *has* data (contacts, photos, battery level) and it *can do* things (call, take a photo, show the time). You don't need to know how the circuits work — you press buttons (methods).

## **Classes: Blueprints for Objects**

A **class** is a **blueprint** that describes what data and methods its objects will have. Each object created from a class is an **instance**.

> **Analogy.** A class is the **cookie cutter**; the objects are the **cookies**. One cutter, many cookies — each with its own decorations (data).

## **Defining a Class**

````python
class Student:
    def __init__(self, name, student_id):
        self.name = name                 # attributes: the object's data
        self.student_id = student_id
        self.scores = []

    def add_score(self, score):          # methods: the object's actions
        self.scores.append(score)

    def average(self):
        if not self.scores:
            return 0
        return sum(self.scores) / len(self.scores)

    def is_passing(self):
        return self.average() >= 75
````

* `class Student:` starts the blueprint. Class names use **PascalCase**.
* `__init__` is the **constructor** — Python runs it automatically when an object is created, to set up its **attributes**.
* `self` refers to **the particular object** being created or used.
* Methods are functions defined inside the class; their first parameter is always `self`.

## **Creating and Using Objects**

````python
ana = Student("Ana", "2026-001")       # create an object (an instance)
ben = Student("Ben", "2026-002")

for score in (88, 92, 79):
    ana.add_score(score)
ben.add_score(70)

print(ana.name, ana.scores, round(ana.average(), 2), ana.is_passing())
print(ben.name, ben.scores, ben.average(), ben.is_passing())
````

Each object keeps **its own** data: adding a score to Ana does not affect Ben.

```text
ana                                   ben
├── name       → "Ana"                ├── name       → "Ben"
├── student_id → "2026-001"           ├── student_id → "2026-002"
└── scores     → [88, 92, 79]         └── scores     → [70]
```

## **A Readable Printout: `__str__`**

Defining `__str__` controls what `print()` shows for an object:

````python
class Product:
    def __init__(self, code, name, price, stock=0):
        self.code = code
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, quantity):
        self.stock += quantity

    def sell(self, quantity):
        if quantity > self.stock:
            raise ValueError(f"Only {self.stock} {self.name} left.")
        self.stock -= quantity
        return quantity * self.price

    def __str__(self):
        return f"{self.name} (₱{self.price}) — {self.stock} in stock"


halo = Product("HALO", "Halo-halo", 65)
halo.restock(10)
print("Sale: ₱", halo.sell(3))
print(halo)

try:
    halo.sell(20)
except ValueError as error:
    print("Rejected:", error)
````

## **Why Use Classes?**

Compare keeping a product's data in separate variables and functions with keeping it in an object:

| Without a class | With a class |
| --------------- | ------------ |
| `halo_name`, `halo_price`, `halo_stock` — three separate variables per product | one `Product` object per product |
| `sell(stock, price, qty)` — easy to pass the wrong values | `halo.sell(3)` — the object uses its own data |
| rules (e.g., no overselling) repeated wherever stock changes | the rule lives **inside** `sell()` |

Classes become valuable when a program has **many similar things** that each have their own data and behavior.
