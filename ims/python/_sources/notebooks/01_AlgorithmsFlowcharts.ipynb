Before writing a single line of code, a programmer must know **exactly what steps** will solve the problem. That plan is called an **algorithm**, and one of the clearest ways to draw it is a **flowchart**.

> **Analogy.** An algorithm is like the **directions you give a friend** who is visiting campus for the first time: *"Exit the jeepney at the main gate, walk straight past the guardhouse, turn left at the library, the building with the blue roof is ours."* Clear, ordered, and specific — anyone following them arrives at the same place.

```{mermaid}
flowchart LR
    P["Problem"] --> A["Algorithm<br/>(steps in words)"]
    A --> F["Pseudocode / flowchart<br/>(structured plan)"]
    F --> C["Program<br/>(Python code)"]
    C --> T["Test & refine"]
```

## **What Is an Algorithm?**

An **algorithm** is a **finite sequence of well-defined steps** that solves a problem or performs a task.

Donald Knuth described five properties every algorithm must have:

| Property | Meaning | Counter-example |
| -------- | ------- | --------------- |
| **Finiteness** | it must end after a finite number of steps | "keep stirring" with no stopping condition |
| **Definiteness** | every step is precise and unambiguous | "add a little salt" |
| **Input** | it receives zero or more inputs | — |
| **Output** | it produces at least one result | a procedure that computes but never reports anything |
| **Effectiveness** | each step is basic enough to actually be carried out | "find the largest prime number" |

### **Example: Computing a student's final grade**

```text
1. Get the quiz score, lab score, and exam score.
2. Compute final = 0.30 × quiz + 0.30 × lab + 0.40 × exam.
3. If final is at least 75, the remark is "Passed"; otherwise "Failed".
4. Display the final grade and the remark.
```

This algorithm has **input** (three scores), **definite** steps, **output** (grade and remark), and it **finishes** after four steps.

## **Pseudocode**

**Pseudocode** describes an algorithm using structured, English-like statements. It is not a real programming language, so it ignores details like punctuation — but it uses consistent keywords so the logic is clear.

Common conventions used in this course:

| Purpose | Keyword |
| ------- | ------- |
| start / end | `START`, `END` |
| input / output | `INPUT`, `OUTPUT` (or `READ`, `DISPLAY`) |
| assignment | `SET total ← 0` |
| decision | `IF … THEN … ELSE … ENDIF` |
| repetition | `WHILE … DO … ENDWHILE`, `FOR i ← 1 TO n … ENDFOR` |
| procedure | `PROCEDURE name(params) … RETURN … ENDPROCEDURE` |

```text
START
    INPUT quiz, lab, exam
    SET final ← 0.30 × quiz + 0.30 × lab + 0.40 × exam
    IF final ≥ 75 THEN
        SET remark ← "Passed"
    ELSE
        SET remark ← "Failed"
    ENDIF
    OUTPUT final, remark
END
```

Notice the **indentation**: steps inside the `IF` belong to it. Python uses exactly the same idea.

## **Flowcharts**

A **flowchart** is a diagram of an algorithm. Each kind of step has a standard **symbol**, and **arrows** show the order in which steps happen.

| Symbol | Shape | Meaning |
| ------ | ----- | ------- |
| Terminal | rounded rectangle (oval) | start or end |
| Input / Output | parallelogram | read input or display output |
| Process | rectangle | a calculation or assignment |
| Decision | diamond | a yes/no question; two exits |
| Predefined process | rectangle with double sides | a call to a separate, named procedure |
| Connector | small circle | joins parts of a chart across a page |
| Flow line | arrow | the direction of flow |

```{mermaid}
flowchart LR
    T([Terminal]) --> IO[/Input or Output/] --> PR[Process] --> D{Decision?}
    D -->|Yes| PP[[Predefined process]]
    D -->|No| CN((A))
```

### **Flowchart for the final-grade algorithm**

```{mermaid}
flowchart TD
    S([Start]) --> I[/Input quiz, lab, exam/]
    I --> C["final ← 0.30·quiz + 0.30·lab + 0.40·exam"]
    C --> D{final ≥ 75?}
    D -->|Yes| P["remark ← 'Passed'"]
    D -->|No| F["remark ← 'Failed'"]
    P --> O[/Output final, remark/]
    F --> O
    O --> E([End])
```

**Rules for good flowcharts:** one start and one end; flow generally top-to-bottom; every decision has exactly two labeled exits; arrows never cross if it can be avoided; each box contains a single, clear step.

## **The Three Control Structures**

In 1966, Corrado Böhm and Giuseppe Jacopini showed that **any** algorithm can be built from just three building blocks. These are the foundation of **structured programming**, and you will see them in every program you write.

### **1. Sequence — steps one after another**

> **Analogy:** brushing your teeth — wet the brush, apply toothpaste, brush, rinse. Always in that order.

```{mermaid}
flowchart TD
    A[Step 1] --> B[Step 2] --> C[Step 3]
```

### **2. Selection — choosing between paths**

> **Analogy:** at a fork in the road, *if* it is raining you take the covered walkway, *otherwise* you take the shortcut across the field.

```{mermaid}
flowchart TD
    Q{Condition?} -->|True| A[Do this]
    Q -->|False| B[Do that]
    A --> N[Continue]
    B --> N
```

### **3. Iteration (repetition) — repeating steps**

> **Analogy:** checking each of the 40 answer sheets in a pile — *while* there are sheets left, take one, check it, put it on the done pile.

```{mermaid}
flowchart TD
    I[Initialize] --> Q{More items?}
    Q -->|Yes| W[Process one item] --> Q
    Q -->|No| N[Continue]
```

Every structure has **one entry and one exit**, so structures can be **stacked** (sequence) and **nested** (a loop inside a decision, a decision inside a loop) like building blocks.

## **Worked Example 1 — Jeepney Fare (selection)**

> **Problem.** The minimum jeepney fare is ₱13.00 for the first 4 km, plus ₱1.80 for each additional kilometre. Students, seniors, and persons with disabilities get a **20% discount**. Compute the fare.

**Pseudocode**

```text
START
    INPUT distance, is_discounted
    IF distance ≤ 4 THEN
        SET fare ← 13.00
    ELSE
        SET fare ← 13.00 + (distance − 4) × 1.80
    ENDIF
    IF is_discounted THEN
        SET fare ← fare × 0.80
    ENDIF
    OUTPUT fare
END
```

**Flowchart**

```{mermaid}
flowchart TD
    S([Start]) --> I[/Input distance, is_discounted/]
    I --> D1{distance ≤ 4?}
    D1 -->|Yes| A["fare ← 13.00"]
    D1 -->|No| B["fare ← 13.00 + (distance − 4) × 1.80"]
    A --> D2{is_discounted?}
    B --> D2
    D2 -->|Yes| C["fare ← fare × 0.80"]
    D2 -->|No| O[/Output fare/]
    C --> O
    O --> E([End])
```

### **Desk checking (tracing)**

Before trusting an algorithm, **trace** it by hand with sample inputs and a table of variable values:

| Test | distance | is_discounted | fare after step 1 | final fare |
| ---- | -------- | ------------- | ----------------- | ---------- |
| 1 | 3 | No | 13.00 | **13.00** |
| 2 | 10 | No | 13.00 + 6 × 1.80 = 23.80 | **23.80** |
| 3 | 10 | Yes | 23.80 | 23.80 × 0.80 = **19.04** |

Choose test cases that exercise **every path** through the flowchart, including boundaries such as exactly 4 km.

## **Worked Example 2 — Class Average (iteration)**

> **Problem.** Read the scores of *n* students and display the class average and the number of students who passed (score ≥ 75).

```text
START
    INPUT n
    SET total ← 0, passed ← 0, count ← 1
    WHILE count ≤ n DO
        INPUT score
        SET total ← total + score
        IF score ≥ 75 THEN
            SET passed ← passed + 1
        ENDIF
        SET count ← count + 1
    ENDWHILE
    OUTPUT total / n, passed
END
```

```{mermaid}
flowchart TD
    S([Start]) --> I[/Input n/]
    I --> Z["total ← 0, passed ← 0, count ← 1"]
    Z --> L{count ≤ n?}
    L -->|Yes| R[/Input score/]
    R --> T["total ← total + score"]
    T --> P{score ≥ 75?}
    P -->|Yes| PP["passed ← passed + 1"]
    P -->|No| C["count ← count + 1"]
    PP --> C
    C --> L
    L -->|No| O[/Output total ÷ n, passed/]
    O --> E([End])
```

Notice the **selection nested inside the iteration** — the three structures combine to solve a bigger problem.

**Trace** with n = 3 and scores 80, 70, 90:

| count | score | total | passed |
| ----- | ----- | ----- | ------ |
| 1 | 80 | 80 | 1 |
| 2 | 70 | 150 | 1 |
| 3 | 90 | 240 | 2 |
| 4 | — | loop ends | average = 80.0, passed = 2 |

## **Worked Example 3 — Finding the Largest Value**

> **Problem.** Find the highest score in a list.

```text
START
    INPUT scores
    SET highest ← first score
    FOR each score in scores DO
        IF score > highest THEN
            SET highest ← score
        ENDIF
    ENDFOR
    OUTPUT highest
END
```

> **Analogy:** picking the tallest student in a line — you remember the tallest person seen so far, walk down the line, and swap whenever you meet someone taller.

## **From Algorithm to Program (preview)**

In Week 4 you will start writing Python. Notice how closely Python follows the pseudocode for Example 1:

````python
def jeepney_fare(distance, is_discounted):
    if distance <= 4:
        fare = 13.00
    else:
        fare = 13.00 + (distance - 4) * 1.80
    if is_discounted:
        fare = fare * 0.80
    return round(fare, 2)


for distance, discounted in [(3, False), (10, False), (10, True)]:
    print(distance, "km", "discounted" if discounted else "regular", "→ ₱", jeepney_fare(distance, discounted))
````

The outputs match the desk-check table — the program is a faithful translation of the algorithm.

## **Stepwise Refinement**

Large problems are easier when broken down **top-down**: start with a few big steps, then **refine** each step into smaller ones until each is simple enough to code (Wirth, 1971).

```text
Prepare the class report
├── 1. Read all student records
├── 2. Compute each student's final grade
│     ├── 2.1 Get quiz, lab, exam
│     ├── 2.2 Apply the weights
│     └── 2.3 Decide Passed / Failed
└── 3. Display the summary
      ├── 3.1 Class average
      └── 3.2 Number passed and failed
```

## **Common Mistakes**

* **Ambiguous steps** — "compute the grade" instead of the exact formula.
* **Missing initialization** — using `total` before setting it to 0.
* **Infinite loops** — forgetting to update the loop counter.
* **Decisions with one exit** — every diamond needs both a *Yes* and a *No* path.
* **Untested boundaries** — what happens at exactly 4 km, or a score of exactly 75?

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| algorithm | finite, definite steps with input and output |
| pseudocode | structured English description of an algorithm |
| flowchart | diagram of an algorithm using standard symbols |
| sequence, selection, iteration | the three building blocks of every algorithm |
| desk checking | tracing an algorithm by hand to verify it |
| stepwise refinement | break a problem into smaller and smaller steps |
