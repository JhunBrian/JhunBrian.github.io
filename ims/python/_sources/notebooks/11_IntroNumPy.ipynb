Python lists can hold anything, which makes them flexible — but slow and memory-hungry for numerical work. Data science needs to compute with **millions** of numbers at once. **NumPy** (*Numerical Python*) provides the **`ndarray`**: a fast, compact, **n-dimensional array** of numbers. Almost every data science library — pandas, Matplotlib, scikit-learn — is built on top of it.

> **Analogy.** A Python list is like a **shopping bag**: you can throw in anything — apples, a notebook, a phone — but finding and processing items one by one takes time. A NumPy array is like an **egg tray**: every slot holds the same kind of item, in a fixed, orderly layout, so you can handle the whole tray at once.

```{mermaid}

    A["NumPy array<br/>[78, 92, 64, 85]<br/>one data type"] --> N["memory: one compact<br/>block of numbers"]
```

````python
import numpy as np

print(np.__version__)
````

## **Why Arrays? Speed and Memory**

````python
import sys
import time

n = 1_000_000
py_list = list(range(n))
np_array = np.arange(n)

start = time.perf_counter()
doubled_list = [x * 2 for x in py_list]
list_time = time.perf_counter() - start

start = time.perf_counter()
doubled_array = np_array * 2
array_time = time.perf_counter() - start

list_bytes = sys.getsizeof(py_list) + sum(sys.getsizeof(x) for x in py_list)
print(f"List:  {list_time * 1000:7.2f} ms, about {list_bytes / 1e6:5.1f} MB")
print(f"Array: {array_time * 1000:7.2f} ms, about {np_array.nbytes / 1e6:5.1f} MB")
````

NumPy is typically **tens of times faster** and uses **a fraction of the memory**, because:

1. **One data type** — the array knows every element is, say, a 64-bit integer, so it needs no per-item type checks.
2. **Contiguous memory** — the numbers sit side by side, which processors read very efficiently.
3. **Vectorization** — the loop happens inside fast, compiled C code instead of the Python interpreter.

## **Drawbacks and Limits**

Arrays are not better for everything.

| Drawback | Explanation |
| -------- | ----------- |
| **one data type** | mixing numbers and text forces everything into a less useful type (often strings) |
| **fixed size** | "appending" creates a whole new array — slow inside loops |
| **overflow** | fixed-size integers can silently wrap around |
| **less flexible** | no `append`, `insert` in place, or mixed records like a list of dicts |
| **learning curve** | shapes, axes, and broadcasting take practice |

````python
mixed = np.array([1, "two", 3.0])
print(mixed, mixed.dtype)          # everything became text

small = np.array([120, 100], dtype=np.int8)     # int8 holds -128..127
print(small + small)               # wraps around silently!
````

> **Rule of thumb:** use **lists** for small, mixed, or growing collections; use **arrays** for large amounts of **numbers** you will compute with.

## **Creating Arrays**

````python
scores = np.array([78, 92, 64, 85, 70])         # from a list
print(scores)

print(np.zeros(5))                     # five zeros
print(np.ones((2, 3)))                 # 2 rows × 3 columns of ones
print(np.full(4, 75))                  # four copies of 75
print(np.arange(0, 20, 5))             # like range(): 0, 5, 10, 15
print(np.linspace(0, 1, 5))            # 5 evenly spaced numbers from 0 to 1
````

### **Random numbers (reproducibly)**

Create a **random number generator** with a **seed** so that results can be reproduced:

````python
rng = np.random.default_rng(112)

print(rng.integers(1, 7, size=10))           # dice rolls
print(rng.normal(loc=80, scale=8, size=5).round(1))   # simulated scores
print(rng.choice(["Ana", "Ben", "Carlo"], size=2, replace=False))
````

## **Array Attributes**

| Attribute | Meaning |
| --------- | ------- |
| `ndim` | number of dimensions (axes) |
| `shape` | size along each dimension |
| `size` | total number of elements |
| `dtype` | data type of the elements |
| `itemsize` | bytes per element |
| `nbytes` | total bytes |

````python
table = np.array([[78, 92, 64], [85, 70, 99]])

print("ndim:", table.ndim)
print("shape:", table.shape)
print("size:", table.size)
print("dtype:", table.dtype)
print("itemsize:", table.itemsize, "| nbytes:", table.nbytes)
````

## **Data Types**

| dtype | Holds | Range / precision |
| ----- | ----- | ----------------- |
| `int8`, `int16`, `int32`, `int64` | integers | `int8`: −128 to 127 … `int64`: about ±9.2 × 10¹⁸ |
| `uint8` | non-negative integers | 0 to 255 (e.g., image pixels) |
| `float32`, `float64` | decimals | about 7 / 15–16 significant digits |
| `bool` | `True` / `False` | — |
| `str_` (`<U…`) | fixed-length text | — |

NumPy chooses a dtype automatically, or you can specify one. Convert with `astype()`:

````python
a = np.array([1, 2, 3])
b = np.array([1.5, 2, 3])
c = np.array([1, 2, 3], dtype=np.float32)
print(a.dtype, b.dtype, c.dtype)

print(np.array([3.7, 8.2]).astype(int))          # truncates toward zero
print(np.array(["78", "92"]).astype(float) + 1)
print((a + b).dtype)                              # int + float → float (upcasting)
````

:::{admonition} Paradigm lens · vectorization
:class: tip

`np_array * 2` describes **what** to compute for the whole array — no loop, no counter, no changing state. This **vectorized** style is a declarative, functional way of thinking: the array is transformed into a new array, and NumPy decides **how** to loop efficiently.
:::

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| `ndarray` | fast, compact, n-dimensional array of one data type |
| advantages | speed, memory, vectorized operations |
| drawbacks | one dtype, fixed size, overflow, less flexible than lists |
| creation | `array`, `zeros`, `ones`, `full`, `arange`, `linspace`, `default_rng` |
| attributes | `ndim`, `shape`, `size`, `dtype`, `nbytes` |
| dtypes | `int64`, `float64`, `bool`, …; convert with `astype()` |
