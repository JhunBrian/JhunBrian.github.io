Real programs work with **many** values: all the scores in a class, the letters in a name, the coordinates of a place. Python stores ordered collections in **sequences**. This lesson covers the three most common: **lists**, **tuples**, and **strings**.

| Type | Example | Ordered | Mutable (changeable)? | Typical use |
| ---- | ------- | ------- | --------------------- | ----------- |
| `list` | `[78, 92, 64]` | yes | **yes** | a collection that grows or changes |
| `tuple` | `(8.48, 124.65)` | yes | no | a fixed group of related values |
| `str` | `"DS112"` | yes | no | text |

> **Analogy.** A **list** is a **class attendance sheet** — you can add, remove, and correct names. A **tuple** is a **printed ID card** — the details are fixed once issued. A **string** is a **word written in ink** — to change it, you write a new one.

## **Operations Shared by All Sequences**

| Operation | Meaning | Example (`s = "DATA"`) |
| --------- | ------- | ---------------------- |
| `s[i]` | item at position *i* (from 0) | `s[0]` → `'D'` |
| `s[-1]` | last item | `'A'` |
| `s[a:b]` | slice from *a* up to (not including) *b* | `s[1:3]` → `'AT'` |
| `len(s)` | number of items | `4` |
| `x in s` | membership | `'T' in s` → `True` |
| `s + t`, `s * n` | join, repeat | `"DA" + "TA"`, `"ha" * 3` |
| `for x in s` | loop over items | — |

```text
 index:    0    1    2    3
         +----+----+----+----+
         | D  | A  | T  | A  |
         +----+----+----+----+
 negative: -4   -3   -2   -1
```

````python
scores = [78, 92, 64, 85, 70, 99]

print(scores[0], scores[-1])
print(scores[1:4])          # items 1, 2, 3
print(scores[:3], scores[3:])
print(scores[::2])          # every second item
print(scores[::-1])         # reversed
print(len(scores), 85 in scores)
````

## **Lists**

### **Changing a list**

````python
names = ["Ana", "Ben", "Carlo"]

names.append("Diana")          # add to the end
names.insert(1, "Bea")         # add at a position
names[0] = "Anna"              # replace
removed = names.pop()          # remove and return the last item
names.remove("Ben")            # remove by value
print(names, "| removed:", removed)

scores = [78, 92, 64, 85]
scores.sort()                  # sorts in place
print(scores, sum(scores), min(scores), max(scores))
print(sorted(scores, reverse=True), scores)   # sorted() returns a new list
````

| Method | Effect |
| ------ | ------ |
| `append(x)` | add `x` at the end |
| `insert(i, x)` | add `x` at position `i` |
| `extend(other)` | add all items of another sequence |
| `remove(x)` | remove the first `x` |
| `pop(i)` | remove and return item `i` (default: last) |
| `sort()`, `reverse()` | reorder in place |
| `index(x)`, `count(x)` | position of `x`, number of `x` |

### **Aliasing: two names, one list**

Assigning a list to another variable does **not** copy it:

````python
a = [1, 2, 3]
b = a              # b is another name for the SAME list
c = a.copy()       # c is a new list with the same items
a.append(4)
print(a, b, c)
````

> **Analogy.** `b = a` gives the same **Google Doc link** to a classmate — any edit is seen by both. `a.copy()` **downloads a copy** — changes no longer affect each other.

### **List comprehensions**

A **comprehension** builds a new list from an existing sequence in one readable line:

```python
[expression for item in sequence if condition]
```

````python
scores = [78, 92, 64, 85, 70, 99]

curved = [s + 5 for s in scores]
passing = [s for s in scores if s >= 75]
labels = ["P" if s >= 75 else "F" for s in scores]
print(curved, passing, labels, sep="\n")
````

:::{admonition} Paradigm lens · functional / declarative
:class: tip

A comprehension describes **what** the new list contains — *"each score plus five"*, *"the scores that are at least 75"* — rather than the steps of creating an empty list and appending. It is a functional `map`/`filter` written in a declarative style, and it does not modify the original list.
:::

### **Nested lists**

Lists can contain lists — a simple way to store a **table**:

````python
grades = [
    ["Ana", 88, 92],
    ["Ben", 75, 80],
    ["Carlo", 91, 85],
]

for name, quiz, exam in grades:
    print(f"{name:<6} {(quiz + exam) / 2:.1f}")

print(grades[2][1])        # row 2, column 1
````

## **Tuples**

A tuple is like a list that **cannot change**. Use one for values that belong together and should stay fixed: coordinates, RGB colors, a date, a database record.

````python
ustp = (8.4859, 124.6567)          # (latitude, longitude)
lat, lon = ustp                    # unpacking
print(lat, lon)

try:
    ustp[0] = 0
except TypeError as error:
    print("TypeError:", error)

a, b = 1, 2
a, b = b, a                        # swap using tuple packing/unpacking
print(a, b)

single = (5,)                      # a one-item tuple needs a comma
print(type(single))
````

**Why use tuples?** They protect data from accidental change, can be used as **dictionary keys** (next lesson), and signal to readers *"this group of values is fixed."*

## **Strings**

A string is an **immutable sequence of characters**. All sequence operations work on strings, plus many **string methods**.

````python
review = "  The room was CLEAN and the staff were friendly!  "

print(review.strip())                      # remove surrounding spaces
print(review.lower())
print(review.strip().split())              # list of words
print(review.count("the"), review.find("staff"))
print(review.replace("CLEAN", "spotless").strip())
print("-".join(["2026", "08", "10"]))
print("DS112".startswith("DS"), "2026".isdigit())
````

| Method | Effect |
| ------ | ------ |
| `lower()`, `upper()`, `title()` | change case |
| `strip()` | remove leading/trailing whitespace |
| `split(sep)` | break into a list |
| `sep.join(list)` | combine a list into one string |
| `replace(old, new)` | substitute text |
| `find(x)`, `count(x)` | locate / count a substring |
| `startswith()`, `endswith()`, `isdigit()`, `isalpha()` | tests that return `True`/`False` |

Because strings are immutable, methods **return new strings** — the original is unchanged.

### **Example: cleaning and counting words**

````python
text = "Great location. Great staff! The breakfast was great, the room was small."

cleaned = text.lower()
for mark in ".,!":
    cleaned = cleaned.replace(mark, "")

words = cleaned.split()
print(len(words), "words")
print("'great' appears", words.count("great"), "times")
print("longest word:", max(words, key=len))
````

## **Choosing Between Lists, Tuples, and Strings**

| | List | Tuple | String |
| - | ---- | ----- | ------ |
| **Advantages** | flexible: add, remove, sort; many methods | safe from change; usable as dict keys; slightly faster | rich text methods; natural for text |
| **Disadvantages** | can be changed by accident; aliasing surprises | cannot be updated — must create a new one | only characters; every change creates a new string |
| **Use when** | the collection changes | the values are a fixed record | you store text |

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| indexing / slicing | `s[i]`, `s[a:b]`, negative indexes count from the end |
| list | mutable ordered collection; `append`, `sort`, `pop` |
| aliasing | `b = a` shares the list; use `.copy()` |
| comprehension | build a new list declaratively |
| tuple | immutable record; packing and unpacking |
| string | immutable text; methods return new strings |
