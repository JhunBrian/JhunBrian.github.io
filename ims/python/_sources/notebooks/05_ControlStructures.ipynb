## **Selection with `if`**

```{mermaid}
flowchart LR
    Q{condition?} -->|True| A[indented block runs]
    Q -->|False| N[skip the block]
    A --> N2[continue]
    N --> N2
```

````python
temperature = 38.2

if temperature >= 37.5:
    print("Fever detected.")
    print("Please visit the clinic.")

print("Check complete.")       # not indented: always runs
````

### **`if` … `else` and `if` … `elif` … `else`**

Python checks conditions **from top to bottom** and runs only the **first** branch whose condition is true.

````python
final = 83.4

if final >= 90:
    remark = "Excellent"
elif final >= 80:
    remark = "Very Good"
elif final >= 75:
    remark = "Passed"
else:
    remark = "Failed"

print(final, remark)
````

```{mermaid}
flowchart TD
    A{final ≥ 90?} -->|Yes| E[Excellent]
    A -->|No| B{final ≥ 80?}
    B -->|Yes| V[Very Good]
    B -->|No| C{final ≥ 75?}
    C -->|Yes| P[Passed]
    C -->|No| F[Failed]
```

> **Order matters.** If the `final >= 75` test came first, an 83.4 would be labeled "Passed" and never reach "Very Good".

### **Nested conditions and logical operators**

Conditions can be nested, but combining them with `and` / `or` is often clearer:

````python
final, absences = 82, 4

# nested
if final >= 75:
    if absences <= 5:
        print("Passed")
    else:
        print("Failed due to absences")
else:
    print("Failed")

# combined
status = "Passed" if final >= 75 and absences <= 5 else "Failed"     # conditional expression
print(status)
````

### **The jeepney fare in Python**

````python
distance, is_discounted = 10, True

if distance <= 4:
    fare = 13.00
else:
    fare = 13.00 + (distance - 4) * 1.80

if is_discounted:
    fare *= 0.80

print(f"₱{fare:.2f}")
````

## **Iteration with `while`**

A `while` loop repeats **as long as** its condition is true. Use it when you **don't know in advance** how many repetitions are needed.

> **Analogy.** *"While the rice is not yet cooked, keep the heat on and check every few minutes."* You don't know the exact number of checks; you stop when the condition changes.

````python
balance = 1000
month = 0

while balance < 1500:          # how many months to grow 1,000 to 1,500 at 5% per month?
    balance *= 1.05
    month += 1

print(month, "months →", round(balance, 2))
````

Something inside the loop must eventually make the condition **false**; otherwise you get an **infinite loop** (stop it with **Kernel → Interrupt**).

### **Input validation with `while`**

A very common pattern keeps asking until the input is valid:

```python
score = float(input("Enter a score (0-100): "))
while score < 0 or score > 100:
    print("Invalid score. Try again.")
    score = float(input("Enter a score (0-100): "))
```

The cell below simulates the same pattern with a list of "typed" values:

````python
typed_values = ["125", "-4", "88"]          # what a user might type, in order
position = 0

score = float(typed_values[position])
while score < 0 or score > 100:
    print(f"{score} is invalid. Try again.")
    position += 1
    score = float(typed_values[position])

print("Accepted:", score)
````

## **Iteration with `for`**

A `for` loop repeats **once for each item** in a sequence. Use it when you have a collection or a known number of repetitions.

> **Analogy.** *"For each answer sheet in the pile, check it."* The loop ends when the pile runs out.

````python
scores = [78, 92, 64, 85, 70]

for score in scores:
    print(score, "passed" if score >= 75 else "failed")
````

### **`range()`**

`range(start, stop, step)` produces numbers from `start` up to — **but not including** — `stop`.

| Call | Numbers produced |
| ---- | ---------------- |
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 20, 5)` | 0, 5, 10, 15 |
| `range(10, 0, -3)` | 10, 7, 4, 1 |

````python
for week in range(1, 6):
    print("Week", week)

print(list(range(10, 0, -3)))
````

### **Common loop patterns**

| Pattern | Idea |
| ------- | ---- |
| **accumulator** | start a total at 0 and add to it each time |
| **counter** | start at 0 and add 1 when a condition holds |
| **maximum / minimum** | remember the best value seen so far |
| **search** | stop as soon as the item is found |

````python
scores = [78, 92, 64, 85, 70, 99]

total = 0                 # accumulator
passed = 0                # counter
highest = scores[0]       # maximum so far

for score in scores:
    total += score
    if score >= 75:
        passed += 1
    if score > highest:
        highest = score

print(f"Average: {total / len(scores):.2f}, passed: {passed}, highest: {highest}")
````

## **`break`, `continue`, and loop `else`**

* `break` **exits** the loop immediately.
* `continue` **skips** to the next repetition.
* A loop's `else` block runs only if the loop finished **without** `break` — handy for searches.

````python
ids = ["2026-001", "2026-002", "2026-003"]
target = "2026-002"

for student_id in ids:
    if student_id == target:
        print("Found", target)
        break
else:
    print(target, "not found")

for score in [80, -1, 95]:
    if score < 0:
        continue          # skip invalid data
    print("Processing", score)
````

## **Nested Loops**

A loop inside a loop repeats the **inner** loop completely for **each** repetition of the outer loop.

````python
for row in range(1, 6):
    line = ""
    for col in range(1, 6):
        line += f"{row * col:4}"
    print(line)
````

:::{admonition} Paradigm lens · structured programming
:class: tip

`if`, `while`, and `for` are exactly the **selection** and **iteration** blocks of structured programming. Each has **one entry and one exit** (`break` and `continue` are controlled exits within the loop), and indentation makes the nesting visible. Python has **no `goto`** — spaghetti code is not even possible.
:::

## **Choosing the Right Loop**

| Situation | Use |
| --------- | --- |
| go through every item in a list | `for item in items` |
| repeat a fixed number of times | `for i in range(n)` |
| repeat until something happens (valid input, a target is reached) | `while condition` |

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| `if` / `elif` / `else` | choose one branch; checked top to bottom |
| conditional expression | `a if condition else b` |
| `while` | repeat while a condition is true; update something inside |
| `for` + `range()` | repeat for each item / each number |
| accumulator, counter, max | the most common loop patterns |
| `break`, `continue`, `else` | exit early, skip, "not found" handling |
