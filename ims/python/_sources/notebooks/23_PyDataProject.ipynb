## **1. Start with a Good Question**

A good data question is **specific**, **answerable with data**, and **useful to someone**.

## **5. Test Your Functions**

A function that gives a wrong answer silently is worse than one that crashes. Check your functions on **small inputs whose answers you know**:

````python
import numpy as np
import pandas as pd


def classify_rain(mm):
    '''Classify daily rainfall in millimetres.'''
    return np.select([mm >= 30, mm > 0], ["heavy", "light"], default="dry")


def monthly_total(df):
    '''Total rainfall per month from a DataFrame with 'date' and 'rain_mm' columns.'''
    return df.groupby(df["date"].dt.to_period("M"))["rain_mm"].sum()


# --- tests with hand-checked answers ---
assert classify_rain(np.array([0.0])).tolist() == ["dry"]
assert classify_rain(np.array([5.0, 30.0, 45.5])).tolist() == ["light", "heavy", "heavy"]

sample = pd.DataFrame({"date": pd.to_datetime(["2026-01-01", "2026-01-15", "2026-02-01"]),
                       "rain_mm": [10.0, 5.0, 2.5]})
totals = monthly_total(sample)
assert totals.iloc[0] == 15.0 and totals.iloc[1] == 2.5

print("All tests passed.")
````

## **6. Analyze and Visualize**

Use the tools from the course deliberately:

| Question type | Tools |
| ------------- | ----- |
| How much / how many? | `sum`, `mean`, `value_counts` |
| Which group is highest? | `groupby` + `agg`, `sort_values`, `nlargest` |

Every chart needs a **title**, **labeled axes with units**, and a **one-sentence takeaway** written underneath.
