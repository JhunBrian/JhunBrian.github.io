
| Week | Topic | You should be able to… |
| ---- | ----- | ---------------------- |
| 2 | Programming languages; algorithms and flowcharts | compare language levels and translators; write pseudocode; draw and trace flowcharts |
| 3 | Programming paradigms; Jupyter Notebook | explain structured, procedural, and functional programming with examples; use Jupyter cells, Markdown, and the kernel |
| 4 | Variables, expressions, statements | use types, operators, conversion, f-strings; follow PEP 8 |
| 5 | Control structures | write `if`/`elif`/`else`, `while`, and `for` loops; trace their output |
| 6 | Functions | define functions with parameters, defaults, and return values; decompose problems; use `lambda`, `map`, `filter` |
| 7 | Lists, tuples, strings, dictionaries, sets | choose and use the right structure; slicing; comprehensions; counting |
| 8 | Files, exceptions, objects and classes | read/write CSV files; handle exceptions; define a simple class |

## **Self-Check Questions**

### **Concepts**

1. What is the difference between a compiler and an interpreter? Which is Python, and what is bytecode?
2. List Knuth's five properties of an algorithm and give a non-example for two of them.
3. Explain structured, procedural, and functional programming, each with an everyday analogy.
4. What is a side effect? Why do functional programmers avoid them?
5. Why does `input()` need conversion before arithmetic?
6. When would you use a `while` loop instead of a `for` loop?
7. What is the difference between `return` and `print()`?
8. Give one advantage and one disadvantage of lists, tuples, dictionaries, and sets.
9. Why should you use `with` when opening files?

### **Tracing code**

10. What is printed?

    ```python
    total = 0
    for i in range(1, 10, 3):
        if i % 2 == 0:
            continue
        total += i
    print(total)
    ```

11. What is printed, and why?

    ```python
    a = [1, 2, 3]
    b = a
    c = a[:]
    b.append(4)
    print(a, c)
    ```

12. What does `{w: len(w) for w in "to be or not".split() if len(w) > 2}` produce?

### **Writing code**

13. Draw a flowchart and write a Python function `water_bill(cubic_meters)` with a minimum charge of ₱150 for the first 10 m³ and ₱25 per additional m³.
14. Write a function that receives a list of scores and returns a dictionary with keys `"average"`, `"highest"`, `"lowest"`, and `"passed"` (count of scores ≥ 75).
15. Read `data/grades.csv`, skip rows with missing values, and print the name of the student with the highest final exam score.
16. Write a class `Wallet` with an owner and a balance, and methods `deposit()` and `spend()` that rejects spending more than the balance.

## **Study Tips**

* Re-run the code in each lesson and **predict the output before running**.
* Redo the seatworks without looking at your previous answers.
* Practice **tracing** loops with a table of variable values.
* Draw flowcharts by hand; the examination may ask you to read or produce them.

## **References**

Downey, A. B. (2024). *Think Python: How to think like a computer scientist* (3rd ed.). O'Reilly Media. https://allendowney.github.io/ThinkPython/

Knuth, D. E. (1997). *The art of computer programming: Vol. 1. Fundamental algorithms* (3rd ed.). Addison-Wesley.

Liang, Y. D. (2013). *Introduction to programming using Python*. Pearson.

Maulion, M. R. (2022). *DS 112 computer programming* [Instructional material]. University of Science and Technology of Southern Philippines.

Python Software Foundation. (2026). *The Python tutorial*. https://docs.python.org/3/tutorial/
