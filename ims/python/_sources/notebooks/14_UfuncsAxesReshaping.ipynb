````python
import numpy as np

grades = np.array([
    [88, 92, 79, 85],
    [75, 80, 70, 72],
    [91, 85, 88, 94],
    [62, 70, 58, 66],
])
names = np.array(["Ana", "Ben", "Carlo", "Diana"])
````

## **Universal Functions (ufuncs)**

A **ufunc** applies an operation **element by element**, very fast. Arithmetic operators are ufuncs too: `a + b` calls `np.add(a, b)`.

| Category | Examples |
| -------- | -------- |
| arithmetic | `np.add`, `np.subtract`, `np.multiply`, `np.divide`, `np.power`, `np.mod` |
| rounding | `np.round`, `np.floor`, `np.ceil` |
| math | `np.sqrt`, `np.exp`, `np.log`, `np.abs`, `np.sin` |
| comparison | `np.maximum`, `np.minimum`, `np.greater` |

````python
x = np.array([1, 4, 9, 16])
print(np.sqrt(x), np.log10(np.array([10, 100, 1000])))
print(np.round(np.array([2.345, 7.891]), 1), np.ceil(2.1), np.floor(-2.1))
print(np.maximum(grades[:, 0], grades[:, 1]))     # element-wise larger of quiz and lab
print(np.add(x, 1), np.power(x, 0.5))
````

## **Axis Operations and Aggregation**

Aggregation functions accept an `axis` argument:

* `axis=0` → collapse **down** the rows → one result **per column**;
* `axis=1` → collapse **across** the columns → one result **per row**;
* no axis → one result for the **whole** array.

```text
              quiz lab mid final        axis=1 (per student)
        Ana  [ 88  92  79   85 ]   →    mean 86.0
        Ben  [ 75  80  70   72 ]   →    mean 74.25
        ...
              ↓   ↓   ↓    ↓
 axis=0      79  81.75 73.75 79.25   (per assessment)
```

````python
print("per assessment:", grades.mean(axis=0))
print("per student:   ", grades.mean(axis=1))
print("overall:       ", grades.mean())

print("highest per assessment:", grades.max(axis=0))
print("best student per assessment:", names[grades.argmax(axis=0)])
print("std per student:", grades.std(axis=1).round(2))
print("running total of Ana's scores:", np.cumsum(grades[0]))
print("90th percentile of finals:", np.percentile(grades[:, 3], 90))
````

| Function | Result |
| -------- | ------ |
| `sum`, `mean`, `median`, `std`, `var` | totals and statistics |
| `min`, `max`, `argmin`, `argmax` | extremes and their positions |
| `cumsum`, `cumprod` | running totals / products |
| `percentile`, `quantile` | position-based statistics |
| `any`, `all` | logical summaries |

## **Matrix Arithmetic**

`*` multiplies **element by element**. **Matrix multiplication** uses `@` (or `np.dot`): each result is the sum of a row times a column.

$$\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} \begin{bmatrix} 5 \\ 6 \end{bmatrix} = \begin{bmatrix} 1\cdot5 + 2\cdot6 \\ 3\cdot5 + 4\cdot6 \end{bmatrix} = \begin{bmatrix} 17 \\ 39 \end{bmatrix}$$

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A * B)          # element-wise
print(A @ B)          # matrix product
print(A @ np.array([5, 6]))

weights = np.array([0.2, 0.3, 0.2, 0.3])
print("final grades:", (grades @ weights).round(2))     # one matrix product does all students at once
````

Matrix products also solve systems of equations. For example, a canteen sells **x** plates of rice and **y** plates of adobo: 2x + 3y = 270 and x + y = 110.

````python
coefficients = np.array([[2, 3], [1, 1]])
totals = np.array([270, 110])
print(np.linalg.solve(coefficients, totals))
````

## **Reshaping**

`reshape` changes the shape **without changing the data**. The total number of elements must stay the same; `-1` means "figure out this dimension".

````python
daily = np.arange(1, 29)                 # 28 days of readings
weeks = daily.reshape(4, 7)              # 4 weeks × 7 days
print(weeks)
print(daily.reshape(-1, 14).shape)        # 14 columns, rows computed automatically
print(weeks.ravel()[:10])                 # back to 1D (a view when possible)
print(weeks.flatten().shape)              # back to 1D (always a copy)

col = np.array([1, 2, 3])[:, np.newaxis]  # add an axis: shape (3,) → (3, 1)
print(col.shape)
````

## **Transposition**

**Transposing** swaps rows and columns: `a.T`.

````python
print(grades.T)                          # rows are now assessments
print(grades.shape, "→", grades.T.shape)
````

## **Multidimensional Arrays**

Arrays can have **three or more** dimensions. Suppose we store grades for **3 sections × 4 students × 4 assessments**:

```text
shape (3, 4, 4)
axis 0 → section   axis 1 → student   axis 2 → assessment
```

````python
rng = np.random.default_rng(112)
all_sections = rng.integers(55, 100, size=(3, 4, 4))

print(all_sections.shape, all_sections.ndim)
print("Section 2, student 1:", all_sections[1, 0])
print("Average per section:", all_sections.mean(axis=(1, 2)).round(1))
print("Average per assessment, each section:\n", all_sections.mean(axis=1).round(1))
print("Swap axes →", np.transpose(all_sections, (2, 0, 1)).shape)   # assessment, section, student
````

## **1D vs. 2D vs. Multidimensional Arrays**

| | 1D | 2D | Multidimensional (3D+) |
| - | -- | -- | ---------------------- |
| **Picture** | a row of values | a table / matrix | a stack of tables (cube, …) |
| **Examples** | daily temperatures, one exam's scores | gradebook, spreadsheet, grayscale image | sections × students × assessments; color images (height × width × RGB); video |
| **Upsides** | simplest; easy to index and plot | natural for tables; matrix math; row/column statistics | represent several groupings at once; one operation over everything |
| **Drawbacks** | cannot represent relationships between variables | all rows must have the same length | hard to visualize and to read; axis mistakes are easy; harder to label |

> **Tip:** when data has **labels** (student names, column names, dates), pandas (Weeks 13–16) is usually easier than a raw 2D array — and it is built on NumPy.

:::{admonition} Paradigm lens · declarative / vectorized
:class: tip

`grades.mean(axis=0)` says **what** you want — *"the mean of each column"* — and NumPy decides **how** to loop. Code written this way is shorter, faster, and closer to the mathematical description of the problem. This is the same declarative style you will use in pandas.
:::

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| ufunc | fast element-wise function (`np.sqrt`, `np.maximum`, …) |
| axis | `axis=0` per column, `axis=1` per row |
| aggregation | `sum`, `mean`, `std`, `argmax`, `cumsum`, `percentile` |
| matrix arithmetic | `*` element-wise vs. `@` matrix product; `np.linalg.solve` |
| reshape | change shape, same data; `-1` to infer; `ravel`, `flatten`, `newaxis` |
| transpose | `a.T`, `np.transpose(a, axes)` |
