# **NumPy Portfolio**

**DS112 — Computer Programming**

**Assessment:** NumPy portfolio · **Week 12** · **CO3**

**ILO :** Familiarize and implement built-in functions from the NumPy package and apply array operations to a realistic problem.

## **Overview**

Create a **portfolio notebook** that shows what you can do with NumPy on a realistic numeric dataset of your choice. The portfolio should read like a short **technical article**: a question, the data, the analysis, and conclusions.

## **Requirements**

1. **Data.** Use a public numeric dataset (for example from PAGASA, PSA open data, Kaggle, or the UCI repository) **or** a simulated dataset with a clearly described generating process and a fixed seed. Cite the source. Load it into NumPy arrays (`np.loadtxt`, `np.genfromtxt`, or from pandas with `.to_numpy()`).
2. **Questions.** State at least **three** questions you will answer.
3. **Techniques.** Demonstrate, where they make sense for your data:
   * creating and reshaping arrays, dtypes and conversion;
   * indexing, slicing, boolean masks, fancy indexing;
   * insertion/deletion, searching, sorting with `argsort`;
   * concatenation and splitting;
   * broadcasting;
   * ufuncs and aggregation along axes;
   * at least one matrix operation (`@`, `np.linalg`).
4. **Explanations.** Before each code cell, explain *what* you are doing and *why*; after it, interpret the result.
5. **Conclusions.** Answer your questions and state limitations of your data and analysis.
6. **Performance note.** Compare one vectorized computation with an equivalent Python loop and report the speed-up.
