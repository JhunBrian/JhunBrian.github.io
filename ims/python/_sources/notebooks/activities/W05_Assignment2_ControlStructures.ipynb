## **Problem 1 — Water Bill**

A water district charges by consumption (m³) per month:

| Consumption | Rate |
| ----------- | ---- |
| first 10 m³ | minimum charge of ₱180.00 |
| 11–20 m³ | ₱21.50 per m³ above 10 |
| 21–30 m³ | ₱24.00 per m³ above 20 |
| above 30 m³ | ₱27.25 per m³ above 30 |

Senior citizens receive a **5% discount** on bills of ₱500 or less. Read the consumption and senior status, and print an itemized bill. Reject negative consumption with a message.

## **Problem 2 — Number Guessing Game**

The program picks a secret number from 1 to 50 (use `random.Random(seed)` so results can be reproduced). The player has **at most 7 tries**. After each guess print *Too high*, *Too low*, or *Correct!*, and at the end print the number of tries used. Reject guesses outside 1–50 **without** counting them as tries.

## **Problem 3 — Class Statistics**

Read scores until the user types `-1` (a **sentinel**). Reject values outside 0–100. Then print: the number of scores, average, highest, lowest, number passed (≥ 75), and a **histogram** using `*` for the ranges 0–59, 60–74, 75–89, and 90–100.

## **Problem 4 — Patterns with Nested Loops**

Given a height *n* (for example 5), print a **number pyramid** and a **multiplication table** from 1 to *n*:

```text
    1
   121
  12321
 1234321
123454321
```

## **Bonus — Prime Report**

Print all prime numbers between two values *a* and *b* and how many there are, using `for`/`else` or `break`.
