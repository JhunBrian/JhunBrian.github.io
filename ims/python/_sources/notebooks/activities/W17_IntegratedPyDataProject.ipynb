# **Integrated PyData Project**

**DS112 — Computer Programming**

**Assessment:** Final project (released Week 13, presented Week 17) · **Week 13–17** · **CO1, CO2, CO3**

**ILO :** Apply NumPy and pandas to acquire, transform, aggregate, and analyze structured data and present a tested, reproducible PyData programming solution.

## **Overview**

In teams (as assigned by your instructor), build a **complete, reproducible PyData solution** to a question about a real dataset, and present it in Week 17. The project integrates everything in DS112: algorithms, functions, data structures, files and exceptions, NumPy, and pandas. Read the lesson *The Integrated PyData Project* for the workflow and tips.

## **Technical Requirements**

* **Data** — at least **1,000 rows** from a real, citable source (or two joined sources). No personal or sensitive data.
* **Modular code** — a `.py` module with at least **five** documented functions (for loading, cleaning, and computing) imported by the notebook.
* **Robustness** — exception handling for file loading and data conversion.
* **NumPy** — at least one meaningful vectorized computation (normalization, scoring, matrix operation, simulation…).
* **pandas** — cleaning, `groupby` aggregation, at least one `merge`, at least one `pivot_table`.
* **Testing** — a cell (or small test file) with at least **five** `assert` checks verifying your functions on small, hand-checked inputs.
* **Visualization** — at least **five** effective charts.
* **Reproducibility** — relative paths, a `requirements.txt`, a README explaining how to run the project, and *Restart & Run All* without errors.

## **Report Contents**

1. Title, team members, and roles
2. Introduction — the problem, audience, and questions
3. Data — source, description, data dictionary, limitations
4. Methods — cleaning and analysis steps (with a flowchart of the pipeline)
5. Results — answers with charts and tables
6. Conclusions and recommendations
7. Reproducibility instructions
8. References and AI acknowledgement
