# **pandas Portfolio**

**DS112 — Computer Programming**

**Assessment:** pandas portfolio · **Week 16** · **CO3**

**ILO :** Familiarize and implement pandas functions to acquire, transform, aggregate, and analyze structured data.

## **Overview**

Create a **portfolio notebook** that tells a data story using pandas on a **real, public dataset** (for example Philippine open data, PAGASA climate data, PSA statistics, or a Kaggle dataset). The portfolio demonstrates the full workflow: acquire → clean → transform → aggregate → visualize → conclude.

## **Requirements**

1. **Dataset and source** — at least **500 rows** and **5 columns**, properly cited, with a data dictionary (table of column meanings and types).
2. **Questions** — at least **four** questions that matter to a specific audience.
3. **Cleaning** — handle missing values, wrong types, duplicates, and inconsistent text (`.str` methods); document every decision.
4. **Transformation** — derived columns, binning, dates, and at least one method chain.
5. **Aggregation** — `groupby` with named aggregation and at least one `transform`.
6. **Combination** — at least one `merge` (with a second table you obtain or construct) and one `concat`.
7. **Reshaping** — at least one `pivot_table` and one `melt` or `crosstab`.
8. **Visualization** — at least **five** charts of different kinds, each answering a question.
9. **Conclusions and limitations** — clear answers to your questions and what your data cannot tell you.
10. **Reproducibility** — relative paths, a `requirements.txt` or list of package versions, and *Restart & Run All* without errors.
