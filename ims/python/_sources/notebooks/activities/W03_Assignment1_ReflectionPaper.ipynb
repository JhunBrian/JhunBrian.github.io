## **Guide Questions**

Address **all** of the following, in an organized essay (not a numbered list of answers):

1. **Languages.** Before this course, how did you imagine computers "understand" instructions? What changed after learning about machine code, high-level languages, and interpreters?
2. **Algorithms.** Describe one everyday task you performed recently as an algorithm. Was it easy to make every step **definite**? What did this teach you about how precise programs must be?
3. **Paradigms.** Explain structured, procedural, and functional programming **in your own analogies**. Which way of thinking feels most natural to you, and why?
4. **Python and data science.** Why do you think Python became the main language of data science? Use at least one characteristic from the lesson.
5. **Jupyter.** How might notebooks change the way you learn and experiment, compared with writing code in a plain file?
6. **Self-assessment.** What was the most difficult concept so far? What will you do to improve, and what do you hope to be able to build by the end of the semester?
