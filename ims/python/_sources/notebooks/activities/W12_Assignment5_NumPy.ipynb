## **Problem 1 — Exam Item Analysis (2D)**

Generate an answer matrix for **40 students × 20 items** with `rng = np.random.default_rng(<your student number>)`, where each entry is 1 (correct) with probability 0.7. Compute:

1. each student's score and the class mean, median, and standard deviation;
2. each item's **difficulty** (proportion correct) and the three hardest items;
3. the **top 25%** and **bottom 25%** of students (by score), and each item's **discrimination index** = difficulty in the top group − difficulty in the bottom group;
4. the items that should be revised (difficulty < 0.3 or discrimination < 0.1).

## **Problem 2 — Image as an Array (3D)**

Create an **8 × 8 × 3** "image" of integers 0–255 (`uint8`) with your generator. Then:

1. convert it to grayscale with the weights 0.299 R + 0.587 G + 0.114 B;
2. flip it horizontally and vertically using slicing;
3. increase brightness by 40 **without overflow** (clip at 255);
4. explain why `uint8` arithmetic can overflow and how you avoided it.

## **Problem 3 — Matrix Problem**

A canteen sells three meal sets. Each set uses rice, viand, and drink portions given by a 3 × 3 matrix, and the cost of each portion is a vector. Use `@` to compute the **cost of each set**, and use `np.linalg.solve` to find how many of each set were sold given the total portions used in a week (make up consistent numbers and verify your answer).

## **Problem 4 — Comparison**

Write a short table comparing **1D, 2D, and 3D arrays** using examples from Problems 1–3: when is each shape the natural choice, and what are the drawbacks?
