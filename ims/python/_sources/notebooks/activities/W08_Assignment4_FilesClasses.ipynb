## **Overview**

Build an **inventory manager** for a school supply store that reads its data from a CSV file, handles bad data gracefully, and represents each product as an **object**.

## **Requirements**

### **Part A — The data file**

Create `inventory.csv` with at least **12** products and the columns `code,name,category,price,stock`. Deliberately include **three bad rows**: one with a missing price, one with a non-numeric stock, and one with a negative price.

### **Part B — The `Product` class**

| Member | Behavior |
| ------ | -------- |
| `__init__(code, name, category, price, stock)` | stores attributes; raises `ValueError` if price ≤ 0 or stock < 0 |
| `value()` | price × stock |
| `restock(qty)` | adds stock; rejects qty ≤ 0 |
| `sell(qty)` | reduces stock and returns the sale amount; raises `ValueError` if there is not enough stock |
| `__str__()` | a readable one-line description |

### **Part C — Loading with exception handling**

Write `load_products(path)` that returns a list of `Product` objects. Rows that cannot be converted or that fail validation are **skipped**, and a message names the row number and the reason. A missing file prints a friendly message and returns an empty list.

### **Part D — Operations and report**

1. Simulate a day of transactions from a list such as `[("sell", "NB01", 3), ("restock", "PN02", 20), ("sell", "XX99", 1), ...]` — including at least one **unknown code** and one **oversell**, both handled without crashing.
2. Write `report.csv` with each product's code, name, stock, and value, and a **low-stock** flag (stock < 5).
3. Print the total inventory value per category.

### **Part E — Reflection (Markdown)**

* Where did exception handling prevent your program from crashing?
* Why is it useful that `Product` validates its own data?
