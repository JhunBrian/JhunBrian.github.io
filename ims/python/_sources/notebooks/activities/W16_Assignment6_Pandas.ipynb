## **Overview**

Analyze the course's **canteen** datasets (`data/canteen_sales.csv` and `data/menu.csv`, synthetic) together with a **third table you create**, and present your findings to the "canteen manager".

## **Tasks**

1. **Load and inspect.** Read both files with appropriate dtypes and parsed dates. Report shapes, dtypes, and missing values.
2. **Create a third table.** Make a `staff.csv` table assigning a cashier to each `branch` and **weekday** (you design it). Read it back with pandas.
3. **Merge.** Combine sales, menu, and staff. Validate the merges and report any unmatched rows.
4. **Compute.** Add `revenue`, `weekday`, and `week` columns. Use `np.where` or `pd.cut` to classify days as *slow*, *normal*, or *busy* (justify your thresholds).
5. **Aggregate.** Answer with `groupby`/`agg`:
   * revenue per branch, per category, and per cashier;
   * the best-selling item per branch;
   * each item's **share** of its branch's revenue (`transform`).
6. **Pivot.** Build a pivot table of revenue by **week × category** with totals, and a `crosstab` of day class by branch.
7. **Visualize.** At least **four** clear charts (a trend, a comparison, a distribution, and one of your choice), each with a title, labeled axes, and a one-sentence takeaway.
8. **Recommendations.** Write three recommendations to the manager, each supported by a number or chart from your analysis.
