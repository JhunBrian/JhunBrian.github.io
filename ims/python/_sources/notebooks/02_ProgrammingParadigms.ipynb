Imagine three people asked to make **adobo** for a family gathering:

* **Lola** cooks from memory, step by step, tasting and adjusting as she goes.
* **A restaurant kitchen** splits the work: one station prepares the meat, another the sauce, another the plating — each station has one job and passes its result on.
* **A food factory** runs the ingredients through a line of machines: wash → cut → marinate → cook → pack. Each machine takes something in and produces something new, without changing the machine itself.

All three produce adobo, but they **organize the work differently**.

Programming works the same way. A **programming paradigm** is a **style or way of thinking** about how to organize a program — how to structure its steps, its data, and its logic. A paradigm is not a language; it is an **approach**. Many languages, including Python, let you use several paradigms.

> **Key idea:** a paradigm answers the question *"How should I organize my solution?"*

## **The Family of Paradigms**

Paradigms are often grouped into two big families:

```{mermaid}
flowchart TB
    P["Programming paradigms"] --> I["Imperative<br/>describe HOW, step by step"]
    P --> D["Declarative<br/>describe WHAT result is wanted"]
    I --> S["Structured"]
    I --> PR["Procedural"]
    I --> O["Object-oriented"]
    D --> F["Functional"]
    D --> Q["Query / logic<br/>(SQL, Prolog)"]
```

This course focuses on the three paradigms you will use most as a beginning data scientist:

| Paradigm | Guiding question | Organizes the program around… |
| -------- | ---------------- | ------------------------------ |
| **Structured** | How do I control the flow of steps clearly? | sequence, selection, and iteration — no jumping around |
| **Procedural** | What smaller tasks (procedures) make up the job? | named, reusable procedures (functions) |
| **Functional** | What transformations turn the input into the output? | functions that take values in and return new values out |

We will also take a short look at **object-oriented** and **declarative** programming, which you will meet again later in the course and in DS212.

To compare paradigms fairly, we will solve **the same problem** in each one.

> **Running problem.** You have a list of quiz scores. Keep only the **passing** scores (75 and above), add a **5-point bonus** to each (but never above 100), and compute the **average** of the bonus-adjusted passing scores.
>
> Scores: `78, 92, 64, 85, 70, 99`

## **The Starting Point: Imperative Programming**

**Imperative** programming describes a computation as a **sequence of commands that change the program's state** (the values stored in its variables). It answers: *"What steps should the computer perform, in what order?"*

> **Analogy — a whiteboard.** Imagine a whiteboard with a few labeled boxes (`total`, `count`). Each instruction tells you to **erase and rewrite** a box: *"add 83 to total", "add 1 to count"*. At the end, you read the answer from the board. The **state** of the whiteboard changes with every step.

All the paradigms below except functional are **refinements of imperative programming** — they add rules and organization on top of it.

## **Structured Programming**

### **The problem it solved: spaghetti code**

Early programs often used `GOTO` statements that jumped to any line of the program. As programs grew, the flow became a tangle that was almost impossible to follow — programmers called it **spaghetti code**.

```text
10  INPUT score
20  IF score < 75 GOTO 60
30  LET score = score + 5
40  IF score > 100 GOTO 80
50  GOTO 90
60  PRINT "skip"
70  GOTO 10
80  LET score = 100
90  PRINT score
95  GOTO 10
```

Try following it: line 20 jumps to 60, which jumps back to 10; line 40 jumps to 80, which falls into 90, which jumps to 10... Now imagine 10,000 lines like this.

> **Analogy — a road map with teleporters.** Spaghetti code is like a city where any street might suddenly **teleport** you to another part of town. Structured programming replaces the teleporters with **ordinary roads, intersections, and roundabouts** — you can always see where you came from and where you are going.

In 1968, Edsger Dijkstra published his famous letter *"Go To Statement Considered Harmful,"* arguing that programs should be built only from **clean, nestable control structures**. Two years earlier, Böhm and Jacopini had proven this was always possible.

### **The rules of structured programming**

1. Use only **three control structures**: **sequence**, **selection** (`if`), and **iteration** (`while`, `for`).
2. Every structure has **one entry and one exit**.
3. Structures may be **nested** inside each other, but never **jump** into or out of the middle of one.
4. Use **indentation** so the structure is visible.

> **Analogy — LEGO blocks.** You have only three kinds of blocks, and each has exactly one connector on top and one at the bottom. You can stack them, or put one inside another — and anything you build stays sturdy and easy to understand.

### **The running problem, structured**

```text
START
    SET scores ← [78, 92, 64, 85, 70, 99]
    SET total ← 0
    SET count ← 0
    FOR each score IN scores DO                 ← iteration
        IF score ≥ 75 THEN                       ← selection
            SET adjusted ← score + 5             ← sequence
            IF adjusted > 100 THEN
                SET adjusted ← 100
            ENDIF
            SET total ← total + adjusted
            SET count ← count + 1
        ENDIF
    ENDFOR
    OUTPUT total / count
END
```

```{mermaid}
flowchart TD
    S([Start]) --> Z["total ← 0, count ← 0"]
    Z --> L{More scores?}
    L -->|Yes| G[Take next score]
    G --> P{score ≥ 75?}
    P -->|No| L
    P -->|Yes| A["adjusted ← score + 5"]
    A --> C{adjusted > 100?}
    C -->|Yes| H["adjusted ← 100"]
    C -->|No| T["total ← total + adjusted<br/>count ← count + 1"]
    H --> T
    T --> L
    L -->|No| O[/Output total ÷ count/]
    O --> E([End])
```

Every path in the flowchart moves **forward** or **loops back to the loop's single decision** — there are no jumps into the middle of anything.

### **Strengths and limits**

| Strengths | Limits |
| --------- | ------ |
| easy to read, trace, and debug | everything is still in one long block |
| the flowchart and the code match closely | the same logic (e.g., capping at 100) may be repeated in many places |
| the basis of almost every modern language | large programs need more organization → **procedural** |

## **Procedural Programming**

**Procedural** programming keeps the structured rules but organizes the program into **procedures** (also called **subroutines** or **functions**): named blocks of steps that perform one task, can receive **parameters**, and can **return** a result.

> **Analogy — a barangay hall.** Instead of one person doing everything, the hall has windows: one issues **clearances**, one handles **cedulas**, one records **complaints**. Each window does **one job**, follows its own steps, and returns a document. The **barangay captain** (the main program) just sends each request to the right window.

> **Another analogy — a recipe with sub-recipes.** A cookbook's lasagna recipe says *"prepare the white sauce (see page 12)."* You don't rewrite the white-sauce steps inside every recipe that needs it — you **call** the sub-recipe.

### **Decomposition**

The heart of procedural programming is **decomposition**: breaking a problem into smaller procedures.

```{mermaid}
flowchart TB
    M["main program<br/>compute bonus average"] --> P1["is_passing(score)"]
    M --> P2["add_bonus(score)"]
    M --> P3["average(values)"]
    P2 --> P4["cap(value, limit)"]
```

### **The running problem, procedural**

```text
PROCEDURE is_passing(score)
    RETURN score ≥ 75
ENDPROCEDURE

PROCEDURE add_bonus(score)
    RETURN cap(score + 5, 100)
ENDPROCEDURE

PROCEDURE cap(value, limit)
    IF value > limit THEN RETURN limit ELSE RETURN value ENDIF
ENDPROCEDURE

PROCEDURE average(values)
    SET total ← 0
    FOR each v IN values DO SET total ← total + v ENDFOR
    RETURN total / LENGTH(values)
ENDPROCEDURE

MAIN
    SET adjusted ← empty list
    FOR each score IN scores DO
        IF is_passing(score) THEN
            APPEND add_bonus(score) TO adjusted
        ENDIF
    ENDFOR
    OUTPUT average(adjusted)
ENDMAIN
```

In a flowchart, a call to a procedure is drawn with the **predefined process** symbol:

```{mermaid}
flowchart TD
    S([Start]) --> L{More scores?}
    L -->|Yes| P[["is_passing(score)"]]
    P -->|True| B[["add_bonus(score)"]]
    B --> A["append to adjusted"]
    A --> L
    P -->|False| L
    L -->|No| AV[["average(adjusted)"]]
    AV --> O[/Output result/]
    O --> E([End])
```

### **Why procedures help**

| Benefit | Example |
| ------- | ------- |
| **Reuse** | `average()` works for scores, prices, temperatures… |
| **Readability** | the main program reads like a summary of the solution |
| **Easier testing** | test `add_bonus(98)` by itself → should give 100 |
| **Easier change** | change the bonus rule in **one** place |
| **Teamwork** | different people can write different procedures |

> **Important:** procedures can still **change shared data** (state). A procedure that modifies a global variable can surprise other parts of the program. Functional programming tackles exactly this problem.

Procedural programming is the **main paradigm of this course**: from Week 6 onward, you will design almost every program as a set of well-named functions. Languages such as C, Pascal, and Fortran are classic procedural languages, and Python supports the style naturally.

## **Functional Programming**

**Functional** programming treats computation as **evaluating functions** — in the mathematical sense. A function takes values **in** and gives a **new value out**, and it does **not change anything else**.

> **Analogy — a juice machine.** Put in mangoes, get mango juice. Put in the same mangoes tomorrow, get the same juice. The machine does not secretly change the fruit on your counter or keep a tally on the wall — it only **transforms input into output**.

> **Analogy — an assembly line.** Raw materials enter, pass through stations — *wash*, *cut*, *filter out the bad pieces*, *package* — and a finished product comes out. Each station is a function; the product of one station becomes the input of the next.

### **Core ideas**

| Idea | Meaning | Analogy |
| ---- | ------- | ------- |
| **Pure functions** | same input → same output, no side effects | a calculator: `7 × 8` is always 56, and pressing it doesn't change anything else |
| **Immutability** | data is not changed in place; new data is created | photocopying a form and filling in the copy, keeping the original clean |
| **Functions as values** | functions can be passed to other functions | giving a machine a different **attachment** (blade, whisk, grinder) |
| **Composition** | build big transformations by chaining small ones | stations in an assembly line |

### **Side effects — the thing functional programming avoids**

A **side effect** is anything a function does **besides** returning a value: changing a global variable, modifying a list it received, printing, writing to a file.

```text
Impure (has a side effect):                Pure (no side effect):

SET bonus_total ← 0                        FUNCTION add_bonus(score)
PROCEDURE add_bonus(score)                     RETURN MIN(score + 5, 100)
    SET bonus_total ← bonus_total + 5      ENDFUNCTION
    RETURN score + 5
ENDPROCEDURE
```

The impure version secretly changes `bonus_total` every time it is called — call it twice by accident and the total is wrong. The pure version can be called a thousand times safely.

### **Three classic tools: map, filter, reduce**

Functional programs rarely write loops directly. Instead, they use general-purpose **higher-order functions** that take another function as input:

| Tool | What it does | Analogy |
| ---- | ------------ | ------- |
| **filter** | keep only the items that pass a test | a sieve that lets only the good grains through |
| **map** | apply a function to **every** item | a stamping machine that stamps every envelope |
| **reduce** | combine all items into one value | a blender that turns many fruits into one smoothie |

### **The running problem, functional**

```{mermaid}
flowchart LR
    A["[78, 92, 64, 85, 70, 99]"] -->|"filter: score ≥ 75"| B["[78, 92, 85, 99]"]
    B -->|"map: min(score + 5, 100)"| C["[83, 97, 90, 100]"]
    C -->|"reduce: average"| D["92.5"]
```

```text
result ← average( map(add_bonus, filter(is_passing, scores)) )
```

There is **no counter, no running total, no variable being overwritten** — just data flowing through transformations. The original `scores` list is never modified.

### **Why it matters in data science**

Data analysis is naturally a **pipeline of transformations**: *load → clean → filter → transform → summarize*. Libraries such as **pandas** and **NumPy** are designed around this style, which is why functional thinking will make Weeks 10–16 much easier.

| Strengths | Limits |
| --------- | ------ |
| easy to test (pure functions) | some problems are naturally step-by-step |
| safe to reuse and combine | can feel unfamiliar at first |
| pipelines read like a description of the data flow | long chains can become hard to debug |

## **A Brief Look at Two More Paradigms**

### **Object-oriented programming (OOP)**

OOP organizes a program around **objects** that bundle **data** (attributes) together with the **actions** (methods) that work on that data.

> **Analogy — students in a class.** Each student *has* data (name, ID, scores) and *can do* things (submit work, compute their own average). Instead of one central program keeping track of everyone's scores, **each student object manages its own**.

```text
student ← Student("Ana", [78, 92, 64])
student.passing_average()
```

You will meet objects in Week 8, and OOP is the focus of **DS212**.

### **Declarative programming**

Declarative programming describes **what result you want**, and lets the system decide **how** to produce it.

> **Analogy — ordering food.** At a restaurant, you say *"one chicken adobo, extra rice."* You don't tell the cook how to light the stove. Using a ride-hailing app is similar: you set the **destination**, not every turn on the route.

The classic declarative language is **SQL**:

```sql
-- SQLite syntax: MIN(a, b) returns the smaller of two values
SELECT AVG(MIN(score + 5, 100))
FROM quiz_scores
WHERE score >= 75;
```

In Python, data libraries such as pandas let you write in a similar style: `scores[scores >= 75]` describes **which** scores you want without writing the loop.

## **Side-by-Side Comparison**

| | Structured | Procedural | Functional | Object-oriented | Declarative |
| - | ---------- | ---------- | ---------- | --------------- | ----------- |
| **Main idea** | clear control flow | decompose into procedures | transform data with pure functions | objects with data + behavior | describe the result |
| **Basic unit** | control structures | procedure / function | pure function | object / class | query / rule |
| **State changes** | yes | yes (often local) | avoided | inside objects | hidden from the programmer |
| **Analogy** | LEGO blocks | barangay windows | juice machine / assembly line | students managing their own records | ordering at a restaurant |
| **Example languages** | Pascal, C | C, Fortran, Python | Haskell, Lisp; Python & R support it | Java, C#, Python | SQL, Prolog |
| **In this course** | Weeks 2–5 | Weeks 6–9 (main) | Weeks 6–7, 10–16 | Week 8 (intro) | pandas, Weeks 13–16 |

## **Python as a Multi-Paradigm Language**

Python does not force a single paradigm. You can write the running problem in **each** style. You will learn the syntax from Week 4 onward — for now, simply **read** the code and match it to the pseudocode above.

````python
scores = [78, 92, 64, 85, 70, 99]

# ---------- Structured (imperative) ----------
total = 0
count = 0
for score in scores:
    if score >= 75:
        adjusted = score + 5
        if adjusted > 100:
            adjusted = 100
        total = total + adjusted
        count = count + 1
print("Structured:", total / count)
````

````python
# ---------- Procedural ----------
def is_passing(score):
    return score >= 75


def cap(value, limit):
    return limit if value > limit else value


def add_bonus(score):
    return cap(score + 5, 100)


def average(values):
    return sum(values) / len(values)


adjusted = []
for score in scores:
    if is_passing(score):
        adjusted.append(add_bonus(score))
print("Procedural:", average(adjusted))
````

````python
# ---------- Functional ----------
from functools import reduce

passing = filter(lambda s: s >= 75, scores)
bonus = list(map(lambda s: min(s + 5, 100), passing))
total = reduce(lambda a, b: a + b, bonus)
print("Functional:", total / len(bonus))
````

````python
# ---------- Declarative style (with pandas) ----------
import pandas as pd

s = pd.Series(scores)
print("Declarative:", (s[s >= 75] + 5).clip(upper=100).mean())

print("Original scores unchanged:", scores)
````

All four produce **92.5** — the paradigms differ in **organization**, not in the answer.

Python itself is primarily **imperative** with strong support for **objects**, and it borrows many **functional** tools such as `lambda`, `map()`, `filter()`, and comprehensions. Declarative-style Python appears mainly through libraries such as pandas. As the original course notes put it, it is useful to distinguish between *Python being a declarative language* and *writing declarative-style Python*.

## **Choosing a Paradigm**

There is no single "best" paradigm. Good programmers **mix** them:

| Situation | A natural choice |
| --------- | ---------------- |
| a short script that runs once | structured / imperative |
| a program with repeated tasks | procedural — put each task in a function |
| a data-cleaning pipeline | functional — chain transformations |
| a system of interacting things (students, courses, sensors) | object-oriented |
| selecting and summarizing table data | declarative (pandas, SQL) |

## **Summary**

* A **paradigm** is a way of organizing a program, not a language.
* **Imperative** programs change state step by step; **declarative** programs describe the result.
* **Structured** programming uses only sequence, selection, and iteration, each with one entry and one exit — no spaghetti jumps.
* **Procedural** programming decomposes a problem into named, reusable procedures.
* **Functional** programming builds results by passing data through pure functions, avoiding side effects.
* **Python** supports all of these, so you can choose — and combine — the style that fits the problem.
