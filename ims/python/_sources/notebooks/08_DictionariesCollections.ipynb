Lists find items by **position**. Often, however, we want to find information by a **name** or **key**: a student by ID, a price by item code, a word's count by the word itself. For that, Python provides **dictionaries**. This lesson also covers **sets** and the handy tools in the **`collections`** module.

> **Analogy.** A **dictionary** is like a **phone's contact list**: you look up *"Ana"* (the **key**) and get her number (the **value**). You never need to remember that Ana is contact number 17.

## **Dictionaries**

A dictionary stores **key–value pairs**. Keys must be unique and immutable (strings, numbers, tuples); values can be anything.

```python
prices = {"RICE": 15, "ADOBO": 80, "HALO": 65}

print(prices["ADOBO"])            # look up by key
prices["BUKO"] = 40               # add a new pair
prices["RICE"] = 18               # update
del prices["HALO"]                # remove
print(prices)
print(len(prices), "ADOBO" in prices, "HALO" in prices)
```

### **Safe lookups**

Looking up a missing key raises `KeyError`. Use `get()` with a default instead:

```python
try:
    prices["PIZZA"]
except KeyError as error:
    print("KeyError:", error)

print(prices.get("PIZZA"), prices.get("PIZZA", 0))
```

### **Looping over a dictionary**

```python
for code in prices:                       # keys
    print(code, end=" ")
print()

for code, price in prices.items():        # key–value pairs
    print(f"{code:<6} ₱{price}")

print(list(prices.keys()), list(prices.values()))
```

### **Records: dictionaries inside lists**

A list of dictionaries is a natural way to store a **table of records** — and it is exactly how pandas thinks of rows later in the course.

```python
students = [
    {"id": "2026-001", "name": "Ana", "scores": [88, 92, 95]},
    {"id": "2026-002", "name": "Ben", "scores": [75, 70, 80]},
    {"id": "2026-003", "name": "Carlo", "scores": [91, 85, 89]},
]

for s in students:
    avg = sum(s["scores"]) / len(s["scores"])
    print(f'{s["id"]}  {s["name"]:<6} {avg:.2f}')

by_id = {s["id"]: s for s in students}           # dictionary comprehension: index records by ID
print(by_id["2026-003"]["name"])
```

### **The counting pattern**

```python
votes = ["adobo", "sinigang", "adobo", "pancit", "adobo", "sinigang"]

counts = {}
for vote in votes:
    counts[vote] = counts.get(vote, 0) + 1
print(counts)
print(max(counts, key=counts.get), "wins")
```

## **Sets**

A **set** is an **unordered collection of unique items**. It is ideal for removing duplicates and for membership tests and comparisons.

> **Analogy.** A set is like the **guest list at an event** — each person is on it once, and the order doesn't matter.

```python
ds1a = {"Ana", "Ben", "Carlo", "Diana"}
math_club = {"Carlo", "Diana", "Ella"}

print(ds1a & math_club)        # intersection: in both
print(ds1a | math_club)        # union: in either
print(ds1a - math_club)        # difference: in DS1A only
print("Ella" in ds1a)

emails = ["a@x.com", "b@x.com", "a@x.com"]
print(len(emails), "emails,", len(set(emails)), "unique")
```

## **The `collections` Module**

| Tool | What it is | Use for |
| ---- | ---------- | ------- |
| `Counter` | a dictionary that counts | frequencies, "most common" |
| `defaultdict` | a dictionary with an automatic default value | grouping items |
| `namedtuple` | a tuple with named fields | readable fixed records |
| `deque` | a list optimized for adding/removing at both ends | queues |

```python
from collections import Counter, defaultdict, namedtuple, deque

counts = Counter(votes)
print(counts.most_common(2))

by_section = defaultdict(list)
for name, section in [("Ana", "DS1A"), ("Ben", "DS1B"), ("Carlo", "DS1A")]:
    by_section[section].append(name)
print(dict(by_section))

Item = namedtuple("Item", ["code", "name", "price"])
adobo = Item("ADOBO", "Chicken adobo", 80)
print(adobo.name, adobo.price)

queue = deque(["Ana", "Ben"])
queue.append("Carlo")          # join the line
print(queue.popleft(), "is served; waiting:", list(queue))
```

### **Example: the most common words in hotel reviews**

The course dataset `data/reviews.csv` contains 10,000 hotel reviews. Using only the tools from this lesson:

```python
import csv
from collections import Counter

STOPWORDS = {"the", "and", "a", "to", "was", "i", "we", "is", "of", "in", "it", "for",
             "our", "my", "with", "were", "very", "but", "at", "this", "had", "not", "you",
             "on", "they", "that", "are", "there", "room", "hotel", "stay", "be", "as", "so", "would"}

counter = Counter()
with open("data/reviews.csv", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        for word in row["reviews.text"].lower().split():
            word = word.strip(".,!?\"'()")
            if word.isalpha() and word not in STOPWORDS:
                counter[word] += 1

print(counter.most_common(10))
```

## **Choosing a Data Structure**

| Structure | Ordered | Mutable | Duplicates | Look up by | Advantages | Disadvantages |
| --------- | ------- | ------- | ---------- | ---------- | ---------- | ------------- |
| `list` | yes | yes | allowed | position | flexible; many methods | searching for a value is slow for large lists |
| `tuple` | yes | no | allowed | position | protected; can be a dict key | cannot be changed |
| `str` | yes | no | allowed | position | text methods | characters only |
| `dict` | insertion order | yes | keys unique | **key** | very fast lookups by key | uses more memory; keys must be immutable |
| `set` | no | yes | **not allowed** | membership | fast `in` tests; set math | no order, no indexing |

> **Ask yourself:** *Do I look things up by position or by name? Can it change? Do duplicates matter?*

## **Summary**

| Concept | Key idea |
| ------- | -------- |
| dictionary | key → value; `d[key]`, `get()`, `items()` |
| list of dicts | a table of records |
| counting pattern | `d[k] = d.get(k, 0) + 1`, or `Counter` |
| set | unique, unordered; union, intersection, difference |
| `collections` | `Counter`, `defaultdict`, `namedtuple`, `deque` |
