An **Integrated Development Environment (IDE)** is a program that brings together the tools a programmer needs: an editor to write code, a way to run it, and tools to inspect results and fix errors.

For data science, the most popular environment is the **Jupyter Notebook**. A notebook is a single document that mixes **code**, **results**, **explanations**, **equations**, and **charts** — like a **laboratory notebook** where you record your procedure, your observations, and your conclusions side by side.

```{mermaid}
flowchart LR
    B["Your browser<br/>(the notebook page)"] <--> S["Jupyter server"]
    S <--> K["Kernel<br/>(the running Python process)"]
    S <--> F[".ipynb file<br/>on your computer"]
```

* The **notebook** is the page you see and edit.
* The **kernel** is the Python process that actually runs your code and **remembers** your variables.
* The file is saved with the extension **`.ipynb`** (*IPython notebook*).

> **Analogy.** The notebook is the **chalkboard** in front of the class; the kernel is the **teacher's memory** of everything written so far. Erasing part of the board does not make the teacher forget — and restarting the kernel is like the teacher forgetting everything at once.

## **Cells**

A notebook is a list of **cells**. There are two main types:

* **Code cells** contain Python. Running one sends the code to the kernel and shows the **output** below it. The value of the **last line** is displayed automatically.
* **Markdown cells** contain formatted text: headings, lists, tables, links, images, and math.

Try reading these code cells and their outputs:

````python
# A code cell: the last expression is displayed automatically
price = 85
quantity = 3
price * quantity
````

````python
# print() shows values anywhere in the cell
print("Subtotal:", price * quantity)
print("With 12% VAT:", round(price * quantity * 1.12, 2))
````

## **Command Mode and Edit Mode**

Jupyter has two modes:

* **Edit mode** (green or blue border, a cursor in the cell) — you are typing inside a cell.
* **Command mode** (no cursor) — keyboard shortcuts act on whole cells. Press **Esc** to enter command mode and **Enter** to edit.

| Shortcut | Mode | Action |
| -------- | ---- | ------ |
| `Shift + Enter` | both | run the cell and move to the next |
| `Ctrl + Enter` | both | run the cell and stay |
| `Alt + Enter` | both | run the cell and insert a new one below |
| `A` / `B` | command | insert a cell **a**bove / **b**elow |
| `M` / `Y` | command | change the cell to **M**arkdown / code |
| `D`, `D` | command | delete the cell |
| `Z` | command | undo cell deletion |
| `0`, `0` | command | restart the kernel |
| `Tab` | edit | auto-complete a name |
| `Shift + Tab` | edit | show help for a function |

## **Writing in Markdown**

Markdown turns plain text into formatted text. Use Markdown cells to **explain your thinking** — a notebook without explanations is just a script.

| You type | You get |
| -------- | ------- |
| `# Title`, `## Section` | headings |
| `**bold**`, `*italic*` | **bold**, *italic* |
| `` `code` `` | `code` |
| `- item` or `1. item` | bulleted / numbered lists |
| `[USTP](https://www.ustp.edu.ph)` | a link |
| `![chart](images/chart.png)` | an image |
| `$\bar{x} = \frac{\sum x_i}{n}$` | math: $\bar{x} = \frac{\sum x_i}{n}$ |

Tables use pipes and dashes:

```text
| Name | Score |
| ---- | ----- |
| Ana  | 92    |
```

Equations use LaTeX between dollar signs. For example, the final-grade formula:

$$\text{final} = 0.30 \cdot \text{quiz} + 0.30 \cdot \text{lab} + 0.40 \cdot \text{exam}$$

## **Execution Order and the Kernel**

The number in brackets beside a code cell, such as `[3]`, shows the **order** in which cells were run — not their position on the page.

```text
[2]  total = price * quantity       ← price was defined in cell [5]?!
[5]  price = 120
```

Because the kernel **remembers** variables, running cells out of order can produce results that cannot be reproduced. Two habits prevent this:

1. **Kernel → Restart & Run All** before you submit or share a notebook — it runs every cell from top to bottom in a fresh kernel.
2. Keep notebooks **in order**: definitions near the top, used further down.

## **From Flowchart to Notebook**

A good notebook mirrors the problem-solving process: **problem → algorithm → code → test → explanation**. Here is the jeepney-fare algorithm from the previous lessons turned into a short notebook section.

**Markdown cell — the problem and algorithm**

> Compute the jeepney fare: ₱13.00 for the first 4 km, ₱1.80 per additional km, with a 20% discount for students.
>
> 1. If distance ≤ 4, fare = 13.00; otherwise fare = 13.00 + (distance − 4) × 1.80.
> 2. If discounted, multiply by 0.80.

**Code cell — the translation**

````python
distance = 10          # in a real notebook you could use input(); here we set the value directly
is_discounted = True

if distance <= 4:
    fare = 13.00
else:
    fare = 13.00 + (distance - 4) * 1.80

if is_discounted:
    fare = fare * 0.80

print(f"Fare for {distance} km: ₱{fare:.2f}")
````

**Markdown cell — the check**

> The result ₱19.04 matches the desk-check table from the *Algorithms and Flowcharts* lesson.

## **Saving and Sharing**

* **Save** often (`Ctrl + S`). Jupyter also autosaves and keeps **checkpoints** (File → Revert to checkpoint).
* **Export** with **File → Save and Export Notebook As…** to HTML or PDF for submission.
* **Download** the `.ipynb` file to share the editable notebook.
* Do not rename `.ipynb` files to other extensions; they are JSON text files that Jupyter needs to read.
