## **Python Syntax and Coding Conventions**

Before thinking about **objects**, **classes**, **inheritance**, or **polymorphism**, we need to be comfortable reading Python itself.

Consider:

```python
student_name = "Ana"
units = 21
rate_per_unit = 250.0

total_fee = units * rate_per_unit

print(f"{student_name}'s tuition is ₱{total_fee:.2f}")
```

Even this tiny program contains several fundamental Python ideas.

```{mermaid}
flowchart LR
    A["Values<br/>'Ana', 21, 250.0"] --> B["Variables"]
    B --> C["Expression<br/>units * rate_per_unit"]
    C --> D["Variable<br/>total_fee"]
    D --> E["Output<br/>print(...)"]
```

### **Python statements**

A **statement** is an instruction that Python executes.

```python
name = "Maria"
age = 20
print(name)
```

There are three statements here:

```python
name = "Maria"
```

```python
age = 20
```

```python
print(name)
```

Python normally executes them **top to bottom**.

```{mermaid}
flowchart TD
    A["name = 'Maria'"] --> B["age = 20"]
    B --> C["print(name)"]
```

This becomes particularly important in Jupyter because cells can be executed in a different order.

### **Python is case-sensitive**

Python distinguishes uppercase and lowercase letters.

```python
student = "Ana"
Student = "Ben"
STUDENT = "Carlo"

print(student)
print(Student)
print(STUDENT)
```

These are **three different identifiers**.

That means this:

```python
score = 95
print(Score)
```

produces an error because `score` and `Score` are not the same name.

This matters later with classes because conventionally we'll write:

```python
class Student:
    pass
```

but an object might be:

```python
student = Student()
```

`Student` and `student` intentionally mean different things.

### **Indentation is syntax**

In many languages, braces indicate blocks.

Conceptually:

```text
if condition {
    do something
}
```

Python instead uses **indentation**.

```python
temperature = 38

if temperature > 37:
    print("High temperature")
    print("Consider resting")

print("Program finished")
```

Visualize the structure:

```{mermaid}
flowchart TD
    A["if temperature > 37"] --> B["Indented block"]
    B --> C["print('High temperature')"]
    B --> D["print('Consider resting')"]

    A --> E["Outside the block"]
    E --> F["print('Program finished')"]
```

The indentation isn't merely decoration.

It communicates **program structure** to Python.

Wrong:

```python
if temperature > 37:
print("High temperature")
```

Correct:

```python
if temperature > 37:
    print("High temperature")
```

The standard convention is **4 spaces per indentation level**.

### **Comments**

Comments allow programmers to explain code.

```python
# Calculate the student's final score
final_score = 90 * 0.4 + 85 * 0.6
```

Python ignores everything after `#` on that line.

But avoid comments that merely repeat the code.

Poor:

```python
# Set age to 20
age = 20
```

Better:

```python
# Minimum age required by the registration policy
minimum_age = 18
```

The second explains **why** the value exists.

## **Naming variables**

Compare these two programs.

* **Version A**

```python
x = 15
y = 120
z = x * y
```

* **Version B**

```python
quantity = 15
price_per_item = 120
total_cost = quantity * price_per_item
```

Both work.

But Version B communicates the meaning of the computation.

That's going to become increasingly important in OOP because we'll eventually have names such as:

```python
student

student_id
calculate_average
enrollment_status
```

instead of:

```python
x
a
data1
thing
func
```

### **`snake_case`**

Python convention generally uses **snake_case** for variables and functions.

```python
student_name = "Ana"
total_score = 95
number_of_students = 40
```

Rather than:

```python
studentName = "Ana"
TotalScore = 95
numberofstudents = 40
```

For functions:

```python
def calculate_average():
    pass
```

Later, you'll notice that **classes use a different convention**:

```python
class StudentRecord:
    pass
```

So remember:

| Element  | Convention   | Example               |
| -------- | ------------ | --------------------- |
| Variable | `snake_case` | `student_name`        |
| Function | `snake_case` | `calculate_average()` |
| Class    | `PascalCase` | `StudentRecord`       |
| Constant | `UPPER_CASE` | `MAX_STUDENTS`        |

This visual distinction helps us recognize what we're reading before understanding the implementation.

### **Whitespace**

These are technically very different in readability:

```python
total=price*quantity+tax
```

and

```python
total = price * quantity + tax
```

Prefer the second.

Similarly:

```python
x=10
name="Alice"
average=(x+20)/2
```

should normally be:

```python
x = 10
name = "Alice"
average = (x + 20) / 2
```

Readable code matters especially in OOP because programs will eventually contain multiple interacting classes and methods.

## **Multi-line expressions**

Long expressions can be wrapped using parentheses.

Instead of:

```python
final_grade = prelim_grade * 0.30 + midterm_grade * 0.30 + final_exam_grade * 0.40
```

we could write:

```python
final_grade = (
    prelim_grade * 0.30
    + midterm_grade * 0.30
    + final_exam_grade * 0.40
)
```

Same computation.

Much easier to inspect.

## **Python's readability philosophy**

A useful principle throughout this course is:

> Code is read far more often than it is written.

Suppose you encounter this:

```python
a = 80
b = 90
c = (a + b) / 2
```

You have to mentally discover what everything represents.

Compare:

```python
quiz_score = 80
exam_score = 90

average_score = (quiz_score + exam_score) / 2
```

The program almost explains itself.

And that idea connects directly to OOP.

Eventually, instead of scattered values such as:

```python
student_name = "Ana"
student_id = "2026-001"
student_grade = 92
```

we'll learn to model the concept itself:

```python
student = Student(...)
```

## **Variables and Assignment**

A **variable** is a name that refers to a value stored in memory. The `=` operator **binds** a name to a value; it does not mean "equal" as in mathematics.

```python
units = 21
```

reads as: *let the name `units` refer to the value `21`.*

In Python, every value is an **object**, and a variable is simply a **label** attached to that object:

```text
units  ───►  21   (an int object)
name   ───►  "Ana"   (a str object)
```

A variable can be reassigned to a different value — even of a different type:

````python
units = 21
print(units, type(units))

units = "twenty-one"
print(units, type(units))

# multiple assignment and swapping
quiz, exam = 85, 92
quiz, exam = exam, quiz
print(quiz, exam)
````

## **Data Types**

Every value in Python has a **type**, which determines what the value can do.

| Type    | Example                      | Description                               | Mutable? |
| ------- | ---------------------------- | ----------------------------------------- | -------- |
| `int`   | `21`, `-5`                   | whole numbers                             | no       |
| `float` | `3.14`, `250.0`              | decimal numbers                           | no       |
| `str`   | `"Ana"`, `'DS212'`           | text                                      | no       |
| `bool`  | `True`, `False`              | truth values                              | no       |
| `NoneType` | `None`                    | "no value"                                | no       |
| `list`  | `[85, 92, 78]`               | ordered, changeable collection            | **yes**  |
| `tuple` | `(14.5, 121.0)`              | ordered, unchangeable collection          | no       |
| `dict`  | `{"name": "Ana", "age": 20}` | key–value pairs                           | **yes**  |
| `set`   | `{"DS", "IT", "CS"}`         | unordered collection of unique items      | **yes**  |

Use `type()` to inspect a value and `isinstance()` to test it:

````python
values = [21, 3.14, "DS212", True, None, [1, 2], (1, 2), {"a": 1}, {1, 2}]

for value in values:
    print(f"{str(value):<10} → {type(value).__name__}")

print(isinstance(3.14, float))
````

### **Mutable vs. immutable**

A **mutable** object can be changed after it is created; an **immutable** object cannot. This distinction becomes very important in OOP, because objects that share a mutable value can affect each other.

````python
scores = [85, 92]
same_scores = scores        # both names refer to the SAME list object
same_scores.append(78)
print(scores)               # the change is visible through both names

name = "ana"
upper_name = name.upper()   # strings are immutable: a NEW string is created
print(name, upper_name)
````

### **Type conversion**

Values can be converted from one type to another. This is especially necessary for user input, which always arrives as a `str`.

````python
text_score = "87"
score = int(text_score)
print(score + 3)

print(float("2.5"), str(100), bool(0), bool("hello"), list("abc"))

try:
    int("eighty")
except ValueError as error:
    print("ValueError:", error)
````

## **Operators and Expressions**

An **expression** is a combination of values, variables, and operators that Python evaluates to produce a value.

### **Arithmetic operators**

| Operator | Meaning           | Example   | Result |
| -------- | ----------------- | --------- | ------ |
| `+`      | addition          | `7 + 2`   | `9`    |
| `-`      | subtraction       | `7 - 2`   | `5`    |
| `*`      | multiplication    | `7 * 2`   | `14`   |
| `/`      | division          | `7 / 2`   | `3.5`  |
| `//`     | floor division    | `7 // 2`  | `3`    |
| `%`      | remainder (modulo)| `7 % 2`   | `1`    |
| `**`     | exponent          | `7 ** 2`  | `49`   |

````python
total_minutes = 135
hours = total_minutes // 60
minutes = total_minutes % 60
print(f"{hours} h {minutes} min")

# operators also work on other types
print("DS" + "212", "ha" * 3, [1, 2] + [3])
````

### **Assignment, comparison, logical, membership, and identity operators**

| Category   | Operators                         | Example                         |
| ---------- | --------------------------------- | ------------------------------- |
| Assignment | `=`, `+=`, `-=`, `*=`, `/=`       | `count += 1`                    |
| Comparison | `==`, `!=`, `<`, `>`, `<=`, `>=`  | `score >= 75`                   |
| Logical    | `and`, `or`, `not`                | `passed and not late`           |
| Membership | `in`, `not in`                    | `"DS" in programs`              |
| Identity   | `is`, `is not`                    | `value is None`                 |

`==` asks *"do these have the same value?"*, while `is` asks *"are these the same object?"* — a distinction we will meet again with objects.

````python
programs = ["DS", "IT", "CS"]
print("DS" in programs, "EE" not in programs)

a = [1, 2, 3]
b = [1, 2, 3]
print(a == b, a is b)    # equal values, different objects

result = None
print(result is None)
````

### **Operator precedence**

When an expression contains several operators, Python follows a fixed order (highest first): `**` → unary `-` → `*`, `/`, `//`, `%` → `+`, `-` → comparisons → `not` → `and` → `or`.

```python
2 + 3 * 4        # 14, not 20
(2 + 3) * 4      # 20
```

When in doubt, use **parentheses** — they make your intention clear to both Python and the reader.

## **Input and Output**

### **Output with `print()` and f-strings**

`print()` displays values. **f-strings** (formatted string literals) let us embed expressions and control formatting.

````python
name = "Ana"
gwa = 1.4567
tuition = 18250

print("Name:", name, sep=" ")
print(f"{name} has a GWA of {gwa:.2f}")
print(f"Tuition: ₱{tuition:,.2f}")
print(f"|{name:<10}|{name:>10}|{name:^10}|")
print("no newline here", end=" → ")
print("continues")
````

### **Input with `input()`**

`input()` pauses the program, waits for the user to type a line, and **always returns a string**:

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))     # convert before doing arithmetic
print(f"Hello {name}, next year you will be {age + 1}.")
```

```text
Enter your name: Ana
Enter your age: 19
Hello Ana, next year you will be 20.
```

> **Note:** `input()` is shown here as text rather than as a runnable cell, because this book is built automatically and nobody is present to type an answer. In your own notebooks, run it normally.

### **Reading and writing files**

Programs frequently read data from files and save results. The `with` statement opens a file and closes it automatically:

````python
import tempfile
from pathlib import Path

folder = Path(tempfile.mkdtemp())
report = folder / "scores.txt"

with open(report, "w", encoding="utf-8") as file:
    file.write("Ana,92\n")
    file.write("Ben,85\n")

with open(report, encoding="utf-8") as file:
    for line in file:
        name, score = line.strip().split(",")
        print(f"{name} scored {int(score)}")
````

## **Jupyter Notebook Workflow**

We write our lessons, exercises, and laboratory work in **Jupyter notebooks**. A notebook is a sequence of **cells** connected to a running Python process called the **kernel**.

```{mermaid}
flowchart LR
    N["Notebook<br/>(.ipynb file)"] --> C1["Markdown cell<br/>explanations"]
    N --> C2["Code cell<br/>Python code"]
    C2 -->|"Shift + Enter"| K["Kernel<br/>runs code, keeps variables"]
    K -->|"output"| C2
```

### **Cell types**

* **Code cells** contain Python. The value of the **last line** is displayed automatically.
* **Markdown cells** contain formatted text, tables, links, and equations — use them to explain your thinking.

### **Useful shortcuts** (command mode, press `Esc` first)

| Shortcut          | Action                          |
| ----------------- | ------------------------------- |
| `Shift + Enter`   | run cell, move to next          |
| `A` / `B`         | insert cell above / below       |
| `M` / `Y`         | change cell to Markdown / Code  |
| `D`, `D`          | delete cell                     |
| `0`, `0`          | restart the kernel              |

### **Execution order matters**

The number beside a cell, such as `[5]`, shows the **order** in which cells ran — not their position on the page. Running cells out of order can produce results that nobody else can reproduce.

```text
[3]  total = price * quantity     ← uses price from cell [7]?!
[7]  price = 120
```

> **Good habit:** before submitting, choose **Kernel → Restart & Run All** and confirm the notebook runs from top to bottom without errors.
