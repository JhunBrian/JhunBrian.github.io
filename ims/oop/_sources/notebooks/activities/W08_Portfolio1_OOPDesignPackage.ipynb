# **Portfolio 1: OOP Design and Package**

**DS212 — Object-Oriented Programming**

**Assessment:** Portfolio 1 · **Week 8** · **CO2**

**ILO :** Organize object-oriented code into packages and relate OOP principles to real-world applications.

## **Overview**

Design a small object-oriented system for a **real-world problem of your choice**, document the design, and implement it as a **Python package** with several modules. This is the first item of your portfolio and may become the starting point of your final project.

## **Requirements**

### **Part 1 — Problem and requirements**

* A short problem statement (who, what problem, why it matters).
* At least **five** functional requirements written as user stories.

### **Part 2 — Design**

* **CRC cards** for at least **five** classes.
* A **class diagram** showing attributes, methods, visibility, multiplicity, and at least **three different** kinds of relationships (association, inheritance, aggregation, composition, dependency).
* A **sequence diagram** for the most important scenario.
* A short table explaining where you used **abstraction**, **encapsulation**, **inheritance**, and **polymorphism/composition** — and why.

### **Part 3 — Package**

```text
<your_package>/
├── __init__.py          public API, __all__, __version__
├── <module_1>.py
├── <module_2>.py
└── <module_3>.py        at least three modules organized by responsibility
```

* Modules are organized by **responsibility** (no `utils2.py`, no single giant module).
* Imports inside the package use **relative imports**; no circular imports.
* Every public class has a docstring.

### **Part 4 — Demonstration notebook**

A notebook that imports your package and walks through a **meaningful scenario** (not just object creation), including at least one invalid operation handled gracefully.

### **Part 5 — Reflection**

* Which responsibility assignment did you change after drawing the sequence diagram, and why?
* How would your design need to change if the system grew to ten times as many users or records?
