# **Final OOP Project and Presentation**

**DS212 — Object-Oriented Programming**

**Assessment:** Final project · **Week 17** · **CO1, CO2, CO3**

**ILO :** Refine, document, package, and present the completed object-oriented application.

## **Overview**

Complete, document, package, and present your object-oriented application. The final project is assessed on the **application** itself (Laboratory/Portfolio rubric), the **written documentation** (written rubric), and the **presentation** (presentation rubric).

## **Final Deliverables**

### **1. Application** (assessed with the Laboratory/Portfolio rubric, /50)

* [ ] final source code in the GitHub repository with a clean, meaningful history;
* [ ] tagged final release (`v1.0.0`) with built distributions attached;
* [ ] passing unit and integration tests (and CI, if set up);
* [ ] installable with `pip install` in a fresh environment (from the wheel or TestPyPI);
* [ ] a demo notebook or script that runs from top to bottom.

### **2. Documentation** (assessed with the written rubric, /25)

* [ ] `README.md` (description, installation, usage, tests, license, authors);
* [ ] a **design document** with the final class and sequence diagrams, how the four core concepts are applied, and what changed since the proposal and why;
* [ ] docstrings for all public classes and methods;
* [ ] a reproducibility section: exact steps to set up and run the project from scratch;
* [ ] each member's contributions, references, and an AI acknowledgement, if applicable.

### **3. Presentation** (assessed with the presentation rubric, /30)

A **10–15 minute** presentation followed by questions:

1. the problem and its users;
2. requirements and scope;
3. design — class diagram and the reasoning behind key decisions;
4. **live demonstration** of the main scenarios, including graceful handling of invalid input;
5. quality — tests, packaging, and Git workflow;
6. reflection — challenges, lessons learned, and future improvements.

Every team member must present a part and be ready to answer design questions.
