**ILO :** Structure a distributable Python package with clear modules, metadata, dependencies, documentation, and version control.

## **Overview**

Turn your **Portfolio 1** system (or the Week 10–12 work) into a **distributable, version-controlled Python project**.

## **Requirements**

### **Project structure**

```text
<project>/
├── pyproject.toml
├── README.md
├── LICENSE
├── CHANGELOG.md
├── .gitignore
├── src/<package>/
│   ├── __init__.py
│   └── ... modules ...
├── tests/
│   └── (at least a placeholder; full tests in Week 14)
└── examples/
    └── demo.ipynb
```

### **Metadata and dependencies**

* `pyproject.toml` with a unique name, version, description, readme, `requires-python`, license, authors, `dependencies`, and a `dev` extra.
* A `requirements.txt` pinning the exact environment used for the demo notebook.

### **Documentation**

* `README.md` with: description, installation, quick-start example, how to run the demo and tests, license, and authors.
* Docstrings for every public module, class, and method (PEP 257).
* `CHANGELOG.md` with entries for each version.

### **Version control**

* A Git repository with **at least eight meaningful commits** (no single "upload project" commit).
* At least **one feature branch** merged into `main`.
* At least **two tagged releases** (for example `v0.1.0` and `v0.2.0`) following semantic versioning, with matching versions in `pyproject.toml`, `__version__`, and the changelog.

### **Evidence**

In a notebook or Markdown file, include:

1. the output of `git log --oneline --graph --decorate --all`;
2. the project tree;
3. proof that `pip install -e .` works in a **fresh** virtual environment and that the demo notebook runs.
