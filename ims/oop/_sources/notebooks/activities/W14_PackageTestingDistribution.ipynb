# **Package Testing and Distribution Exercise**

**DS212 — Object-Oriented Programming**

**Assessment:** Graded laboratory exercise · **Week 14** · **CO3**

**ILO :** Test a Python package locally and prepare a controlled distribution workflow using GitHub and TestPyPI.

## **Overview**

Add a **test suite** to your Portfolio 2 package and rehearse a complete **release** to **TestPyPI**.

## **Requirements**

### **Part A — Unit tests**

* At least **15** tests using `unittest` or `pytest`, organized in `tests/` by module.
* Tests cover **normal**, **edge**, and **error** cases for every public class.
* At least one test uses a **fixture** / `setUp()` and at least one is **parametrized** (pytest) or loops over cases with `subTest()` (unittest).
* At least one **integration test** exercising several classes together.
* All tests pass: include the full output of `python -m pytest -v` (or `python -m unittest -v`).

### **Part B — Local installation**

1. Build the distributions with `python -m build`; show the contents of `dist/`.
2. Run `python -m twine check dist/*`.
3. In a **fresh** virtual environment, install the **wheel** and run a short import-and-use check.

### **Part C — GitHub**

* Push the repository to GitHub.
* Add a **GitHub Actions** workflow that runs the tests on every push; include a screenshot or link showing a passing run.
* Create a **GitHub Release** from your latest tag, with the changelog notes and the built files attached.

### **Part D — TestPyPI**

1. Upload the package to **TestPyPI** using an **API token** (never commit or show the token).
2. In another fresh environment, install it from TestPyPI and run your quick-start example.
3. Include the TestPyPI project URL and the output of the install command.

### **Part E — Reflection**

* Which test found a real bug in your package? How did you fix it?
* What problem would you have missed by testing only from your development folder instead of a fresh environment?
