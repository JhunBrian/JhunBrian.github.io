## **Overview**

Design and implement a small set of classes whose objects **protect their own state**. The focus is on **abstraction** (a clear public interface) and **encapsulation** (controlled, validated access to data).

### **Scenario: Dormitory Room Reservation**

The university dormitory assigns **rooms** to **residents** for a semester.

* A **room** has a number, a capacity (1–4), a monthly rate, and a list of residents.
* A **resident** has a student ID, a name, and a **balance** of unpaid fees.
* A room cannot exceed its capacity, a resident cannot be assigned to two rooms, and a balance can never become negative.
* The dormitory office needs a list of residents with unpaid balances.

## **Requirements**

### **Part A — Abstraction**

1. Write an **abstract base class** `Billable` (using `abc`) with abstract methods `amount_due()` and `pay(amount)`.
2. Draw a **class diagram** (Mermaid or image) showing your classes, their **visibility** (`+`, `-`, `#`), and relationships.

### **Part B — Encapsulation**

Implement at least **three** classes (for example `Resident`, `Room`, `Dormitory`) that satisfy:

| Requirement | Example |
| ----------- | ------- |
| at least **two read-only properties** | `student_id`, `room.number` |
| at least **two properties with validating setters** | `capacity` must be 1–4; `monthly_rate` must be positive |
| internal data stored in **protected** attributes (`_name`) | `_residents`, `_balance` |
| at least one **private helper** method | `_check_capacity()` |
| a collection exposed **read-only** (return a tuple or copy) | `room.residents` |
| invalid operations raise `ValueError` with a clear message | assigning to a full room |
| `Resident` implements `Billable` | `amount_due()`, `pay()` |

### **Part C — Demonstration**

Write a short scenario that:

1. creates a dormitory with at least **three** rooms and **six** residents;
2. assigns residents to rooms, including one attempt to **overfill** a room and one attempt to assign the **same** resident twice (both rejected gracefully);
3. bills residents for one month, records several payments (including one **invalid** payment), and prints the list of residents with unpaid balances;
4. shows that outside code **cannot** directly change a balance or a room's list of residents.

### **Part D — Reflection** (Markdown)

* Which data did you make read-only, and why?
* Give one example where encapsulation **prevented** an invalid state in your demonstration.
* How does your `Billable` abstraction make it easy to bill something new (for example, a `LaundryService`) later?
