# **Integrated OOP Project Implementation**

**DS212 — Object-Oriented Programming**

**Assessment:** Final project milestone 2 · **Week 16** · **CO3**

**ILO :** Implement the OOP project using appropriate class relationships, exception handling, testing, and version control.

## **Overview**

Implement your approved project design as a working, tested, version-controlled package.

## **Requirements**

### **Implementation**

* All classes from the (revised) design are implemented in a `src`-layout package.
* Class relationships in the code **match** the class diagram (update the diagram if the design changed, and record why).
* A simple **interface** (command-line menu, notebook interface, or simple GUI) lets a user perform the main scenarios.
* Data is **persisted** (CSV, JSON, or SQLite) where the project requires it.

### **Robustness**

* A custom exception hierarchy for your domain.
* Validation at object boundaries; failed operations leave objects in a **consistent** state.
* The interface catches domain exceptions and shows **friendly messages** — users never see a raw traceback.

### **Testing**

* Unit tests for every model class and at least **two** integration tests.
* Every acceptance criterion from the proposal is covered by at least one test (include a traceability table: *requirement → test name*).

### **Version control and teamwork**

* Work is done on **feature branches** merged through **pull requests** with at least one reviewer.
* Every member has meaningful commits (the instructor will check the history).
* A short **progress report**: completed features, remaining work, problems encountered, and changes from the proposal.
