**ILO :** Implement inheritance hierarchies, method overriding, and super() to reuse and extend class behavior.

## **Overview**

Design and implement an **inheritance hierarchy** in which subclasses **reuse**, **extend**, and **override** the behavior of their parent classes.

### **Scenario: Campus Transport Fleet**

The university operates several kinds of vehicles: **shuttle buses**, **service vans**, and **electric tricycles** (*e-trikes*). All vehicles have a plate number, a capacity, and a mileage log. They differ in how they are **refuelled or recharged**, how their **operating cost** per kilometre is computed, and which **maintenance** they need.

## **Requirements**

### **Part A — Hierarchy design**

1. Draw a class diagram with an abstract base class `Vehicle` and **at least three** concrete subclasses, with **at least two levels** of inheritance (for example `Vehicle → FuelVehicle → ShuttleBus`).
2. In a Markdown table, list every method and state whether each subclass **inherits**, **extends** (calls `super()`), or **overrides** it.

### **Part B — Implementation**

| Requirement | Detail |
| ----------- | ------ |
| constructors | every subclass constructor calls `super().__init__()` |
| overriding | `operating_cost(km)` is overridden in each concrete class with a different formula |
| extending | `describe()` is extended using `super().describe()` |
| abstract method | `Vehicle` declares at least one abstract method |
| multiple inheritance | create a **mixin** (for example `GPSTrackableMixin` with `last_location()` and `update_location()`) and combine it with at least one vehicle class |
| MRO | print the `__mro__` of the mixed class and explain the order |
| class attributes | at least one class attribute that a subclass **overrides** (for example `WHEELS`) |

### **Part C — Demonstration**

1. Create a fleet of at least **five** vehicles of different types.
2. Log trips, then print each vehicle's description and total operating cost using **one loop** (no `isinstance` checks).
3. Show `isinstance()` and `issubclass()` results for at least three cases.

### **Part D — Design considerations** (Markdown)

* Where did inheritance clearly **reduce duplication**?
* Identify one relationship in your design that should **not** be inheritance, and explain why (*is-a* vs. *has-a*).
* What would break if `Vehicle` changed the signature of `operating_cost()`?
