## **Overview**

Build a **mini experiment tracker** — a tool data scientists use to record what was run, with which parameters, how long it took, and what result it produced. You will use advanced OOP features where they genuinely simplify the design.

## **Requirements**

### **Part A — Function decorators**

Implement and use:

| Decorator | Behavior |
| --------- | -------- |
| `@timed` | measures and stores the run time of the decorated function |
| `@track(experiment)` | records the function name, arguments, result, and time in an `Experiment` object (a decorator **with an argument**) |
| `@validate_params(**rules)` | checks keyword arguments against simple rules (for example `learning_rate=(0, 1)`) and raises `ValueError` otherwise |

All decorators must use `functools.wraps`.

### **Part B — Class and static methods**

Create an `Experiment` class with:

* `@classmethod from_config(cls, config: dict)` and `@classmethod from_json(cls, text)` alternative constructors;
* a `@staticmethod` helper (for example `slugify(name)` for file-safe names);
* an instance method `summary()` that prints a table of tracked runs and identifies the **best** run.

### **Part C — Controlled class creation**

Create a family of `Model` classes (at least **three** simple models, for example mean, median, and linear-trend predictors). Use **either** a custom metaclass **or** `__init_subclass__` to:

* automatically **register** each model by name; and
* **reject** at class-definition time any model that does not implement `fit()` and `predict()`.

Justify in Markdown why you chose a metaclass or `__init_subclass__`.

### **Part D — Demonstration**

Run every registered model on a small dataset with at least **two** parameter settings, tracking each run with your decorators. Print the experiment summary.

### **Part E — Reflection**

* Which decorator saved you the most repeated code?
* Show one place where using a decorator or metaclass would have made the code **worse**, and explain why you avoided it.
