## **Overview**

Build a small, flexible **data-processing pipeline** in which interchangeable components share a common interface. You will compare an inheritance-only design with a design based on **composition** and **polymorphism**.

### **Scenario: Survey Data Pipeline**

The guidance office collects student-wellness survey responses. Each response is a dictionary such as:

```python
{"id": 17, "year": 2, "sleep_hours": 6.5, "stress": 4, "comment": "  Exams next week!! "}
```

Some responses contain missing values, out-of-range numbers, duplicates, or messy text. The office wants to **clean**, **transform**, and **export** the data in different ways depending on the report.

## **Requirements**

### **Part A — Polymorphic steps**

1. Define a common interface (an abstract base class **or** a documented duck-typed protocol) with a method `apply(records)` that returns new records.
2. Implement at least **five** steps, for example `DropMissing`, `DropDuplicates`, `ClampRange(field, low, high)`, `CleanText(field)`, `AddField(name, function)`.
3. Implement at least **two** exporters with the same `export(records)` method (for example CSV text and a Markdown table).

### **Part B — Composition**

1. Create a `Pipeline` class that **has** a list of steps and **has** an exporter. Steps can be added in any order.
2. Build **two different** pipelines from the same step classes for two different reports, without writing new subclasses.

### **Part C — Duck typing**

Write one step class that does **not** inherit from your base class (if you used one) and show that the pipeline still works with it.

### **Part D — Comparison** (Markdown)

1. Sketch (diagram or pseudo-code) how the same features would look using **inheritance only** (for example `CleanAndExportCSV(Pipeline)`), and count how many classes you would need.
2. Explain when inheritance would still be the better choice.

### **Part E — Demonstration**

Use a dataset of at least **15** responses (create it yourself, including dirty records). Show the output of both pipelines.
