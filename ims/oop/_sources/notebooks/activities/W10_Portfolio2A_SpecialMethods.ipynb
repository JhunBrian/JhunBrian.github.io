## **Overview**

Implement a **data-oriented class** whose objects behave like built-in Python types by using special methods appropriately.

### **Choose one domain** (or propose your own to your instructor)

| Option | Class | Natural operations |
| ------ | ----- | ------------------ |
| A | `TimeSeries` — dated numeric observations | `len`, indexing/slicing by position, iteration, `+`/`-` between aligned series, `*` by a scalar, `==`, `in` (dates) |
| B | `Fraction`-like `Rational` | arithmetic `+ - * /` (including with `int`), comparisons, `hash`, `abs`, `float()` |
| C | `Matrix` | `+`, `-`, `*` by scalar, `@` (matrix multiplication), `==`, indexing `m[i, j]`, `len`, `repr` |
| D | `Playlist` of `Song` objects | `len`, indexing/slicing, iteration, `+` (concatenate), `in`, sorting songs by duration |

## **Requirements**

| Group | Minimum requirement |
| ----- | ------------------- |
| Representation | `__repr__` (unambiguous) **and** `__str__` (user-friendly); optionally `__format__` |
| Comparison | `__eq__` plus ordering (use `functools.total_ordering`); `__hash__` if objects are immutable, otherwise explain why they are unhashable |
| Arithmetic | at least **three** arithmetic operators, including one **reflected** method; return `NotImplemented` for unsupported types |
| Container / sequence | `__len__`, `__getitem__` (including **slices** where meaningful), `__iter__`, `__contains__` |
| Lifecycle / context | one of: a `__enter__`/`__exit__` context manager, or a `classmethod` alternative constructor that validates input |
| Validation | invalid construction or operations raise clear exceptions |

### **Justification table**

For **every** special method you implement, write one line explaining **why** the operator's meaning is natural for your class. Also list one operator you **chose not** to overload and why.

### **Demonstration**

Show each special method in action, including at least **three** cases that correctly raise `TypeError` or `ValueError`. Show your class working with built-ins such as `sorted()`, `sum()`, `max()`, `len()`, and `set()` where applicable.
