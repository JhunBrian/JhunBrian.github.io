## **Problem 1 — Designing an exception hierarchy**

An **online laboratory reservation** system has these failure situations:

* a student books a computer that does not exist;
* a computer is already reserved for the requested time;
* a reservation is outside laboratory hours (7:00–19:00);
* a student exceeds **two** reservations per day;
* the reservation file cannot be read.

1. Design a custom exception hierarchy (class diagram) with **one base exception**.
2. Decide which exceptions should carry **attributes** (for example, the conflicting time slot) and implement them.
3. Explain which situations should use a **built-in** exception instead of a custom one.

## **Problem 3 — Find and fix the exception-handling problems**

The following code "works" but handles exceptions badly. Identify **at least five** problems, explain each, and rewrite the code.

```python
class GradeImporter:
    def __init__(self, path):
        self.path = path
        self.grades = {}

    def load(self):
        try:
            f = open(self.path)
            for line in f:
                name, grade = line.split(",")
                self.grades[name] = float(grade)
            f.close()
        except:
            print("error")

    def average(self):
        try:
            return sum(self.grades.values()) / len(self.grades)
        except Exception:
            return 0

    def get(self, name):
        try:
            return self.grades[name]
        except KeyError:
            raise Exception("not found")
```

Test your rewritten version with: a missing file, a line with a missing comma, a non-numeric grade, an empty file, and an unknown student name.
