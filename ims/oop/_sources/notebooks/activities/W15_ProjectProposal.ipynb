# **Project Proposal and Class Design**

**DS212 — Object-Oriented Programming**

**Assessment:** Final project milestone 1 · **Week 15** · **CO3**

**ILO :** Define a real-world problem, identify objects and responsibilities, and create an OOP project design.

## **Overview**

Your team will propose an **object-oriented Python application** that addresses a defined real-world problem. The proposal is reviewed before implementation begins; use the feedback to refine your design.

* **Team size:** as assigned by your instructor.
* **Scope:** realistic for two weeks of implementation — small and working beats large and broken.

## **Required Contents**

Follow the structure presented in the lesson *OOP project planning and design*.

1. **Title, team members, and roles** (project lead, design lead, quality lead, documentation lead).
2. **Problem definition** — problem statement, stakeholders, and scope (*in* and *out* of scope).
3. **Requirements**
   * at least **six** user stories, each with acceptance criteria;
   * at least **three** non-functional requirements (robustness, testing, reproducibility, privacy).
4. **Class-responsibility modeling**
   * noun/verb analysis of the requirements;
   * CRC cards for **every** class (at least **six** classes).
5. **UML design**
   * a complete class diagram (attributes, methods, visibility, relationships, multiplicity);
   * at least **two** sequence diagrams for the main scenarios;
   * a table mapping the **four core concepts** and **composition** to specific parts of the design.
6. **Project structure** — planned package layout and repository structure.
7. **Plan** — milestones (Gantt chart or table) and task assignments per member.
8. **Risks** — at least three risks and how you will handle them.
9. **Prototype** — a short notebook that implements and demonstrates the **two or three most important classes**.
10. **References** and an **AI acknowledgement**, if applicable.
