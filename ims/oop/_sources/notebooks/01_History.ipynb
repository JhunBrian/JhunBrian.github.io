## **Before Object-Oriented Programming**

Before understanding **Object-Oriented Programming**, it helps to understand the problem it was trying to address.

Early computer programs were relatively small. A programmer could write a sequence of instructions describing exactly what the computer should do.

Conceptually:

```text
INPUT
  ↓
INSTRUCTION
  ↓
INSTRUCTION
  ↓
INSTRUCTION
  ↓
OUTPUT
```

This style of programming is closely associated with **procedural programming**.

A program is organized primarily around **procedures or functions** that operate on data.

For example:

```python
name = "Alice"
balance = 5000

def deposit(balance, amount):
    return balance + amount

balance = deposit(balance, 1000)
```

Here, the **data** and the **functions operating on that data** are separate.

We have:

```text
DATA
name
balance

       ↓ used by

FUNCTIONS
deposit()
withdraw()
transfer()
```

This approach works well for many problems.

However, as software systems became larger, developers needed better ways to organize increasingly complex combinations of:

* data,
* behavior,
* state,
* relationships,
* responsibilities,
* and interacting components.

This leads us toward one of the central ideas behind OOP:

> **What if a program could be organized around the entities participating in the problem rather than primarily around sequences of procedures?**

## **From Procedures to Objects**

Consider a university enrollment system.

We might have data describing:

```text
Student
Course
Instructor
Enrollment
Department
```

Each entity also has behavior.

A `Student` might:

```text
enroll()
drop_course()
view_grades()
```

A `Course` might:

```text
add_student()
remove_student()
calculate_class_average()
```

An `Instructor` might:

```text
assign_grade()
create_activity()
view_class_list()
```

Instead of keeping the data and operations completely separate, we can organize them together.

```{mermaid}
flowchart LR

A["Student Data"] --> S["Student Object"]
B["Student Behavior"] --> S

C["Course Data"] --> T["Course Object"]
D["Course Behavior"] --> T

E["Instructor Data"] --> I["Instructor Object"]
F["Instructor Behavior"] --> I
```

This is the basic direction of **object-oriented thinking**.

An object combines:

```text
STATE + BEHAVIOR
```

or, more concretely:

```text
DATA + OPERATIONS ON THAT DATA
```

## **Where Did OOP Come From?**

Object-oriented programming was not invented all at once.

Its development occurred over decades and was influenced by ideas from:

* simulation,
* interactive computing,
* computer graphics,
* programming languages,
* software engineering,
* biological systems,
* and networked computers.

A simplified historical progression is:

```{mermaid}
timeline
    title Development of Object-Oriented Programming

    1960s : Simula
          : Classes and objects
          : Simulation

    1970s : Smalltalk
          : Message passing
          : Everything as objects
          : Interactive computing

    1980s : C++
          : OOP enters mainstream systems programming

    1990s : Java
          : OOP becomes widely taught and adopted

    1991+ : Python
          : Multi-paradigm programming
          : Practical object-oriented programming

    Today : OOP across software engineering
          : Web systems
          : Data science
          : Machine learning libraries
```

The exact history is richer than this simplified timeline, but two languages are especially important to understanding the origins of OOP:

**Simula** and **Smalltalk**. Historical accounts identify Simula as foundational to classes, objects, inheritance, and dynamic lookup, while Smalltalk developed these ideas into a more thoroughly object-oriented environment. ([Cambridge University Press][1])

## **Simula: Objects Begin with Simulation**

### **The Simulation Problem**

In the 1960s, Norwegian computer scientists **Ole-Johan Dahl** and **Kristen Nygaard** worked on the programming language **Simula**.

As its name suggests, Simula was originally created for **simulation**. ([Cambridge University Press][1])

Imagine simulating a port.

You might need to represent:

```text
Ship
Cargo
Dock
Crane
Worker
Truck
```

Each entity has its own state.

For example:

```text
Ship
----------------
name
capacity
location
cargo
```

And its own behavior:

```text
Ship
----------------
load()
unload()
move()
```

Representing these entities naturally encouraged programmers to think about computational structures that represented **things with both state and behavior**.

## **Simula and the Class**

One of Simula's major contributions was the development of concepts we now associate with:

* classes,
* objects,
* inheritance,
* subtyping,
* and dynamic method lookup. ([Cambridge University Press][1])

The idea of a **class** is particularly important.

Think of a class as a description of a category of objects.

```{mermaid}
classDiagram

class Ship {
    name
    capacity
    location
    load()
    unload()
    move()
}

Ship --> ship1 : instance
Ship --> ship2 : instance
Ship --> ship3 : instance
```

The class describes what a `Ship` has and what a `Ship` can do.

Individual ships are **objects or instances** of that class.

Conceptually:

```text
CLASS
  ↓
Ship
  ↓
 ┌──────────┬──────────┬──────────┐
 ↓          ↓          ↓
Ship A     Ship B     Ship C
```

This distinction between **class** and **object** remains central to many modern OOP languages.

## **Smalltalk: A Different Vision of Computing**

Simula established many foundational mechanisms, but another language strongly shaped the philosophy of object-oriented programming:

**Smalltalk**.

Smalltalk was developed during the 1970s at **Xerox PARC**, associated particularly with the work of **Alan Kay, Dan Ingalls, Adele Goldberg**, and others. ([DOI System][2])

Smalltalk wasn't simply trying to improve simulation.

The researchers were exploring ideas about:

* personal computing,
* interactive programming,
* graphical interfaces,
* learning,
* dynamic systems,
* and communication among computational entities.

Alan Kay's historical account describes Smalltalk as part of a broader attempt to rethink computing rather than merely improve conventional programming. ([DOI System][2])

## **Alan Kay and the Meaning of "Object-Oriented"**

Alan Kay is closely associated with the term **object-oriented programming**.

Interestingly, Kay's original conception differed somewhat from the way OOP is commonly taught today.

Modern introductions frequently emphasize:

```text
Encapsulation
Abstraction
Inheritance
Polymorphism
```

Kay placed especially strong emphasis on **objects communicating through messages**, along with protected local state and late binding. ([Userpage][3])

The conceptual model is therefore not merely:

```text
OBJECT = DATA + METHODS
```

but also:

```text
OBJECT ↔ OBJECT ↔ OBJECT
       messages
```

## **Principles of Object-Oriented Programming**

From Simula, Smalltalk, and the languages that followed, a small set of principles emerged that still guide object-oriented design today.

| Principle                      | Idea                                                                                   |
| ------------------------------ | -------------------------------------------------------------------------------------- |
| **Objects combine state and behavior** | Data and the operations that act on it live together in one unit.            |
| **Classes describe objects**   | A class is a reusable description from which many objects are created.                 |
| **Objects communicate through messages** | Objects ask other objects to do things by calling their methods.            |
| **Objects protect their own state** | Other code should change an object only through the object's methods.             |
| **Specialization**             | New classes can be defined as specialized versions of existing classes.               |
| **Substitutability**           | Different objects can respond to the same message in their own way.                    |

These ideas are usually summarized as the **four core concepts** of OOP, which we study in depth later in the course:

```{mermaid}
flowchart LR
    A["Abstraction<br/>show what, hide how"] --- E["Encapsulation<br/>protect state"]
    E --- I["Inheritance<br/>reuse & specialize"]
    I --- P["Polymorphism<br/>one interface, many forms"]
```

Alongside the four concepts, experienced developers rely on a few **design principles**:

* **High cohesion** — each class has one clear purpose, and everything inside it supports that purpose.
* **Low coupling** — classes depend on each other as little as possible, so a change in one does not ripple through the whole program.
* **Single responsibility** — a class should have only one reason to change.

> **Object-oriented programming is less about syntax and more about deciding *which objects exist* and *what each one is responsible for*.**

## **The Role of OOP in Software Development and Data Science**

OOP became the dominant way of organizing large software because it helps teams:

* **model** a problem using concepts that match the real world;
* **divide work**, since different classes can be developed and tested separately;
* **reuse** code through classes, inheritance, and composition;
* **maintain** systems over years, because changes stay inside the class responsible for them.

As a data science student, you already use objects every day — usually without writing a single class:

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("grades.csv")      # df is a DataFrame object
df.describe()                       # calling a method on the object

model = LinearRegression()          # model is an estimator object
model.fit(X, y)                     # the object stores what it learned
model.predict(X_new)
```

The scikit-learn library, for example, is designed so that **every** model is an object with the same `fit()` and `predict()` methods. Learning OOP lets you understand these libraries more deeply and design your own tools in the same style.

Even the simplest Python values are objects with methods:

````python
from datetime import date

title = "object-oriented programming"
scores = [85, 92, 78]
start = date(2026, 8, 10)

print(title.title())                  # a str object's method
scores.append(88)                     # a list object's method
print(scores)
print(start.strftime("%B %d, %Y"))    # a date object's method
print(type(title), type(scores), type(start))
````

## **Python Environment Setup**

Throughout this course, we use **Python 3**, **Jupyter Notebook / JupyterLab**, and a **virtual environment** for each project.

### **Why a virtual environment?**

Different projects may need different versions of the same package. A **virtual environment** is an isolated folder containing its own Python interpreter and packages, so installing something for one project never breaks another.

```{mermaid}
flowchart LR
    S["System Python"] --> V1[".venv for DS212<br/>pandas, jupyter"]
    S --> V2[".venv for Capstone<br/>torch, opencv"]
    S --> V3[".venv for another project"]
```

### **Typical setup**

Create the environment once inside your project folder:

```bash
python -m venv .venv
```

Activate it every time you work on the project:

```bash
# Windows (Command Prompt)
.venv\Scripts\activate

# macOS / Linux
source .venv/bin/activate
```

Install packages, then record them so others can reproduce your setup:

```bash
pip install jupyterlab pandas
pip freeze > requirements.txt
```

Anyone can then recreate the same environment with:

```bash
pip install -r requirements.txt
```

Finally, launch Jupyter from the activated environment:

```bash
jupyter lab
```

You can confirm which Python version a notebook is running:

````python
import platform

print("Python version:", platform.python_version())
````

## **Responsible Computing**

Writing software is not only a technical activity. The code we write — and how we write it — affects other people.

**Academic integrity and AI tools.** The course syllabus allows AI-powered tools only as **assistive** technologies, for example for brainstorming or improving clarity. Any use must be **disclosed** in an AI acknowledgement, and you remain fully responsible for verifying the correctness and originality of your work. Using AI to generate entire assignments or exam answers is academic dishonesty.

**Data privacy.** Data science projects often use personal information. In the Philippines, the **Data Privacy Act of 2012 (RA 10173)** requires that personal data be collected for a legitimate purpose, kept secure, and used only as intended. Never commit real student records, passwords, or API keys to a shared notebook or repository.

**Reproducibility.** Results that cannot be reproduced cannot be trusted. Use virtual environments, record dependencies in `requirements.txt`, and make sure a notebook runs from top to bottom (**Kernel → Restart & Run All**) before you share it.

**Attribution.** Cite the sources, libraries, and code you build on. Respect software licenses.

**Professional conduct.** The ACM Code of Ethics asks computing professionals to *contribute to society*, *avoid harm*, *be honest and trustworthy*, and *respect privacy* — principles that apply to student projects as much as to commercial software.
