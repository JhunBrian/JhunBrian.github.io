## **From a Real-World Scenario to an OOP Design**

**Learning Objective**

Design and implement an original object-oriented system by identifying its entities, responsibilities, and relationships.

## **Part 1 — Create Your Scenario**

Think of a **creative real-world or fictional scenario** involving **at least five interacting entities**.

Your scenario should describe a system in which the entities have meaningful responsibilities and interactions. Avoid choosing entities simply to reach the five-class requirement.

Possible domains could include transportation, games, agriculture, disaster response, entertainment, education, smart cities, space exploration, business, healthcare operations, or an entirely fictional world.

**Creativity is part of the assessment.** Try to develop a scenario that is interesting, specific, and different from common examples discussed in class.

Write a short description of your scenario before creating the diagram.

> **Important:** You are encouraged to develop your own scenario rather than simply reproducing common examples such as a basic library, banking, student enrollment, or shopping system.

## **Part 2 — Create the Class Diagram**

Identify **at least five classes** from your scenario and create a UML class diagram showing their design.

Each class should contain appropriate:

* attributes;
* methods; and
* visibility where applicable.

Your diagram must also show how the classes interact.

**Do not use association as the only relationship in your diagram.** Your design must demonstrate multiple appropriate class relationships, such as:

```text
Association
Inheritance
Composition
Aggregation
Dependency
```

Use relationships because they make sense for your scenario—not simply because you need to include them.

Where appropriate, indicate **multiplicity**, such as:

```text
1
0..1
1..*
0..*
```

For example, your finished diagram should communicate more than:

```text
A ----- B
B ----- C
C ----- D
D ----- E
```

It should demonstrate that you understand **what each relationship means** and why it exists in your system.

## **Part 3 — Implement the Design in Python**

Translate your class diagram into Python.

Your implementation must contain the classes represented in your diagram and demonstrate their relationships through actual program behavior.

The program should:

1. define the classes, attributes, and methods shown in the diagram;
2. create objects from those classes;
3. implement the relationships represented in the class diagram; and
4. demonstrate the interaction of the objects through a working example.

For instance, if your diagram claims that one class **inherits** from another, that inheritance should exist in the Python implementation. If it shows **composition**, an object should actually contain or manage the corresponding object.

## **Part 4 — Demonstrate Your System**

At the end of your notebook, create a short demonstration showing your objects interacting.

Do not simply instantiate the five classes:

```python
a = ClassA()
b = ClassB()
c = ClassC()
d = ClassD()
e = ClassE()
```

Instead, demonstrate a meaningful sequence of events in your scenario.

Think of it as a **mini-simulation of your system**:

```{mermaid}
flowchart LR
    A["Create Objects"] --> B["Objects Interact"]
    B --> C["Methods Change State"]
    C --> D["Other Objects Respond"]
    D --> E["Display Result"]
```

The demonstration should make it possible to see how your class diagram translates into an actual object-oriented program.

## **Submission Requirements**

Submit **one Jupyter Notebook** containing your scenario description, UML class diagram, Python implementation, and a demonstration/output of the system.

### **Assessment Criteria**


| Criterion                             | What will be evaluated                                                                       |
| ------------------------------------- | -------------------------------------------------------------------------------------------- |
| **Scenario Creativity & Originality** | Originality, interesting interactions, and thoughtfulness of the chosen scenario             |
| **Class Design**                      | At least five meaningful entities with appropriate attributes, methods, and responsibilities |
| **Class Relationships**               | Correct and meaningful use of multiple relationship types—not association alone              |
| **Diagram–Code Consistency**          | Classes and relationships shown in the diagram are accurately represented in Python          |
| **Implementation & Interaction**      | Objects interact meaningfully and the program demonstrates the intended scenario             |
| **Code Quality**                      | Readability, organization, naming, and appropriate application of OOP principles             |
