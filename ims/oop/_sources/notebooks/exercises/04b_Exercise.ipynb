### **Task 4 — The `Course` Class**

Create a class `Course` whose constructor receives a `code` and a `start` date (a `datetime.date` object) and starts with an empty list of students.

Add the methods:

| Method                 | Returns                                                             |
| ---------------------- | ------------------------------------------------------------------- |
| `enroll(student)`      | nothing; adds the `Student` object to the course                    |
| `averages()`           | a dictionary `{name: average rounded to 2 decimals}`                |
| `top_student()`        | the **name** of the student with the highest average                |

Use this data:

```python
from datetime import date

ds212 = Course("DS212", date(2026, 8, 10))

data = {
    "Ana": [92, 88, 95],
    "Ben": [78, 74, 80],
    "Carlo": [85, 83, 90],
    "Diana": [70, 65, 72],
    "Ella": [91, 89, 94],
}

for i, (name, scores) in enumerate(data.items(), start=1):
    student = Student(name, f"2026-{i:03d}")
    for score in scores:
        student.add_score(score)
    ds212.enroll(student)

print(ds212.averages())
print(ds212.top_student())
```

Expected output:

```text
{'Ana': 91.67, 'Ben': 77.33, 'Carlo': 86.0, 'Diana': 69.0, 'Ella': 91.33}
Ana
```
