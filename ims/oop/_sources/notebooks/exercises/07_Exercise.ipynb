## **Laboratory Exercise: Barangay Health Center Scheduler**

### **Scenario**

A barangay health center wants a simple scheduling program.

* A **patient** has a name and an age.
* A **doctor** has a name and a specialty, and can see at most **3 patients per day**.
* An **appointment** connects one patient with one doctor on a specific **date**.
* The **schedule** books appointments and can list the patients a doctor will see on a given day.

### **Task 3 — Create the Package**

Create a package named `clinic` with the following structure. In Jupyter, you can create each file with the `%%writefile` cell magic, for example:

```python
%%writefile clinic/people.py
class Patient:
    ...
```

(Create the folder first with `import os; os.makedirs("clinic", exist_ok=True)`.)

```text
clinic/
├── __init__.py        exposes Patient, Doctor, Schedule
├── people.py          Patient, Doctor
└── scheduling.py      Appointment, Schedule
```

Requirements:

* `scheduling.py` imports the classes it needs from `people.py` using a **relative import**.
* `__init__.py` re-exports `Patient`, `Doctor`, and `Schedule`, and defines `__all__` and `__version__ = "0.1.0"`.
* `Schedule.book(patient, doctor, day)` returns an `Appointment` and prints a confirmation, or prints a message and returns `None` if the doctor is fully booked.
* `Schedule.patients_of(doctor, day)` returns the list of patient names for that doctor and day.

### **Task 4 — Use the Package**

In a **new code cell**, import from your package and run:

```python
from datetime import date
from clinic import Doctor, Patient, Schedule
import clinic

cruz = Doctor("Dr. Cruz", "Pediatrics")
schedule = Schedule()
day = date(2026, 10, 5)

for name, age in [("Ana", 7), ("Ben", 5), ("Carlo", 9), ("Diana", 4)]:
    schedule.book(Patient(name, age), cruz, day)

print(f"{cruz.name}, {day}: {', '.join(schedule.patients_of(cruz, day))}")
print(clinic.__version__, clinic.__all__)
```

Expected output:

```text
Booked: Ana with Dr. Cruz on 2026-10-05
Booked: Ben with Dr. Cruz on 2026-10-05
Booked: Carlo with Dr. Cruz on 2026-10-05
Dr. Cruz is fully booked on 2026-10-05
Dr. Cruz, 2026-10-05: Ana, Ben, Carlo
0.1.0 ['Patient', 'Doctor', 'Schedule']
```

> **Tip:** if you edit a module after importing it, restart the kernel before importing again — Python caches imported modules.
