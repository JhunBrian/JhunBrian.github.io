### **Task 1 — A Counting Decorator**

Write a decorator `count_calls` that counts how many times the decorated function is called. Store the count as an attribute `calls` on the wrapper, and use `functools.wraps`.

```python
@count_calls
def read_sensor(sensor_id):
    '''Return a fake reading for a sensor.'''
    return len(sensor_id) * 10

for sid in ("T1", "T22", "T1"):
    read_sensor(sid)

print(read_sensor.calls)
print(read_sensor.__name__, "-", read_sensor.__doc__)
```

Expected output:

```text
3
read_sensor - Return a fake reading for a sensor.
```

### **Task 2 — A Decorator with an Argument**

Write `in_range(low, high)`, a decorator **factory**. The decorated function's **first argument** must be within `[low, high]`; otherwise raise `ValueError` with the message `<function name>: <value> is outside [<low>, <high>]`.

```python
@in_range(-50, 60)
def to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(to_fahrenheit(30))
try:
    to_fahrenheit(75)
except ValueError as error:
    print(error)
```

Expected output:

```text
86.0
to_fahrenheit: 75 is outside [-50, 60]
```

### **Task 4 — A Registry with `__init_subclass__`**

Create a base class `Converter` that keeps a dictionary `Converter.registry`. Using `__init_subclass__(cls, unit, **kwargs)`:

* register each subclass under its `unit` keyword;
* raise `TypeError` if the subclass does not define a `convert(value)` method.

```python
class Kelvin(Converter, unit="K"):
    def convert(self, celsius):
        return round(celsius + 273.15, 2)

class Fahrenheit(Converter, unit="F"):
    def convert(self, celsius):
        return round(celsius * 9 / 5 + 32, 1)

print(sorted(Converter.registry))
print(Converter.registry["K"]().convert(25))
print(Converter.registry["F"]().convert(25))

try:
    class Broken(Converter, unit="X"):
        pass
except TypeError as error:
    print("TypeError:", error)
```

Expected output:

```text
['F', 'K']
298.15
77.0
TypeError: Broken must define convert()
```
