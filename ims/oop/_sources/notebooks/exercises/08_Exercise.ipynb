## **Laboratory Exercise: Money and Wallet**

### **Scenario**

Floating-point numbers alone are a poor way to represent money: they do not know their **currency**, and adding pesos to dollars should not be allowed. You will build a `Money` class and a `Wallet` that holds `Money` objects.

Expected output:

```text
Money(150.00, 'PHP') Money(20.00, 'USD')
₱150.00 USD 20.00
[Money(150.00, 'PHP')]
```

### **Task 2 — Comparison**

Implement `__eq__()` and `__lt__()`, and decorate the class with `functools.total_ordering`.

* Two `Money` objects are equal if **both** amount and currency are equal.
* Comparing different currencies with `<` must raise a `TypeError` with the message `Cannot compare PHP with USD`.
* Comparing with a non-`Money` object should return `NotImplemented`.

```python
print(Money(100) == Money(100), Money(100) == Money(100, "USD"))
print(Money(50) < Money(80), Money(80) >= Money(80))
print(sorted([Money(300), Money(45), Money(120)]))

try:
    Money(1) < Money(1, "USD")
except TypeError as error:
    print("TypeError:", error)
```

Expected output:

```text
True False
True True
[Money(45.00, 'PHP'), Money(120.00, 'PHP'), Money(300.00, 'PHP')]
TypeError: Cannot compare PHP with USD
```

### **Task 3 — Arithmetic**

Implement:

* `__add__()` — adds two `Money` objects of the **same** currency (raise `TypeError("Cannot add PHP and USD")` otherwise);
* `__radd__()` — so that `sum()` works (`sum` starts with `0`, so `0 + money` must return the money unchanged);
* `__mul__()` and `__rmul__()` — multiply by an `int` or `float`;
* `__bool__()` — a zero amount is `False`.

```python
print(Money(150) + Money(85.5))
print(Money(150) * 3, 2 * Money(19.99))
print(sum([Money(100), Money(250), Money(75)]))
print(bool(Money(0)), bool(Money(1)))

try:
    Money(10) + Money(10, "USD")
except TypeError as error:
    print("TypeError:", error)
```

Expected output:

```text
₱235.50
₱450.00 ₱39.98
₱425.00
False True
TypeError: Cannot add PHP and USD
```

### **Task 4 — A Container: `Wallet`**

Create a `Wallet` class that stores `Money` objects in a **private** list and supports:

| Syntax            | Behavior                                        |
| ----------------- | ----------------------------------------------- |
| `wallet.add(m)`   | adds a `Money` object                           |
| `len(wallet)`     | number of items                                 |
| `wallet[i]`       | the *i*-th item                                 |
| `for m in wallet` | loops over the items                            |
| `m in wallet`     | whether an equal `Money` object is in the wallet |
| `wallet.total()`  | the sum of all items (use `sum()`)              |

```python
wallet = Wallet()
for amount in (500, 20, 100):
    wallet.add(Money(amount))

print(len(wallet), wallet[0], wallet[-1])
print(Money(20) in wallet, Money(1000) in wallet)
print([str(m) for m in wallet])
print(wallet.total(), max(wallet))
```

Expected output:

```text
3 ₱500.00 ₱100.00
True False
['₱500.00', '₱20.00', '₱100.00']
₱620.00 ₱500.00
```
