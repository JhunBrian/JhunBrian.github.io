## **Laboratory Exercise: Testing a Shopping Cart**

### **Scenario**

A teammate wrote the `ShoppingCart` class below for the campus bookstore. The requirements are:

1. Adding an item stores its price and quantity; adding the **same item again increases its quantity**.
2. Prices must be **positive** and quantities **at least 1**; otherwise raise `ValueError`.
3. `remove(name)` removes an item; removing an item that is not in the cart raises `KeyError`.
4. `subtotal()` is the sum of price × quantity.
5. Orders with a subtotal of **₱1,000 or more** get a **10% discount**; `total()` applies it.

Your job is to **test** the class. It contains **at least one bug**.

Copy this code into your notebook **without changing it** (yet):

```python
class ShoppingCart:
    DISCOUNT_THRESHOLD = 1000
    DISCOUNT_RATE = 0.10

    def __init__(self):
        self._items = {}

    def add(self, name, price, quantity=1):
        if price <= 0:
            raise ValueError("Price must be positive.")
        if quantity < 1:
            raise ValueError("Quantity must be at least 1.")
        self._items[name] = (price, quantity)

    def remove(self, name):
        del self._items[name]

    def subtotal(self):
        return sum(price * qty for price, qty in self._items.values())

    def total(self):
        subtotal = self.subtotal()
        if subtotal > self.DISCOUNT_THRESHOLD:
            return round(subtotal * (1 - self.DISCOUNT_RATE), 2)
        return subtotal

    def __len__(self):
        return sum(qty for _, qty in self._items.values())
```

### **Task 1 — Normal Cases**

Create a `unittest.TestCase` subclass named `TestShoppingCart` with a `setUp()` method that creates an empty cart.

Write tests for:

* `test_subtotal` — a notebook (₱45 × 2) and a pen (₱12 × 3) give a subtotal of **126**;
* `test_len_counts_quantities` — the same cart has `len(cart) == 5`;
* `test_remove` — after removing the pen, the subtotal is **90**.

Run your tests inside the notebook with:

```python
import sys, unittest
suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestShoppingCart)
unittest.TextTestRunner(verbosity=2, stream=sys.stdout).run(suite)
```

### **Task 2 — Error Cases**

Add tests using `assertRaises`:

* `test_remove_missing_item` — removing an item that is not in the cart raises `KeyError`.

### **Task 3 — Edge Cases: Find the Bugs**

Write tests for the requirements that are most likely to hide mistakes:

* `test_adding_same_item_twice` — adding `("pen", 12, 2)` and then `("pen", 12, 3)` should give `len(cart) == 5`;
* `test_discount_exactly_at_threshold` — a single item of ₱1,000 should cost **900.0**;
* `test_no_discount_below_threshold` — a single item of ₱999 should cost **999**;
* `test_empty_cart` — an empty cart has a total of **0** and length **0**.

Run all your tests. Expected result: **two tests fail**. Record which ones in a Markdown cell and explain, for each, which requirement the code violates.

### **Task 4 — Fix the Bugs**

Fix **only** the lines responsible for the failures, then run the full test suite again. All tests should pass:

```text
Ran 10 tests in 0.00Xs

OK
```

> This is the core idea of **regression testing**: the tests that revealed the bugs stay in the suite, so the bugs cannot silently return.
