# **Packaging — Exercise**

**Learning Objective**

This exercise practices turning code into a **distributable project** with a `src` layout, `pyproject.toml` metadata, a README, and a version-controlled release.

## **Laboratory Exercise: The `tempconv` Package**

### **Scenario**

You will package a small temperature-conversion library so that classmates can install it with `pip`. The library has one module with a `Temperature` class.

This exercise is done **outside** a single notebook: create the files in a folder (VS Code, JupyterLab's file browser, or Python code), then use the notebook to record your commands and outputs.

### **Task 1 — Project Structure**

Create this structure:

```text
tempconv-project/
├── pyproject.toml
├── README.md
├── LICENSE
├── .gitignore
├── src/
│   └── tempconv/
│       ├── __init__.py
│       └── temperature.py
└── tests/
    └── test_temperature.py
```

`temperature.py` should contain a `Temperature` class with:

* a `celsius` property that rejects values below `-273.15` with a `ValueError`;
* read-only properties `fahrenheit` and `kelvin`;
* a `from_fahrenheit(f)` class method;
* a `__repr__` returning `Temperature(25.0)`.

`__init__.py` should export `Temperature` and define `__version__ = "0.1.0"`.

Print the tree of your project from the notebook (you may reuse the `tree()` function from the lesson).

### **Task 2 — `pyproject.toml`**

Write a `pyproject.toml` with:

* the `setuptools` build system;
* a **unique** name: `tempconv-<your-surname>`;
* version `0.1.0`, a description, `readme`, `requires-python = ">=3.9"`, a license, and your name as author;
* a `dev` optional-dependency group containing `pytest`;
* `[tool.setuptools.packages.find]` with `where = ["src"]`.

Verify it from the notebook:

```python
import tomllib

with open("tempconv-project/pyproject.toml", "rb") as f:
    meta = tomllib.load(f)["project"]

print(meta["name"], meta["version"], meta["requires-python"])
```

### **Task 3 — Install and Use**

In a **terminal** with your virtual environment activated, install the project in editable mode:

```bash
cd tempconv-project
pip install -e ".[dev]"
```

Then, in the notebook (restart the kernel first):

```python
from tempconv import Temperature
import tempconv

t = Temperature(25)
print(t, t.fahrenheit, t.kelvin)
print(Temperature.from_fahrenheit(212))
print(tempconv.__version__)
```

Expected output:

```text
Temperature(25.0) 77.0 298.15
Temperature(100.0)
0.1.0
```

### **Task 4 — README**

Write a `README.md` that includes: a one-paragraph description, installation instructions, a short usage example, how to run tests, and the license. Paste its contents into a Markdown cell of your notebook.

### **Task 5 — Version Control and a Release**

Initialize a Git repository, commit your project, and create the first release tag:

```bash
git init
git add .
git commit -m "Initial tempconv package"
git tag -a v0.1.0 -m "First release"
git log --oneline --decorate
```

Paste the output of `git log --oneline --decorate` into your notebook.

Then **add a feature** (for example, a `__eq__` method), bump the version to `0.2.0` in **both** `pyproject.toml` and `__init__.py`, update a `CHANGELOG.md`, commit, and tag `v0.2.0`.
