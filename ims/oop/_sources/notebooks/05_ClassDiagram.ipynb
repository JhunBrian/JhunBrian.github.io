A **class diagram** is a visual representation of the structure of an object-oriented system. It shows the classes that make up a program, the data and behaviors contained in those classes, and the relationships among them.

For this course, class diagrams are especially useful because the syllabus expects students to work with **classes and objects, abstraction, encapsulation, inheritance, polymorphism, and composition**. Later, during project planning, students are explicitly expected to perform **class-responsibility modeling and UML** before implementing their OOP project.  

A class diagram therefore acts as a **bridge between an idea and its Python implementation**:

```text
Real-world problem
       ↓
Identify objects
       ↓
Identify classes
       ↓
Define responsibilities
       ↓
Define relationships
       ↓
Draw class diagram
       ↓
Implement Python classes
```

## **From Objects to Classes**

Suppose we are developing a university enrollment system.

In the real world, we might encounter:

* students,
* instructors,
* courses,
* departments.

An individual student such as `"Ana Cruz"` is an **object**. The general description of what every student should contain and what every student can do is represented by a **class**.

```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

    def enroll(self, course):
        print(f"{self.name} enrolled in {course}")
```

We can create an object from the class:

```python
student1 = Student("Ana Cruz", "2026-001")
```

Conceptually:

```{mermaid}
flowchart LR
    A["Student Class<br/>Blueprint"] -->|instantiate| B["student1<br/>Ana Cruz"]
    A -->|instantiate| C["student2<br/>Juan Santos"]
    A -->|instantiate| D["student3<br/>Maria Reyes"]
```

A **class diagram describes the blueprint**, rather than listing every object created from it.

## **Anatomy of a UML Class**

In the **Unified Modeling Language (UML)**, a class is normally represented as a rectangle divided into three sections:

```{mermaid}
classDiagram
class Student {
    +String name
    +String student_id
    +enroll(course)
    +display_info()
}
```

The three conceptual sections are:

```text
┌────────────────────────────┐
│          Student           │  ← Class name
├────────────────────────────┤
│ + name: String             │  ← Attributes
│ + student_id: String       │
├────────────────────────────┤
│ + enroll(course)           │  ← Methods
│ + display_info()           │
└────────────────────────────┘
```

### **Class Name**

The first section contains the name of the class.

Examples include `Student`, `Course`, `Dataset`, `BankAccount`, and `DataProcessor`.

Class names normally describe a **type of thing or concept**, which is why nouns are commonly used.

### **Attributes**

Attributes describe the **state or information maintained by an object**.

For a `Student`, these might include:

```text
name
student_id
year_level
```

In Python:

```python
self.name = name
self.student_id = student_id
self.year_level = year_level
```

### **Methods**

Methods describe the **behavior or responsibilities** of objects belonging to the class.

For example:

```text
enroll()
drop_course()
display_info()
```

In Python:

```python
def enroll(self, course):
    ...

def drop_course(self, course):
    ...

def display_info(self):
    ...
```

Thus, a useful way to remember a class diagram is:

> **Class = Identity + State + Behavior**

## **Attributes and Methods in UML**

UML commonly writes attributes using:

```text
visibility attribute_name : data_type
```

For example:

```text
+ name : String
+ age : int
- balance : float
```

Methods can be written as:

```text
visibility method_name(parameters) : return_type
```

For example:

```text
+ deposit(amount: float) : void
+ get_balance() : float
```

Consider this diagram:

```{mermaid}
classDiagram
class BankAccount {
    -String account_number
    -float balance
    +deposit(amount)
    +withdraw(amount)
    +get_balance()
}
```

A corresponding Python implementation could be:

```python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

    def get_balance(self):
        return self.balance
```

Notice that a class diagram does **not** attempt to reproduce the entire Python program. It communicates the **design and structure** of the class.

## **Visibility**

UML uses symbols to describe the intended accessibility of members.

| Symbol | Visibility | General meaning                       |
| ------ | ---------- | ------------------------------------- |
| `+`    | Public     | Accessible outside the class          |
| `-`    | Private    | Intended for use inside the class     |
| `#`    | Protected  | Intended for the class and subclasses |

For example:

```{mermaid}
classDiagram
class BankAccount {
    -float balance
    #String account_type
    +deposit(amount)
    +withdraw(amount)
}
```

Python does not enforce access modifiers in exactly the same manner as languages such as Java or C++. Instead, Python commonly uses naming conventions and properties.

For example:

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance
```

This connects class diagrams naturally with **encapsulation and information hiding**, which are explicit topics in the course syllabus. 

## **Relationships Between Classes**

The real power of class diagrams becomes apparent when several classes interact.

Consider a simple university system:

```{mermaid}
classDiagram
class Student {
    +String name
    +String student_id
    +enroll()
}

class Course {
    +String code
    +String title
    +add_student()
}
```

Knowing the individual classes is useful, but an important question remains:

> **How are `Student` and `Course` related?**

Class diagrams allow us to represent these relationships explicitly.

Some of the most important relationships in OOP are:

* association,
* inheritance,
* aggregation,
* composition,
* dependency.

### **Association — "uses / knows about"**

An **association** indicates that objects from two classes are related to or interact with one another.

For example:

> A student enrolls in a course.

```{mermaid}
classDiagram
Student --> Course : enrolls in

class Student {
    +String name
    +enroll(course)
}

class Course {
    +String code
    +String title
}
```

In Python, the relationship could appear as:

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.courses = []

    def enroll(self, course):
        self.courses.append(course)
```

The `Student` object now knows about `Course` objects.

### **Multiplicity**

Sometimes the number of participating objects matters.

Suppose:

* one course may contain many students;
* one student may enroll in many courses.

We can express this with **multiplicity**.

```{mermaid}
classDiagram
Student "0..*" -- "0..*" Course : enrolls in
```

Common multiplicities include:

| Notation | Meaning      |
| -------- | ------------ |
| `1`      | Exactly one  |
| `0..1`   | Zero or one  |
| `*`      | Many         |
| `0..*`   | Zero or many |
| `1..*`   | One or many  |

Multiplicity forces us to think more carefully about our data model.

For example:

```text
Student 1 -------- 1 IDCard
```

is fundamentally different from:

```text
Student * -------- * Course
```

### **Inheritance — "is-a"**

Inheritance represents an **is-a relationship**.

Consider:

> A `Student` is a `Person`.
> An `Instructor` is a `Person`.

The diagram becomes:

```{mermaid}
classDiagram
Person <|-- Student
Person <|-- Instructor

class Person {
    +String name
    +String email
    +display_info()
}

class Student {
    +String student_id
    +enroll()
}

class Instructor {
    +String employee_id
    +teach()
}
```

In Python:

```python
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email


class Student(Person):
    def __init__(self, name, email, student_id):
        super().__init__(name, email)
        self.student_id = student_id


class Instructor(Person):
    def __init__(self, name, email, employee_id):
        super().__init__(name, email)
        self.employee_id = employee_id
```

The arrow points toward the **parent/base class**.

```text
           Person
          /      \
         /        \
    Student     Instructor
```

This representation becomes particularly useful when designing inheritance hierarchies before writing them in Python. The course specifically covers base and derived classes, method overriding, `super()`, multiple inheritance, and inheritance design considerations. 

### **Composition — "has-a"**

Not every relationship should be represented through inheritance.

Consider a `Car` and an `Engine`.

A car is **not** an engine.

Instead:

> A car **has an** engine.

This is a **has-a relationship**, which can be modeled using composition.

```{mermaid}
classDiagram
Car *-- Engine

class Car {
    -Engine engine
    +start()
}

class Engine {
    +start()
    +stop()
}
```

A possible implementation is:

```python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()
```

The `Car` is built using another object.

This leads to an important design question:

```text
Is X a Y?  → Consider inheritance

Does X have/use a Y? → Consider composition/association
```

The distinction is important because the syllabus specifically asks students to examine **composition versus inheritance** when studying polymorphism and flexible OOP designs. 

### **Aggregation vs. Composition**

Aggregation and composition both represent **whole–part relationships**, but they differ in ownership.

#### **Aggregation**

Aggregation is a weaker relationship.

> A `Department` has `Instructor`s, but an instructor can exist independently of a particular department.

```{mermaid}
classDiagram
Department o-- Instructor : has

class Department {
    +String name
}

class Instructor {
    +String name
}
```

The hollow diamond represents aggregation.

#### **Composition**

Composition represents stronger ownership.

For example, imagine an application where a `Survey` creates and owns its `Question` objects:

```{mermaid}
classDiagram
Survey *-- Question : contains

class Survey {
    +String title
    +add_question()
}

class Question {
    +String text
}
```

The filled diamond represents composition.

Conceptually:

```text
Association
    ↓
objects are related

Aggregation
    ↓
weak whole–part relationship

Composition
    ↓
strong whole–part ownership
```

### **Dependency — "temporarily uses"**

A **dependency** means that one class temporarily relies on another without necessarily storing it as part of its permanent state.

For example, suppose `ReportGenerator` receives a `Dataset` only when generating a report:

```python
class ReportGenerator:
    def generate(self, dataset):
        print("Generating report...")
```

This can be represented as:

```{mermaid}
classDiagram
ReportGenerator ..> Dataset : uses

class ReportGenerator {
    +generate(dataset)
}

class Dataset {
    +load()
    +clean()
}
```

The dashed arrow communicates that `ReportGenerator` **depends on** `Dataset`.

## **Reading a Complete Class Diagram**

Consider a simplified learning-management system:

```{mermaid}
classDiagram
Person <|-- Student
Person <|-- Instructor

Course "0..*" -- "0..*" Student : enrolls
Instructor "1" -- "0..*" Course : teaches
Course *-- Assignment : contains

class Person {
    +String name
    +String email
    +display_info()
}

class Student {
    +String student_id
    +enroll(course)
    +submit(assignment)
}

class Instructor {
    +String employee_id
    +create_assignment()
}

class Course {
    +String code
    +String title
    +add_student()
}

class Assignment {
    +String title
    +Date deadline
    +submit()
}
```

Instead of seeing this as a collection of boxes and arrows, read it as a description of the system:

**Inheritance:** `Student` and `Instructor` are specialized forms of `Person`.

**Association:** students enroll in courses, while instructors teach courses.

**Multiplicity:** multiple students can participate in multiple courses.

**Composition:** a course contains assignments in this particular design.

The diagram therefore communicates considerably more information than a simple list of classes.

## **Class Diagrams Before Coding**

A common mistake is to immediately begin writing classes:

```python
class Student:
    ...
```

without first deciding what each class is responsible for.

For small exercises this may work. For larger applications, it can result in classes with unclear responsibilities and unnecessary dependencies.

A better workflow is:

```{mermaid}
flowchart LR
    A["Problem"] --> B["Requirements"]
    B --> C["Identify Objects"]
    C --> D["Identify Classes"]
    D --> E["Assign Responsibilities"]
    E --> F["Define Relationships"]
    F --> G["Class Diagram"]
    G --> H["Python Implementation"]
    H --> I["Test & Refine"]
```

This aligns particularly well with the final project portion of DS212. The syllabus progresses from **problem definition and requirements** to **class-responsibility modeling and UML**, then to project structure, implementation, class collaboration, exception handling, testing, debugging, and version control.  

## **Example: Designing a Data Science Application**

Because DS212 is an OOP course for Data Science students, consider a small dataset-processing application.

### **Requirements**

The application should:

1. load a dataset;
2. clean the dataset;
3. analyze it;
4. generate a report.

We could identify the following responsibilities:

```text
Dataset
    stores and loads data

DataCleaner
    performs cleaning operations

Analyzer
    performs analysis

Report
    represents analysis results
```

A preliminary design might be:

```{mermaid}
classDiagram
Dataset --> DataCleaner : processed by
Analyzer --> Dataset : analyzes
Analyzer --> Report : creates

class Dataset {
    -String filepath
    -data
    +load()
    +get_data()
}

class DataCleaner {
    +remove_missing()
    +remove_duplicates()
}

class Analyzer {
    +calculate_summary()
    +analyze(dataset)
}

class Report {
    -results
    +display()
    +save()
}
```

Notice that we have not written the implementation yet.

The diagram allows us to ask important questions first:

* Should `Dataset` clean itself?
* Should cleaning be handled by `DataCleaner`?
* Should `Analyzer` create the report?
* Should `Report` know about the original dataset?
* Which attributes should be hidden?
* Which objects should own other objects?

These are **design questions**, not merely syntax questions.


## **Class Diagrams Are Models, Not Code**

A class diagram should not be treated as a perfect translation of source code.

Its purpose is to **communicate a design**.

Therefore, you do not necessarily need to include every:

* variable,
* helper method,
* implementation detail,
* temporary value,
* imported function.

Suppose a class contains fifteen internal helper methods. A high-level design diagram might show only the four methods important to understanding how the system works.

The appropriate amount of detail depends on the purpose of the diagram.

> **A useful class diagram communicates the important structure of the system without becoming a graphical copy of the source code.**

This becomes particularly relevant to the course project because the syllabus assesses whether class responsibilities and relationships are organized and understandable, and whether the resulting application is maintainable and reproducible. 

## **Common Design Mistakes**

### **Mistake 1: Treating every noun as a class**

Finding nouns in requirements is a useful starting technique, but not every noun deserves its own class.

```text
"The student enters a name."

Possible concepts:
Student
Name
```

Usually, `Student` should be a class while `name` should simply be an attribute.

```{mermaid}
classDiagram
class Student {
    +String name
}
```

### **Mistake 2: Confusing "is-a" and "has-a"**

This would be questionable:

```{mermaid}
classDiagram
Engine <|-- Car
```

because it means:

> Car **is an** Engine.

Instead:

```{mermaid}
classDiagram
Car *-- Engine
```

which expresses:

> Car **has an** Engine.

### **Mistake 3: Creating one giant class**

```text
UniversitySystem
 ├── students
 ├── courses
 ├── instructors
 ├── enrollment
 ├── grades
 ├── reports
 ├── authentication
 └── everything else
```

OOP encourages assigning **cohesive responsibilities** to appropriate classes.

A better design might distribute them:

```{mermaid}
classDiagram
UniversitySystem --> Student
UniversitySystem --> Course
UniversitySystem --> Enrollment
Course --> Instructor
Enrollment --> Student
Enrollment --> Course
```
