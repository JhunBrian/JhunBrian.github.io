
| Week | Topic                                 | Lesson page                                   | You should be able to…                                                      |
| ---- | ------------------------------------- | --------------------------------------------- | --------------------------------------------------------------------------- |
| 1    | Paradigms, history, principles        | Programming paradigms; History & principles   | compare paradigms; explain why OOP emerged; explain OOP's role in data science |
| 2    | Python review I                       | Python review I                               | use variables, data types, operators, f-strings, and the notebook workflow  |
| 3    | Python review II                      | Python review II                              | write conditionals, loops, and functions; import modules                    |
| 4    | Objects and classes; class diagrams   | Objects and classes; UML class diagram        | define classes, constructors, attributes, and methods; read and draw UML    |
| 5    | Abstraction and encapsulation         | Four core concepts                            | design interfaces and ABCs; use `_`, `__`, and `@property` with validation  |
| 6    | Inheritance and overriding            | Four core concepts                            | build hierarchies; override methods; use `super()`; read the MRO            |
| 7    | Polymorphism and composition          | Four core concepts                            | use duck typing; choose composition vs. inheritance                         |
| 8    | Packages and real-world design        | Packages and real-world OOP design            | organize modules and packages; assign responsibilities with CRC cards       |

## **Key Terms**

Make sure you can define each term **and** give a short Python example.

| | | |
| --- | --- | --- |
| programming paradigm | class | object / instance |
| constructor (`__init__`) | `self` | instance vs. class attribute |
| method (command / query) | UML visibility (`+`, `-`, `#`) | association / multiplicity |
| aggregation vs. composition | dependency | abstraction / abstract base class |
| encapsulation / `@property` | name mangling | inheritance / `super()` |
| method overriding | method resolution order | polymorphism / duck typing |
| module / package / namespace | CRC card | cohesion / coupling |

## **Self-Check Questions**

### **Concepts**

1. Explain the difference between the procedural and object-oriented approaches using the same small problem.
2. What contributions did Simula and Smalltalk make to OOP?
3. What is the difference between `==` and `is`? Give an example where they produce different results.
4. Why is it a mistake to create a list as a **class attribute** when each object needs its own list?
5. When should you use an abstract base class instead of a regular parent class?
6. How do `_balance` and `__balance` differ in Python?
7. Explain the *is-a* / *has-a* test and give one example of each.
8. Why does duck typing not require inheritance?
9. What does `if __name__ == "__main__":` prevent?

### **Reading code**

10. What is printed, and why?

    ```python
    class Counter:
        total = 0

        def __init__(self):
            self.count = 0

        def tick(self):
            self.count += 1
            Counter.total += 1

    a, b = Counter(), Counter()
    a.tick(); a.tick(); b.tick()
    print(a.count, b.count, Counter.total)
    ```

11. Draw the UML class diagram for this code, including visibility and the relationship:

    ```python
    class Engine:
        def start(self):
            ...

    class Car:
        def __init__(self):
            self._engine = Engine()

        def drive(self):
            self._engine.start()
    ```

12. Which `describe()` runs for `TA().describe()`, and what is `TA.__mro__`?

    ```python
    class Person:
        def describe(self): return "person"

    class Student(Person):
        def describe(self): return "student"

    class Employee(Person):
        def describe(self): return "employee"

    class TA(Employee, Student):
        pass
    ```

### **Writing code**

13. Write a `Temperature` class whose `celsius` property rejects values below −273.15 and which provides a read-only `fahrenheit` property.
14. Create an abstract class `Shape` with an abstract `area()` method, and two subclasses. Compute the total area of a list of shapes with a single loop.
15. Write CRC cards for a **parking lot** system with at least four classes, then draw its class diagram.

## **Study Tips**

* **Re-run** the code in each lesson page and change it — predict the output before you run it.
* **Redo** the exercises without looking at your previous answers.
* **Draw** class diagrams by hand; the examination may ask you to read or produce them.
* **Explain** each concept aloud in your own words, with an example from data science.
* Review the **assessment criteria** of your activities to see what "good design" means in this course.

## **References**

Beck, K., & Cunningham, W. (1989). A laboratory for teaching object-oriented thinking. *ACM SIGPLAN Notices, 24*(10), 1–6. https://doi.org/10.1145/74878.74879

Downey, A. B. (2024). *Think Python: How to think like a computer scientist* (3rd ed.). O'Reilly Media. https://allendowney.github.io/ThinkPython/

Fowler, M. (2003). *UML distilled: A brief guide to the standard object modeling language* (3rd ed.). Addison-Wesley.

Liang, Y. D. (2013). *Introduction to programming using Python*. Pearson.

Python Software Foundation. (2026). *The Python tutorial: 9. Classes*. https://docs.python.org/3/tutorial/classes.html
